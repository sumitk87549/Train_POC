## 🚀 Step 3: Load the Multilingual Model

Load the Chatterbox Multilingual model that supports Hindi.

```python
import torch
import torchaudio as ta
from chatterbox.mtl_tts import ChatterboxMultilingualTTS
from IPython.display import Audio, display
import os

# Check if GPU is available
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load the multilingual model
print("Loading Chatterbox Multilingual model...")
model = ChatterboxMultilingualTTS.from_pretrained(device=device)
print("✅ Model loaded successfully!")
print(f"Sample rate: {model.sr} Hz")
```

## 🗣️ Step 4: Test Hindi TTS with Default Voice

Generate speech from Hindi text using the default voice.

```python
# Hindi text for testing
hindi_text = "नमस्ते, आज का दिन बहुत अच्छा है। मुझे उम्मीद है कि आप सभी स्वस्थ और खुश हैं।"

# Generate speech
print(f"Generating speech for: {hindi_text}")
wav = model.generate(hindi_text, language_id="hi")

# Save to file
output_file = "hindi_tts_default.wav"
ta.save(output_file, wav, model.sr)
print(f"✅ Audio saved to {output_file}")

# Play the audio
display(Audio(wav.numpy(), rate=model.sr))
```

## 🎭 Step 5: Test Multiple Hindi Texts

Try different Hindi sentences to see how the model performs.

```python
# Multiple Hindi texts for testing
hindi_samples = [
    "मुझे कृत्रिम बुद्धिमत्ता के बारे में सीखना पसंद है।",
    "आज मौसम बहुत सुहावना है।",
    "क्या आप मेरी मदद कर सकते हैं?",
    "मैं भारत से हूं और मुझे हिंदी बोलना पसंद है।",
    "यह तकनीक बहुत उन्नत है।"
]

# Create output directory if it doesn't exist
os.makedirs("hindi_outputs", exist_ok=True)

# Generate speech for each sample
for i, text in enumerate(hindi_samples, 1):
    print(f"\n[{i}/{len(hindi_samples)}] Generating: {text}")
    
    wav = model.generate(text, language_id="hi")
    output_file = f"hindi_outputs/sample_{i}.wav"
    ta.save(output_file, wav, model.sr)
    
    print(f"✅ Saved to {output_file}")
    print("🔊 Playing audio:")
    display(Audio(wav.numpy(), rate=model.sr))
```

## 📊 Step 8: Batch Processing

Process multiple Hindi texts in batch.

```python
# Add your custom Hindi texts here
batch_texts = [
    "पहला वाक्य",
    "दूसरा वाक्य",
    "तीसरा वाक्य"
]

# Create batch output directory
batch_dir = "batch_outputs"
os.makedirs(batch_dir, exist_ok=True)

print(f"Processing {len(batch_texts)} texts...\n")

for idx, text in enumerate(batch_texts, 1):
    print(f"[{idx}/{len(batch_texts)}] {text}")
    
    wav = model.generate(text, language_id="hi")
    filename = f"{batch_dir}/batch_{idx:03d}.wav"
    ta.save(filename, wav, model.sr)
    
    print(f"  ✅ Saved: {filename}")
    display(Audio(wav.numpy(), rate=model.sr))
    print()

print(f"\n✅ All {len(batch_texts)} files processed!")
```

## 📝 Notes and Tips

**Supported Languages:**
- Hindi (hi), English (en), Spanish (es), French (fr), German (de), and 18+ more!

**Voice Cloning:**
- Upload a 3-10 second audio sample in WAV format
- The model will mimic the voice characteristics
- Works best with clear, noise-free recordings

**Emotion Control:**
- Add tags like `[laugh]`, `[sigh]`, `[cough]` in your text
- Example: "नमस्ते [laugh] क्या आप ठीक हैं?"

**Performance:**
- GPU recommended for faster generation
- First generation takes longer (model loading)
- Subsequent generations are much faster
