## 3. Load the Svara TTS Model

Load the model and create a TTS pipeline.

```python
# Model configuration
MODEL_NAME = "kenpath/svara-tts-v1"

# Determine device
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load the TTS pipeline
print("Loading Svara TTS model...")
tts_pipeline = pipeline(
    "text-to-speech",
    model=MODEL_NAME,
    device=device,
    token=HF_TOKEN
)

print("✓ Model loaded successfully!")
```

## 4. Helper Functions

Functions to generate and play audio.

```python
def generate_speech(text, speaker_id=None, save_path=None):
    """
    Generate speech from text using Svara TTS.
    
    Args:
        text (str): The Hindi text to convert to speech
        speaker_id (int, optional): Speaker/voice ID if model supports multiple speakers
        save_path (str, optional): Path to save the audio file
    
    Returns:
        Audio object for playback
    """
    print(f"Generating speech for: {text[:50]}...")
    
    # Generate speech
    if speaker_id is not None:
        output = tts_pipeline(text, speaker_id=speaker_id)
    else:
        output = tts_pipeline(text)
    
    # Extract audio data
    audio = output["audio"]
    sampling_rate = output["sampling_rate"]
    
    # Save if path provided
    if save_path:
        sf.write(save_path, audio, sampling_rate)
        print(f"✓ Audio saved to: {save_path}")
    
    # Return audio for playback
    return Audio(audio, rate=sampling_rate, autoplay=False)

print("✓ Helper functions defined")
```

## Tips for Experimentation:

1. **Text Variations**: Try different sentence structures, lengths, and content types
2. **Prosodic Markers**: Use punctuation (. , ! ?) to control pauses and intonation
3. **Voice Testing**: If multiple speakers are supported, compare their characteristics
4. **Special Characters**: Test how the model handles numbers, English words mixed with Hindi, etc.
5. **Emotion**: Try expressive text with exclamations, questions, and varied tone

**Common Prosodic Markers to Test:**
- `.` - Full stop (longer pause)
- `,` - Comma (short pause)
- `!` - Exclamation (emphasis/excitement)
- `?` - Question (rising intonation)
- `...` - Ellipsis (trailing off)
