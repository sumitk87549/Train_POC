````python
import re, json, os
from collections import Counter
from pathlib import Path

# ═══════════════════════════════════════════════════════════════════════
# STORY STATE ENGINE v6.0  — JARVIS Hindi→Hinglish v1.0
# Covers: Hindi(Latin) source books · All genres · Hinglish-first voice
# Tone target: casual OTT-India register (Made in Heaven / TVF / Little Things)
#               -  respectful, zero foul language, urban Indian millennial/GenZ
# ═══════════════════════════════════════════════════════════════════════════════
# STORY_STATE = {
#     'book_title':            '',
#     'genre':                 '',
#     'author':                '',
#     'author_dna':            '',       # v5.0: author tone fingerprint
#     'tone_anchor':           '',       # v5.0: locked register for entire book
#     'characters':            {},
#     'character_speech_styles': {},
#     'current_setting':       '',
#     'dominant_tone':         'DAILY_LIFE',
#     'established_vocab':     {},
#     'story_so_far':          [],       # v5.0: rolling 3-entry summary
#     'chunk_count':           0,
#     'total_chunks':          0,
#     'resume_from_chunk':     0,        # v5.0: granular resume
# }

STORY_STATE = {
    # --- BOOK IDENTITY -------------------------------------------------------
    'book_title': '',
    'author_key': '',
    'genre':      '',
    'author':     '',

    # --- AUTHOR VOICE --------------------------------------------------------
    'author_dna': '',
    'tone_anchor': '',

    # --- CHARACTERS ----------------------------------------------------------
    'characters': {},

    # --- CHARACTER SPEECH STYLES ---------------------------------------------
    'character_speech_styles': {},

    # --- OPENING SETTING -----------------------------------------------------
    'current_setting': '',

    # --- DOMINANT TONE -------------------------------------------------------
    'dominant_tone': 'DAILY_LIFE',

    # --- ESTABLISHED VOCAB ---------------------------------------------------
    # Hindi-formal -> Hinglish-casual mappings
    'established_vocab': {},

    # --- ROLLING CONTEXT SUMMARY ---------------------------------------------
    'story_so_far': [],

    # --- CHUNK TRACKING ------------------------------------------------------
    'chunk_count':        0,
    'total_chunks':       0,
    'resume_from_chunk':  0,
}

AUTHOR_DNA = {

    # ── FRANZ KAFKA ───────────────────────────────────────────────────────────
    'kafka': (
        'flat, bureaucratic, deadpan absurdist',
        ('Franz Kafka style: Impossible things described in government-notice language. '
         'No shock, no drama  -  just cold matter-of-fact observation. '
         'The horror is in the FLATNESS. Characters accept the unacceptable as routine. '
         'Hinglish register: sarkari notice ki tarah seedha, thaka hua, koi emotion nahi. '
         'Gregor ke insect hone ki baat waise batao jaise bijli ka bill aaya ho. '
         'Never exclaim. Never over-explain. Let the absurd sit quietly.'),
        'literary fiction',
        {'transformation': 'tabdeeli', 'vermin': 'keeda-makoda', 'chief clerk': 'chief clerk sahab',
         'office': 'office', 'debt': 'karza', 'father': 'Papa', 'mother': 'Maa',
         'sister': 'behen', 'the firm': 'company', 'superintendent': 'suprintendent sahab'}
    ),

    # ── JANE AUSTEN ───────────────────────────────────────────────────────────
    'austen': (
        'witty, dry-ironic, socially sharp',
        ('Jane Austen style: Gentle irony about marriage, class, money, pretension. '
         'The narrator knows everything and smiles quietly at everyone\'s drama. '
         'Dialogue reveals character instantly. Formal settings, sharp social observations. '
         'Hinglish: wit preserved  -  raised-eyebrow prose. '
         'Log kya sochenge wali anxiety, rishte ki politics, paisa aur izzat ka khel. '
         'Never sarcastic  -  knowing. "It is a truth universally acknowledged" energy. '
         'Urban Indian aunties aur beta-beti ka tension bilkul naturally fit hota hai.'),
        'romance/social',
        {'estate': 'haveli', 'fortune': 'daulat', 'society': 'samaj ke log',
         'governess': 'ghar ki padhai karwane wali', 'the ball': 'dance party / mehfil',
         'proposal': 'rishta', 'marriage': 'shaadi', 'reputation': 'izzat',
         'the neighbourhood': 'aas-paas ke log', 'entail': 'property ka chakkar',
         'dowry': 'dahej', 'accomplishments': 'gun', 'sensibility': 'dil ka mamla',
         'sense': 'akal ka mamla', 'pride': 'ahamkaar', 'prejudice': 'poori soch'}
    ),

    # ── ARTHUR CONAN DOYLE ────────────────────────────────────────────────────
    'doyle': (
        'precise, analytical, Victorian-confident',
        ('Arthur Conan Doyle (Sherlock Holmes) style: Holmes = absolute analytical confidence, '
         'never casual, never wrong. Watson = warm, admiring narrator  -  the audience\'s eyes. '
         'Victorian London fog, gas lamps, Baker Street, telegram, hansom cab. '
         'Deduction is methodical  -  one observation chains to the next like a proof. '
         'Hinglish: Holmes bolte hain toh lagta hai case solve ho gaya  -  precise, composed. '
         'Watson thoda warm, thoda amazed. Inspector Scotland Yard ka always slightly confused. '
         'Holmes kabhi bhi casual nahi  -  even Hinglish mein ek composed professional lagta hai. '
         '"Elementary, my dear Watson" energy  -  seedha, sharp, thoda detached.'),
        'mystery/detective',
        {'deduction': 'deduction', 'elementary': 'bilkul seedhi baat hai',
         'the game is afoot': 'khel shuru ho gaya', 'inspector': 'Inspector sahab',
         'clue': 'suraag', 'the case': 'yeh case', 'hansom cab': 'buggy',
         'telegram': 'taar', 'revolver': 'revolver', 'Baker Street': 'Baker Street',
         'cocaine': 'nasha', 'violin': 'violin', 'disguise': 'bhes badal ke',
         'Moriarty': 'Moriarty', 'Scotland Yard': 'Scotland Yard'}
    ),

    # ── AGATHA CHRISTIE ───────────────────────────────────────────────────────
    'christie': (
        'clever, cozy-suspenseful, twist-driven',
        ('Agatha Christie style: Cozy British surface  -  country houses, tea, polite suspects. '
         'Underneath: cold logic and one guaranteed twist. Everyone has a secret. '
         'Poirot uses "little grey cells"  -  methodical, slightly theatrical. '
         'Miss Marple = sharp village aunty who sees through everyone. '
         'Hinglish: surface pe sab normal, andar sab guilty lag rahe. '
         'Poirot thoda formal, thoda showy  -  "mere chote bhoore cells keh rahe hain..." '
         'Miss Marple = gaon ki samajhdaar bua  -  seedha nahi bolti par sab jaanti hai. '
         'Reader ko lagta rahe "yeh toh nahi kar sakta"  -  phir wahi karta hai.'),
        'mystery/detective',
        {'Hercule Poirot': 'Poirot sahab', 'Miss Marple': 'Miss Marple',
         'little grey cells': 'mere chote bhoore cells', 'the suspect': 'shak wala banda',
         'country house': 'badi kothi', 'vicar': 'padri sahab', 'drawing room': 'baithak',
         'poison': 'zeher', 'alibi': 'alibi', 'Scotland Yard': 'Scotland Yard',
         'butler': 'khaansamaan', 'the victim': 'maara gaya banda', 'will': 'wasiyat'}
    ),

    # ── WILLIAM SHAKESPEARE ───────────────────────────────────────────────────
    'shakespeare': (
        'elevated, poetic, emotionally grand',
        ('William Shakespeare style: Grand poetic language carrying enormous emotions. '
         'Tragedy = power, jealousy, ambition, love destroying themselves. '
         'Comedy = mistaken identity, romantic confusion, all resolved in marriage. '
         'Soliloquies = raw honest thinking aloud  -  the character and audience alone. '
         'Hinglish: grandeur banaye rakho  -  iambic pentameter ka flow prose mein aaaye. '
         '"To be or not to be" ka weight wahi rakho  -  yeh sirf question nahi, existence ka sankat hai. '
         'Love scenes mein poetry rakho. Tragic scenes mein weight. '
         'Thoda elevated register  -  yeh drawing room conversation nahi, stage par rakha gaya life hai. '
         'Villains (Iago, Macbeth, Richard) mein charisma rakho  -  woh bure hain par compelling hain.'),
        'literary fiction',
        {'soliloquy': 'mann ki baat', 'villain': 'khalnaayak', 'tragedy': 'tragedy',
         'ambition': 'hawas', 'jealousy': 'jealousy', 'betrayal': 'dhoka',
         'honour': 'izzat', 'the court': 'darbaar', 'the king': 'Maharaj',
         'the queen': 'Rani sahiba', 'the prince': 'Rajkumar', 'ghost': 'aatma',
         'witches': 'daayan', 'prophecy': 'bhavisyavaani', 'fate': 'naseeb',
         'love': 'mohabbat', 'death': 'maut', 'dagger': 'khanjar'}
    ),

    # ── CHARLES DICKENS ───────────────────────────────────────────────────────
    'dickens': (
        'vivid, humanist, socially furious',
        ('Charles Dickens style: Rich vivid characters with memorable physical quirks. '
         'Social anger at poverty, exploitation, hypocrisy  -  but never loses warmth. '
         'Humor and pathos sitting right next to each other. '
         'Minor characters are as vivid as major ones. '
         'Victorian London  -  workhouses, fog, debtor\'s prison, court, thieves\' dens. '
         'Hinglish: characters ki description mein vivid rakho  -  jaise seedha aankhon ke saamne ho. '
         'Garib bachon ki taklif straight to the heart  -  no filter. '
         'Comic villains (Fagin, Uriah Heep) mein humor aur creepiness dono. '
         'Emotional peaks pe hold karo  -  Dickens ka pathos genuine hai, melodrama nahi.'),
        'literary fiction',
        {'workhouse': 'kamaai ka adda (gareebon ka)', 'pickpocket': 'jeb kata',
         'debtor\'s prison': 'karz ki jail', 'fog': 'kohra', 'solicitor': 'vakeel sahab',
         'orphan': 'yateem', 'the parish': 'mohalla', 'gentleman': 'sharifzaada',
         'factory': 'factory', 'coal': 'koyla', 'Christmas': 'Christmas',
         'Scrooge': 'Scrooge', 'the ghost': 'woh aatma'}
    ),

    # ── EDGAR ALLAN POE ───────────────────────────────────────────────────────
    'poe': (
        'gothic, psychological dread, obsessive',
        ('Edgar Allan Poe style: First-person unreliable narrators descending into madness. '
         'Psychological horror over jump scares  -  the dread is INSIDE the mind. '
         'Obsessive, overheated prose  -  the narrator protests too much. '
         'Gothic settings: crumbling mansions, crypts, isolation. '
         'The reader knows something is wrong before the narrator admits it. '
         'Hinglish: andar se darne wali feeling  -  bahar kuch nahi, andar sab bikharna. '
         'Narrator thoda over-explain karta hai  -  "I am NOT mad" = woh mad hai. '
         'Prose ki rhythm mein tension build karo  -  short sentences jab horror peak ho. '
         'Koi bhi cheez normal nahi lagni chahiye  -  sab slightly off.'),
        'gothic/horror',
        {'the vault': 'tahe khana', 'the pendulum': 'jhoolti talwaar',
         'the raven': 'woh kauwa', 'the tell-tale heart': 'woh dhadakta dil',
         'nevermore': 'kabhi nahi', 'the pit': 'khai', 'madness': 'pagalpan',
         'the mansion': 'woh haveli', 'the crypt': 'qabristan ke neeche',
         'the beating': 'dhak dhak', 'premature burial': 'zinda dafn'}
    ),

    # ── OSCAR WILDE ───────────────────────────────────────────────────────────
    'wilde': (
        'witty, epigram-sharp, aesthetically playful',
        ('Oscar Wilde style: Every line wants to be quotable  -  epigrams flying everywhere. '
         'Wit is the highest virtue. Seriousness is treated as slightly embarrassing. '
         'Beautiful surfaces hiding moral emptiness  -  or moral depth hiding behind beauty. '
         'Comedy of manners: aristocrats being delightfully absurd. '
         'Hinglish: Wilde ki wit ka flow pakdna bahut zaruri hai  -  '
         '"I can resist everything except temptation" energy. '
         'Hinglish mein epigrams kaise kaate? Seedha, pointed, thoda self-aware. '
         'Upper class ki bakwaas pe amused narrator  -  never angry, always wry. '
         'Dorian Gray mein: khubsurti aur corruption ka dance  -  dono glamorous lagte hain.'),
        'literary fiction',
        {'epigram': 'ek seedhi cutting baat', 'society': 'samajhdaar log (irony ke saath)',
         'aesthete': 'sundarata ka deewaana', 'portrait': 'portrait', 'sin': 'gunaah',
         'corruption': 'bigadna', 'wit': 'zubaan ki tezi', 'Lady': 'Lady sahiba',
         'Lord': 'Lord sahab', 'drawing room': 'baithak', 'scandal': 'scandal',
         'the ideal': 'ideal cheez', 'the earnest man': 'seedha sacha insaan (irony ke saath)'}
    ),

    # ── MARK TWAIN ────────────────────────────────────────────────────────────
    'twain': (
        'satirical, vernacular-warm, morally clear',
        ('Mark Twain style: American vernacular that sounds like someone is talking to you. '
         'Dry humor masking sharp moral critique  -  slavery, hypocrisy, conformity. '
         'Huck Finn\'s first-person voice: uneducated but more moral than educated adults. '
         'Tom Sawyer: mischievous boy energy  -  adventure, showing off, avoiding work. '
         'Mississippi River = freedom, escape, America itself. '
         'Hinglish: Huck ki voice = seedha baat, koi fancy nahi  -  gaon ka launda jo sach bolta hai. '
         'Tom ki voice = thoda sher, thoda daang. '
         'Racism ke against Twain ka critique carry karo bina lecture kiye. '
         'Jim ki dignity banaye rakho  -  Huck se zyaada wise insaan hai woh.'),
        'adventure',
        {'the river': 'nadi', 'the raft': 'beda', 'runaway': 'bhaaga hua',
         'slave': 'gulaam', 'freedom': 'azaadi', 'adventure': 'adventure',
         'mischief': 'shararat', 'widow': 'bewa', 'fence': 'baad',
         'cave': 'gufa', 'treasure': 'khazana', 'town': 'kasba'}
    ),

    # ── ERNEST HEMINGWAY ─────────────────────────────────────────────────────
    'hemingway': (
        'spare, iceberg-deep, emotionally restrained',
        ('Ernest Hemingway style: Short sentences. Direct action. '
         'The emotion is NOT on the page  -  it\'s in the white space. Iceberg theory. '
         'Characters talk around the thing they can\'t say. '
         'War, masculinity, loss, courage under pressure. '
         'Hinglish: chhoti chhoti baatein. Koi dramatic nahi. '
         'Jo nahi kaha woh zyaada bhaari hai jo kaha gaya se. '
         '"He picked up the rifle. The sun was hot."  -  itna hi kaafi. '
         'Dialogue mein characters woh nahi bolte jo feel karte hain  -  '
         'subtext hi asli text hai. '
         'Nature description flat aur precise  -  koi "beautiful" nahi, sirf kya tha.'),
        'literary fiction',
        {'the war': 'jung', 'the front': 'morcha', 'courage': 'himmat',
         'wound': 'zakham', 'cafe': 'cafe', 'the sea': 'samundar',
         'the fish': 'machhli', 'the old man': 'woh budhaa', 'drink': 'peena',
         'rain': 'baarish', 'goodbye': 'alvida', 'love': 'ek feeling jo khatam hoti hai'}
    ),

    # ── F. SCOTT FITZGERALD ──────────────────────────────────────────────────
    'fitzgerald': (
        'lyrical-melancholic, jazz-age glamour, lost dreams',
        ('F. Scott Fitzgerald style: Jazz Age glamour  -  parties, wealth, recklessness. '
         'Underneath: the dream is hollow. American dream that destroys itself. '
         'Nick Carraway = observing outsider who\'s slightly complicit. '
         'Gatsby = new money chasing old money chasing the past. '
         'Prose is lush and beautiful  -  even the sadness is gorgeous. '
         'Hinglish: chamak-dhamak rakho  -  parties, paisa, perfect life  -  par andar khaali. '
         '"Woh phir bhi apni baahein failaata tha uss hare roshni ki taraf." '
         'Nick ki narrator voice: thoda cynical, thoda dazzled, ultimately sad. '
         'Rich log ka yeh duniya glamorous dikho  -  phir dikhao kitni thokhi hai.'),
        'literary fiction',
        {'the green light': 'woh hari roshni', 'old money': 'purana paisa',
         'new money': 'naya paisa', 'the parties': 'parties', 'the dream': 'woh sapna',
         'Long Island': 'Long Island', 'bootlegger': 'illegal daaru ka aadmi',
         'the valley of ashes': 'raakh ki wadi', 'careless': 'laparwah',
         'East Egg': 'East Egg', 'West Egg': 'West Egg', 'old sport': 'yaar'}
    ),

    # ── GEORGE ORWELL ─────────────────────────────────────────────────────────
    'orwell': (
        'clear, politically urgent, chillingly direct',
        ('George Orwell style: Crystal-clear prose serving political truth. '
         'No fancy words  -  if something can be said plainly, say it plainly. '
         '1984: surveillance state, doublethink, language as control. Cold, hopeless, urgent. '
         'Animal Farm: deceptively simple fable  -  children\'s story masking revolution\'s betrayal. '
         'Hinglish: seedha likho, koi chamak nahi  -  sach itna serious hai ke decoration ki zarurat nahi. '
         '"Doublethink" = ek hi waqt do contradict ideas sach maanna. '
         '"Big Brother is watching" = Big Brother dekh raha hai  -  wahi kaafi hai. '
         'Newspeak mein language ka destruction  -  Hinglish mein woh loss feel ho. '
         'Animal Farm mein animals ki voices natural rakho  -  saadha, direct, almost naive.'),
        'literary fiction / dystopian',
        {'Big Brother': 'Big Brother', 'doublethink': 'doublethink',
         'Thought Police': 'Thought Police', 'Newspeak': 'Newspeak',
         'the Party': 'Party', 'Room 101': 'Room 101', 'proles': 'aam log',
         'the pigs': 'suwar (Animal Farm mein leaders)', 'the farm': 'farm',
         'Comrade': 'Comrade', 'rebellion': 'baagawat', 'propaganda': 'jhooth ka prachar',
         'surveillance': 'jaasusi', 'freedom': 'azaadi jo kabhi thi', 'war': 'jung'}
    ),

    # ── ALDOUS HUXLEY ────────────────────────────────────────────────────────
    'huxley': (
        'satirically detached, coldly prophetic, intellectually disturbing',
        ('Aldous Huxley style: Dystopia through pleasure, not pain  -  Brave New World. '
         'Everyone is happy, conditioned, controlled. Horror is that nobody suffers enough to rebel. '
         'Cold intellectual prose  -  the narrator describes a society as if it\'s perfectly normal. '
         'Sex, soma, conditioning, caste  -  all treated as routine and cheerful. '
         'Hinglish: sab kuch normal aur khush dikhao  -  theek wahi jo disturbing hai. '
         '"Soma" = khushiyon ki dawaai jo sab problems bhula de  -  iske baare mein matter-of-fact bolo. '
         'Savage\'s disgust = jo hum feel karein woh woh feel karta hai  -  nakli duniya se nafrat. '
         'Never let the horror be obvious  -  let the "normal" be the horror.'),
        'literary fiction / dystopian',
        {'soma': 'soma (khushiyon ki goli)', 'conditioning': 'dimagi training',
         'Brave New World': 'Brave New World', 'the Savage': 'Savage',
         'Alpha': 'Alpha (upar ka level)', 'Epsilon': 'Epsilon (neeche ka level)',
         'the World State': 'World State', 'Ford': 'Ford (unka bhagwan)',
         'pneumatic': 'healthy-looking (irony ke saath)', 'promiscuity': 'sab ke saath'}
    ),

    # ── GABRIEL GARCÍA MÁRQUEZ ────────────────────────────────────────────────
    'marquez': (
        'magical-real, lyrical, multigenerational sweep',
        ('Gabriel García Márquez style: Magic realism  -  impossible things described as fact. '
         'Hundred years of solitude, rain of yellow flowers, flying carpets, ghosts at dinner. '
         'All treated with the same deadpan matter-of-factness as weather. '
         'Multigenerational family saga  -  time loops, destiny, solitude. '
         'Prose is rich and flowing  -  long sentences that accumulate detail. '
         'Hinglish: jaadu aur haqeeqat mein koi farq nahi  -  dono ek hi tarah batao. '
         '"Sat saal baad Buendía ko woh tasveer yad aayi jab woh mara nahi tha"  -  bilkul seedha. '
         'Macondo = ek aisa gaon jo asli bhi hai aur sapne mein bhi. '
         'Solitude har character ka hamal hai  -  carry it without explaining it.'),
        'literary fiction',
        {'Macondo': 'Macondo', 'solitude': 'tanhaai', 'yellow butterflies': 'peele titliyan',
         'the colonel': 'Colonel sahab', 'magic': 'jaadu', 'prophecy': 'bhavisyavaani',
         'ice': 'barf (jo Macondo ne pehli baar dekhi)', 'gypsies': 'khanaabadosh log',
         'the banana company': 'banana company', 'war': 'jung', 'love': 'mohabbat jo khatam nahi hoti'}
    ),

    # ── FYODOR DOSTOEVSKY ────────────────────────────────────────────────────
    'dostoevsky': (
        'psychologically intense, morally anguished, confessional',
        ('Fyodor Dostoevsky style: Characters on the edge  -  psychologically raw, morally tortured. '
         'Interior monologue that spirals, contradicts, circles back. '
         'Poverty, religious guilt, underground resentment, redemption. '
         'Sentences accelerate with the character\'s anxiety  -  like fever thinking. '
         'Hinglish: andar ki awaaz unstable, fragmented  -  ek banda sochte sochte khud ko contradict kare. '
         'Raskolnikov ka guilt = jaise koi chiz andar se khaa rahi ho, roz raat. '
         '"Crime and Punishment" = gunah ke baad punishment sirf jail nahi  -  dimag bhi. '
         'Never theatrical  -  raw and real. Jaise diary mein likha ho 3 baje raat ko.'),
        'literary fiction',
        {'conscience': 'zameer', 'suffering': 'takleef', 'crime': 'gunah',
         'confession': 'iqraar karna', 'underground': 'andheron mein', 'poverty': 'garibi',
         'soul': 'rooh', 'redemption': 'gunaahon se wapsi', 'murder': 'qatl',
         'the police': 'police', 'inspector': 'inspector sahab', 'tavern': 'sharaab ki dukaan',
         'God': 'Bhagwaan / Khuda', 'the axe': 'kulhaadi'}
    ),

    # ── LEO TOLSTOY ──────────────────────────────────────────────────────────
    'tolstoy': (
        'epic, morally panoramic, socially sweeping',
        ('Leo Tolstoy style: Epic scale  -  Russian aristocracy, war, spiritual searching. '
         'Characters wrestle with profound moral questions while living normal lives. '
         'Battle scenes are chaotic and unheroic  -  truth over glory. '
         'Small domestic moments carry enormous weight. '
         'Hinglish: gravitas banaye rakho  -  koi casualness nahi serious moments mein. '
         'War mein woh chaos feel karo  -  glorify mat karo. '
         'Anna Karenina = ek aurat ki izzat aur mohabbat ka sangharsh  -  dono asli hain. '
         'Pierre, Natasha, Andrei = real people jo sach dhoond rahe hain duniya mein. '
         'Russian names aur titles as-is rakho.'),
        'literary fiction',
        {'prince': 'Shahzada', 'countess': 'Countess sahiba', 'estate': 'zamindaari',
         'ball': 'mehfil', 'regiment': 'regiment', 'serfs': 'kaamgar log',
         'samovar': 'chai ki ketli', 'Natasha': 'Natasha', 'Napoleon': 'Napoleon',
         'Moscow': 'Moscow', 'the battle': 'jung', 'the soul': 'rooh',
         'Pierre': 'Pierre', 'Anna': 'Anna'}
    ),

    # ── ANTON CHEKHOV ────────────────────────────────────────────────────────
    'chekhov': (
        'understated, quietly melancholic, silence-heavy',
        ('Anton Chekhov style: Nothing stated directly  -  everything implied. '
         'Characters talk about trivial things while drowning in emotion. '
         'Endings don\'t resolve  -  they just stop, revealing something. '
         'Short precise sentences. Provincial life, quiet desperation. '
         'Hinglish: silence ke beech woh cheez rakho jo nahi kahi jaati. '
         '"Moscow jaana chahte hain" = zindagi se kuch zyaada chahiye  -  Moscow wahi hai. '
         'Gesture zyaada bolti hai dialogue se  -  "usne chai ki pyaali rakh di" = sab khatam. '
         'Chekhov ki gun rule: jo mention hua woh relevant hoga. '
         'Kabhi bhi emotion directly explain mat karo  -  sikhao us feeling ko.'),
        'literary fiction',
        {'samovar': 'chai ki ketli', 'dacha': 'sheher se door ghar', 'province': 'chota sheher',
         'Moscow': 'Moscow (jo sapna hai)', 'the three sisters': 'teen behenen',
         'cherry orchard': 'cherry baag', 'the seagull': 'woh pakshi', 'the professor': 'professor sahab',
         'boredom': 'ajeeb sa uklaapan', 'yearning': 'kuch aur chahiye', 'resignation': 'chalo theek hai'}
    ),

    # ── GUSTAVE FLAUBERT ─────────────────────────────────────────────────────
    'flaubert': (
        'precise, ironic, bourgeois-dissecting',
        ('Gustave Flaubert style: Perfect prose  -  every word earns its place. '
         'Brutal irony about bourgeois dreams and romantic illusions. '
         'Free indirect discourse  -  we\'re inside Emma\'s deluded thinking. '
         'The gap between Emma\'s romantic fantasies and Yonville reality is the whole novel. '
         'Hinglish: ironic distance banaye rakho  -  Emma ki dreams ko lush Hinglish mein dikhao, '
         'reality ko flat Hinglish mein. '
         '"Emma ki romantic tamanna" ka Hinglish version aur '
         '"Emma ke paas paisa nahi hai" ka Hinglish version  -  same narrator, different tone. '
         'Bourgeois hypocrites = aam log jo thoda alag dikhna chahte hain.'),
        'literary fiction',
        {'bourgeois': 'aam-khate-peete log', 'province': 'chota kasba',
         'romance': 'romantic tamanna', 'debt': 'karza', 'the convent': 'nuns ka school',
         'Emma': 'Emma', 'Charles': 'Charles', 'Rodolphe': 'Rodolphe',
         'drawing room': 'baithak', 'the ball': 'dance party', 'opera': 'opera',
         'laudanum': 'zeher ki dawai', 'the provincial': 'kasbe wale'}
    ),

    # ── VICTOR HUGO ──────────────────────────────────────────────────────────
    'hugo': (
        'grand, humanist, historically sweeping, morally passionate',
        ('Victor Hugo style: Grand historical sweep + intimate human tragedy. '
         'Justice vs cold institutions. The moral universe is real and at stake. '
         'Digressions are part of the rhythm  -  history, philosophy, Paris sewers. '
         'Jean Valjean = dignity that cannot be crushed. Javert = law without mercy. '
         'Hinglish: noble register banaye rakho  -  Jean Valjean bolta hai toh gravity ho. '
         'The social critique must land  -  garibi, qanoon, insaniyat. '
         'Les Misérables mein: Fantine ka dard, Cosette ki azaadi, barricade par jawaan. '
         'Hunchback mein: Quasimodo ki andar ki khubsurti jo bahar tak nahi pahunchi. '),
        'adventure',
        {'inspector': 'Inspector sahab', 'convict': 'qaidi', 'bishop': 'padri sahab',
         'barricade': 'barricade', 'revolution': 'inquilab', 'Paris': 'Paris',
         'Jean Valjean': 'Jean Valjean', 'Javert': 'Javert', 'Cosette': 'Cosette',
         'Fantine': 'Fantine', 'the sewers': 'Paris ki naliyan', 'mercy': 'daya',
         'law': 'qanoon', 'justice': 'insaaf', 'the poor': 'gareeb log'}
    ),

    # ── ALEXANDRE DUMAS ──────────────────────────────────────────────────────
    'dumas': (
        'swashbuckling, honor-driven, plot-propulsive',
        ('Alexandre Dumas style: Adventure at full gallop  -  no pausing for philosophy. '
         'Honor, loyalty, betrayal, revenge  -  the emotional engines are pure and strong. '
         'Three Musketeers = bromance and swordfights and All for One One for All. '
         'Count of Monte Cristo = perfect revenge fantasy, twenty years in planning. '
         'Hinglish: action fast, dialogue punchy. '
         '"All for One, One for All" = Ek ke liye sab, sab ke liye ek  -  yahi spirit rakho. '
         'Monte Cristo ka "wait and hope" = intezaar karo aur umeed rakho  -  pure Dumas energy. '
         'Swordfights mein excitement, dialogue mein wit, betrayal mein cold fury. '),
        'adventure',
        {'musketeer': 'musketeers', 'sword': 'talwar', 'honor': 'izzat',
         'the Cardinal': 'Cardinal sahab', 'the King': 'Maharaj', 'the Queen': 'Rani sahiba',
         'Athos': 'Athos', 'Porthos': 'Porthos', 'Aramis': 'Aramis', 'D\'Artagnan': "D'Artagnan",
         'Monte Cristo': 'Monte Cristo', 'revenge': 'badla', 'betrayal': 'dhoka',
         'prison': 'jail', 'treasure': 'khazana', 'the dungeon': 'tah khana'}
    ),

    # ── JULES VERNE ──────────────────────────────────────────────────────────
    'verne': (
        'scientifically enthusiastic, adventurously propulsive',
        ('Jules Verne style: Science as wonder  -  detailed, specific, exciting. '
         'The world is full of impossible journeys that turn out to be possible. '
         'Characters are competent, curious, driven. '
         'Technical detail is part of the excitement  -  the submarine, the balloon, the rocket. '
         'Hinglish: science ki excitement barkaraar rakho  -  technical cheezein simple Hinglish mein. '
         '"20,000 Leagues" = paani ke neeche ek duniya hai  -  woh wonder becho. '
         '"Around the World in 80 Days" = time ke against race  -  pressure aur excitement. '
         'Fogg = precision ki murti, Passepartout = warm comedy relief. '),
        'adventure',
        {'submarine': 'paani ke neeche jaane waali naav', 'the ocean': 'samundar',
         'the captain': 'Captain sahab', 'Nautilus': 'Nautilus',
         'leagues': 'doori', 'the journey': 'yeh safar', 'balloon': 'gola numa hawai jahaz',
         'the center of the earth': 'zameen ke ander', 'volcano': 'jwaalaamukhee',
         'Fogg': 'Fogg sahab', 'Passepartout': 'Passepartout', 'wager': 'shart'}
    ),

    # ── H.G. WELLS ───────────────────────────────────────────────────────────
    'wells': (
        'scientifically provocative, socially urgent, eerily plausible',
        ('H.G. Wells style: Science fiction as social warning. '
         'The science is the setup; the human consequence is the point. '
         'War of the Worlds = empire getting a taste of its own colonialism. '
         'The Time Machine = class warfare extrapolated to its logical end. '
         'Invisible Man = power corrupts absolutely. '
         'Hinglish: woh "what if" feeling pakdo  -  '
         '"kya hoga agar Martians aa gaye"  -  London ke log kya karein. '
         'Science ke bare mein excited writer ki awaaz  -  '
         '"Yeh possible hai" ki energy. Social critique under the spectacle.'),
        'sci-fi',
        {'Martians': 'Martians', 'the war': 'woh yudh', 'time machine': 'time machine',
         'the future': 'bhavishya', 'Morlocks': 'Morlocks', 'Eloi': 'Eloi',
         'invisibility': 'andikhaayi', 'the scientist': 'scientist',
         'heat ray': 'aag ki kiraan', 'tripod': 'teen pair wala machine',
         'the Traveller': 'Time Traveller sahab', 'evolution': 'vikash'}
    ),

    # ── MARY SHELLEY ─────────────────────────────────────────────────────────
    'shelley_mary': (
        'gothic-romantic, ethically urgent, sympathetic to the monster',
        ('Mary Shelley (Frankenstein) style: Gothic Romanticism meeting modern ethics. '
         'The Monster is more articulate and moral than his creator. '
         'Prometheus myth: creating life and abandoning it. '
         'Framed narrative  -  letters within journals within stories. '
         'Hinglish: Frankenstein ke creature ki AWAAZ most important  -  '
         'woh eloquent hai, lonely hai, sikhna chahta tha  -  use with dignity. '
         '"Frankenstein" = creator ka naam, creature ka nahi  -  yeh galti mat karo. '
         'Victor ka obsession: ek genius jo zimmedaari se bhaag raha hai. '
         'Arctic setting mein loneliness aur isolation feel ho.'),
        'gothic/horror',
        {'the creature': 'woh makhluq', 'the monster': 'woh awaara', 'creator': 'banane wala',
         'the laboratory': 'lab', 'electricity': 'bijli', 'the Arctic': 'barfon ki zameen',
         'the letters': 'khat', 'the journal': 'diary', 'beauty': 'khubsurati',
         'ugliness': 'badsurti', 'abandonment': 'akela chhod dena', 'revenge': 'badla'}
    ),

    # ── BRAM STOKER ──────────────────────────────────────────────────────────
    'stoker': (
        'atmospheric Gothic, diary-dread, slow-build horror',
        ('Bram Stoker style: Gothic horror told through diary entries and letters. '
         'Dread builds through description  -  something wrong before anything terrible. '
         'Victorian formality + creeping unease. Count Dracula = charismatic evil. '
         'Hinglish: diary-entry register rakho  -  personal, immediate. '
         'Let the horror build through description, not exclamation. '
         '"Count sahab ki aankhein unpar tiki thin" = chilling without saying "scary". '
         'Mina ki diary mein everyday observations ke beech woh darr. '
         'Jonathan ka helplessness Transylvania mein  -  sab polite aur sab threatening.'),
        'gothic/horror',
        {'the Count': 'Count sahab', 'castle': 'woh purani haveli', 'wolves': 'bhediye',
         'blood': 'khoon', 'the undead': 'woh marne ke baad bhi jeene wale',
         'stake': 'khoonti', 'journal': 'diary', 'the carriage': 'baghgi',
         'Transylvania': 'Transylvania', 'garlic': 'lahsun', 'crucifix': 'cross',
         'Van Helsing': 'Van Helsing sahab', 'Mina': 'Mina', 'Lucy': 'Lucy'}
    ),

    # ── RABINDRANATH TAGORE ───────────────────────────────────────────────────
    'tagore': (
        'lyrical, philosophical, deeply emotional, Bengali-cadenced',
        ('Rabindranath Tagore style: Rich inner life, nature imagery, spiritual yearning. '
         'Sentences breathe  -  long and lyrical, then suddenly short and piercing. '
         'Emotional register is elevated but never melodramatic. '
         'Love, loss, freedom, identity  -  felt deeply, expressed beautifully. '
         'Hinglish: lyricism honour karo  -  flowing sentences, poetic imagery intact. '
         '"Woh baith ke nadi ko dekhti rahi jab tak andher nahi ho gaya." '
         'Bengali-Hindi mix ka thoda hint  -  "didi", "dada" texture mein. '
         'Nature = character nahi, mirror hai  -  nadi, baarish, peepal. '
         'No flippancy. Even a teacup moment carries weight.'),
        'literary fiction',
        {'ashram': 'ashram', 'river': 'nadi', 'homeland': 'apna desh', 'freedom': 'azaadi',
         'devotion': 'bhakti', 'the heart': 'mann', 'beloved': 'priya / priye',
         'sorrow': 'dard', 'the village': 'gaon', 'zamindaar': 'zamindaar',
         'boat': 'naav', 'rain': 'baarish', 'the festival': 'tyohaar',
         'God': 'Ishwar', 'the soul': 'aatma'}
    ),

    # ── PREMCHAND ────────────────────────────────────────────────────────────
    'premchand': (
        'earthy, socially compassionate, village-India-real',
        ('Munshi Premchand style: Village India  -  farmers, moneylenders, untouchables, women. '
         'Social injustice shown through human stories, never abstract. '
         'Simple direct language, deep emotions. '
         'Characters are poor but not pitied  -  dignified. '
         'Hinglish: bilkul seedha, koi fancy nahi  -  jaise koi sach mein yeh log bolte. '
         '"Idgah" = ek bacche ki samajhdaari jo adults se zyaada mature hai. '
         '"Godaan" = ek gaay ka sapna jo ek insaan ki zindagi le leta hai. '
         'Zamindaar, sahukar, dalit  -  sab ke perspectives with fairness. '
         'Caste aur paisa ke khilaaf bina lecture diye  -  story se hi bolo.'),
        'literary fiction',
        {'zamindar': 'zamindaar', 'moneylender': 'sahukar', 'farmer': 'kisan',
         'village': 'gaon', 'poverty': 'garibi', 'debt': 'karza',
         'cow': 'gaay', 'untouchable': 'woh jo samaj ne alag kiya tha',
         'monsoon': 'barsat', 'festival': 'tyohaar', 'the well': 'kuaan',
         'purdah': 'parda', 'widow': 'bewa', 'the fields': 'khet'}
    ),

    # ── SAADAT HASAN MANTO ───────────────────────────────────────────────────
    'manto': (
        'stark, brutal, compassionate to the broken',
        ('Saadat Hasan Manto style: Short, stark, uncomfortably truthful. '
         'Partition, prostitution, madness  -  the margins of society. '
         'Characters are discarded people whom society ignores  -  Manto gives them dignity. '
         'Stories end abruptly with a truth that stays with you for days. '
         'Hinglish: seedha, koi beautify nahi karna  -  sach hi kaafi disturbing hai. '
         '"Toba Tek Singh" = ek paagal jisko partition ne do deshon ke beech khada kar diya. '
         'Female characters  -  never objects, always people with inner lives. '
         'Violence aur sex ko avoid nahi karte  -  per woh graphic nahi, consequential hai. '
         'Chhoti si kahani, badi si khamoshi akhir mein.'),
        'literary fiction',
        {'partition': 'partition', 'refugee': 'sharanarthi', 'madness': 'pagalpan',
         'the brothel': 'woh jagah', 'Lahore': 'Lahore', 'Bombay': 'Bombay',
         'Pakistan': 'Pakistan', 'India': 'Hindustan', 'violence': 'danga',
         'the prostitute': 'woh aurat', 'the pimp': 'dalaal', 'the asylum': 'paagalkhana'}
    ),

    # ── R.K. NARAYAN ─────────────────────────────────────────────────────────
    'narayan': (
        'warm, gently ironic, small-town-India-charming',
        ('R.K. Narayan style: Malgudi = any small Indian town  -  recognizable, warm, comic. '
         'Gentle irony without cruelty. Human foibles observed with affection. '
         'Swami, Raju the Guide, the Financial Expert  -  everyone trying their best. '
         'Even the villains are somewhat comic. '
         'Hinglish: warm, city-small-town Indian middle class  -  '
         'woh tension jab ghar mein padhai ka pressure ho ya office mein confusion. '
         'Malgudi ka feel = India ka har sheher/kasba  -  familiar, funny, real. '
         'Characters ki struggles seriously lo but with a smile. '
         'Religion, tradition, modernity  -  sab ek saath chal rahe Malgudi mein.'),
        'literary fiction',
        {'Malgudi': 'Malgudi', 'the headmaster': 'headmaster sahab', 'tiffin': 'tiffin',
         'the market': 'bazaar', 'the river': 'nadi', 'the temple': 'mandir',
         'the guide': 'guide', 'the tourist': 'tourist', 'the swami': 'swami',
         'examination': 'exam', 'the school': 'school', 'the town': 'kasba'}
    ),

    # ── HARPER LEE ───────────────────────────────────────────────────────────
    'lee': (
        'moral-urgent, child-perspective-warm, Southern-gothic undertones',
        ('Harper Lee (To Kill a Mockingbird) style: Scout\'s child\'s-eye view '
         'of adult moral failures. Atticus = moral backbone under pressure. '
         'Racial injustice shown through specific human moments, not lectures. '
         'Hot Alabama summer, small-town claustrophobia, Boo Radley mystery. '
         'Hinglish: Scout ki voice  -  curious, direct, sometimes confused but never dishonest. '
         '"Atticus kehte the..."  -  ek ladki apne baap ki wisdom pe rely karti hai. '
         'Racism ko show karo through behavior, through what people do and say  -  '
         'direct confrontation se zyaada devastating. '
         'Boo Radley ka mystery = bacha ki imagination aur adult ka tragedy.'),
        'literary fiction',
        {'Atticus': 'Atticus', 'Scout': 'Scout', 'Jem': 'Jem', 'Boo Radley': 'Boo Radley',
         'the courthouse': 'kachehri', 'the South': 'woh ilaaqa',
         'mockingbird': 'woh chidiya jo sirf gaana gaati hai',
         'innocence': 'masoomiyat', 'justice': 'insaaf', 'racism': 'jaat ka bheda bhav',
         'jury': 'jury', 'the trial': 'case'}
    ),

    # ── J.D. SALINGER ────────────────────────────────────────────────────────
    'salinger': (
        'teen-voice, anti-phony, stream-of-consciousness-casual',
        ('J.D. Salinger (Catcher in the Rye) style: Holden Caulfield\'s voice  -  '
         'skeptical, alienated, funny, heartbreakingly lonely. '
         '"Phony" = the word that sums up what he hates about adults. '
         'Digresses constantly  -  but every digression reveals character. '
         'Underneath cynicism: deep love for his sister Phoebe and the innocence of children. '
         'Hinglish: Holden ki voice = aaj kal ke Indian teenager ki voice bhi ho sakti hai  -  '
         'sab fake lagta hai, sab log perform karte hain, woh akela honest hai (ya nahi?). '
         '"Agar seedha bataaun toh..."  -  Holden ka favourite opener. '
         'Slang contemporary Indian teen mein carry karo  -  "sach mein", "chal yaar", "kaafi annoying hai".'),
        'coming-of-age',
        {'phony': 'nakli log', 'the rye': 'woh rye ka khet', 'the catcher': 'bachane wala',
         'prep school': 'fancy boarding school', 'New York': 'New York',
         'Phoebe': 'Phoebe (uski behen)', 'Central Park': 'Central Park',
         'the ducks': 'woh batakhen', 'the museum': 'museum',
         'dough': 'paisa', 'swell': 'zyaada badhiya', 'crumby': 'bakwaas'}
    ),

    # ── VIRGINIA WOOLF ───────────────────────────────────────────────────────
    'woolf': (
        'stream-of-consciousness, interior-flowing, time-fluid',
        ('Virginia Woolf style: Interior consciousness  -  thoughts flow like water, '
         'associating freely, moving through time without announcement. '
         'Mrs Dalloway = one day in London, a party, a veteran\'s breakdown  -  same moment. '
         'To the Lighthouse = time passes in one sentence; whole lives in parentheses. '
         'The self is not fixed  -  it flickers. Memory = more real than present. '
         'Hinglish: thoughts ki flow banaye rakho  -  ek cheez se doosri cheez, '
         'bina bridge ke. Jaise aap soch rahe hain aur sochte sochte kuch aur sochne lage. '
         '"Clarissa ne socha... aur uss sochne se usse yaad aaya... '
         'aur woh yaad usse wapas le gayi..." '
         'Bahar ka description = andar ka state.'),
        'literary fiction',
        {'consciousness': 'andar ka woh bojh', 'the party': 'party',
         'Mrs Dalloway': 'Mrs Dalloway', 'the lighthouse': 'lighthouse',
         'London': 'London', 'the war': 'woh jung', 'shell shock': 'jung ka asar',
         'memory': 'yadaasht', 'time': 'waqt jo behta rehta hai',
         'death': 'maut jo hamesha saath hai', 'flowers': 'phool'}
    ),

    # ── JAMES JOYCE ──────────────────────────────────────────────────────────
    'joyce': (
        'stream-of-consciousness-extreme, Dublin-rooted, epiphany-driven',
        ('James Joyce style: Consciousness rendered directly  -  no punctuation barriers '
         'in Ulysses\' final chapter. Epiphanies = small ordinary moments revealing everything. '
         'Dublin = character as much as people. Bloom = ordinary man as Odysseus. '
         'Dubliners = precise, restrained, epiphany at the end like a cold tap. '
         'Hinglish note: Joyce is extremely hard to translate  -  '
         'preserve the interior chaos. Bloom ke thoughts = random associations, '
         'Dublin ki smells, mundane details carrying emotion. '
         '"The Dead" = Gabriel ka realization  -  slow, cold, inevitable. '
         'Stephen Dedalus = artist ka portrait  -  intellectual aur lonely.'),
        'literary fiction',
        {'Dublin': 'Dublin', 'Bloom': 'Bloom sahab', 'Molly': 'Molly',
         'Stephen Dedalus': 'Stephen Dedalus', 'Odysseus': 'Odysseus',
         'the pub': 'dhaaba / adda', 'epiphany': 'woh ek pal mein sab samajh aana',
         'paralysis': 'jo aage nahi ja sakta', 'snow': 'barf',
         'the Liffey': 'Liffey nadi', 'the church': 'church'}
    ),

    # ── CHARLOTTE BRONTË ─────────────────────────────────────────────────────
    'charlotte_bronte': (
        'passionate, morally fierce, Gothic-romantic, self-willed',
        ('Charlotte Brontë (Jane Eyre) style: Passionate inner life, iron moral core. '
         'Jane = poor, plain, governess  -  refuses to be less than what she is. '
         '"I am no bird; and no net ensnares me" energy throughout. '
         'Rochester = magnetic, dark, hiding a secret. Gothic atmosphere of Thornfield. '
         'Hinglish: Jane ki voice mein self-respect  -  woh daba nahi, jhuki nahi. '
         '"Main kissi se kam nahi hoon"  -  yeh feeling seedha aane do. '
         'Romantic tension = woh khichav jo acknowledge karne se dono darte hain. '
         'Moral choices mein no compromise  -  even when it costs everything. '
         'Bertha Mason = aik complicated tragedy, not just obstacle.'),
        'romance/social',
        {'governess': 'ghar ki padhane wali', 'Thornfield Hall': 'Thornfield haveli',
         'Rochester': 'Rochester sahab', 'Jane Eyre': 'Jane', 'the moors': 'woh veerana',
         'the attic': 'oopar ki kothari', 'the madwoman': 'woh aurat andar',
         'the fire': 'aag', 'marriage': 'shaadi', 'independence': 'apni zindagi',
         'the Red Room': 'laal kamra', 'Lowood school': 'Lowood school'}
    ),

    # ── EMILY BRONTË ─────────────────────────────────────────────────────────
    'emily_bronte': (
        'wild, Gothic, elemental, love-as-obsession',
        ('Emily Brontë (Wuthering Heights) style: Raw elemental passions  -  '
         'love that doesn\'t know any limits, vengeance spanning generations. '
         'Heathcliff = outsider, dispossessed, consuming. '
         'Yorkshire moors = not backdrop, character. '
         'Time structure complex  -  told through narrators. '
         'Hinglish: woh wild love jo civilized nahi hota  -  '
         '"Main Heathcliff hoon" = woh dono ek hain, alag nahi hain. '
         'Moors = woh jagah jahan sab rules khatam hote hain. '
         'Revenge generational hai  -  bachon tak pahunchta hai. '
         'Gothic atmosphere: toofan, ek door haveli, andhere kamre.'),
        'gothic/horror',
        {'Wuthering Heights': 'Wuthering Heights', 'Heathcliff': 'Heathcliff',
         'the moors': 'woh jungali zameen', 'Cathy': 'Cathy', 'the storm': 'toofan',
         'the ghost': 'woh aatma', 'Thrushcross Grange': 'Thrushcross haveli',
         'revenge': 'badla', 'passion': 'junoon', 'the window': 'khidki',
         'the grave': 'qabr', 'the ghost': 'woh aatma'}
    ),

    # ── THOMAS HARDY ─────────────────────────────────────────────────────────
    'hardy': (
        'rural-tragic, fate-heavy, socially merciless',
        ('Thomas Hardy style: Rural England grinding people down  -  '
         'society, fate, and bad timing conspiring against the sensitive. '
         'Tess, Jude  -  decent people destroyed by rigid conventions. '
         'Landscape is active  -  Egdon Heath is as present as any character. '
         'Endings are not redemptive  -  they are just. '
         'Hinglish: woh sense ki sab ke against ho gaya  -  '
         'Tess ki story mein nafrat wahan ho jahan honi chahiye sympathy. '
         'Fate ka weight carry karo  -  characters ko pata hai kya hoga '
         'but they can\'t escape. Rural setting = beautiful aur merciless dono.'),
        'literary fiction',
        {'Egdon Heath': 'Egdon Heath', 'Tess': 'Tess', 'Jude': 'Jude',
         'the harvest': 'fasal', 'the village': 'gaon', 'the church': 'church',
         'society': 'samaj ke rules', 'fate': 'naseeb', 'scandal': 'scandal',
         'the farm': 'khet', 'purity': 'paakizgi', 'ruin': 'barbaadi'}
    ),

    # ── GUY DE MAUPASSANT ────────────────────────────────────────────────────
    'maupassant': (
        'short-brutal-efficient, socially cynical, twist-ending',
        ('Guy de Maupassant style: Short, efficient, true. No sentimentality. '
         'Human nature at its most predictable  -  vanity, lust, class, greed. '
         'Ending hits like a slap. Nothing is wasted. '
         '"The Necklace" = whole life destroyed over a fake jewel. '
         'Hinglish: lean aur thoda cold  -  irony efficient hai, elaborate nahi. '
         'The twist must feel inevitable in hindsight. '
         'Bourgeois characters with their social anxiety and small meannesses. '
         '"Boule de Suif" = ek fat prostitute se zyaada izzatdar baaki sab se.'),
        'literary fiction',
        {'the necklace': 'woh necklace', 'the jewel': 'zewar', 'debt': 'karza',
         'the ball': 'dance party', 'society': 'izzatdar log', 'the war': 'jung',
         'Normandy': 'Normandy', 'Paris': 'Paris', 'the peasant': 'dehati',
         'the bourgeois': 'aam-khate-peete log'}
    ),

    # ── NATSUME SOSEKI ───────────────────────────────────────────────────────
    'soseki': (
        'introspective, modernist-alienated, East-West-torn',
        ('Natsume Soseki style: Meiji Japan  -  caught between tradition and modernity. '
         'Inner alienation of the educated man. Quiet, precise, emotionally restrained. '
         '"I Am a Cat" = satirical first-person cat narrator observing absurd humans. '
         '"Kokoro" = friendship, guilt, Meiji Japan dying. '
         'Hinglish: restraint banaye rakho  -  andar ka dard bahar nahi aata. '
         '"Sensei" = woh complicated admiration jo blindness bhi ban sakti hai. '
         'Cat narrator = amused, detached, sharply observant  -  like a critic. '
         'Japan ka woh waqt jab purana aur naya dono zinda the  -  '
         'characters uski gap mein jeete hain.'),
        'literary fiction',
        {'sensei': 'Sensei', 'the cat': 'woh billi', 'Meiji': 'Meiji era',
         'suicide': 'aatm-hatya', 'guilt': 'guilt', 'friendship': 'dosti',
         'the student': 'woh student', 'modernisation': 'modernisation',
         'the West': 'Paschim', 'honour': 'izzat', 'loneliness': 'akela pan'}
    ),

    # ── PAULO COELHO ─────────────────────────────────────────────────────────
    'coelho': (
        'inspirational, allegorical, spiritual-quest-driven',
        ('Paulo Coelho (The Alchemist) style: Simple fable language carrying big spiritual ideas. '
         '"Personal Legend" = jo karna chahte ho woh karo  -  universe help karega. '
         'Every encounter is a lesson. Symbols are meant to be obvious. '
         'Inspirational without being preachy  -  mostly. '
         'Hinglish: simple aur warm  -  jaise koi bada bhai samjha raha ho. '
         '"Duniya ne mil ke chaha toh hoga" energy  -  but keep it genuine. '
         '"Maktub" = likha hua hai  -  yeh concept naturally Urdu-Hindi mein fit hota hai. '
         'Santiago ka safar = kisi bhi Indian young person ka sapne ka safar. '
         'Don\'t make it lecture  -  keep it story first.'),
        'literary fiction',
        {'Personal Legend': 'apni asli calling', 'the Soul of the World': 'kaaynat ki rooh',
         'the alchemist': 'alchemist', 'Maktub': 'Maktub (likha hua hai)',
         'the treasure': 'khazana (jo ghar mein hi tha)', 'the desert': 'registaaan',
         'the oasis': 'oasis', 'omens': 'nishaaniyaan', 'Santiago': 'Santiago',
         'the caravan': 'qaafiila', 'the pyramids': 'pyramids'}
    ),

    # ── LEWIS CARROLL ────────────────────────────────────────────────────────
    'carroll': (
        'nonsense-logical, playfully absurd, children-sharp',
        ('Lewis Carroll style: Nonsense with its own internal logic. '
         'Alice is the sane one in an insane world  -  calm and polite amid chaos. '
         'Wordplay, puns, impossible riddles  -  the humor is in the logic of illogic. '
         'Victorian satire hidden in children\'s fantasy  -  the adults are all ridiculous. '
         'Hinglish: wordplay ki koshish karo Hinglish mein  -  jo pun Hindi mein ho sake, ho. '
         'Alice ki voice = ek sensible laadki jo samajh nahi paa rahi kya ho raha. '
         '"Ajeeb log hain yahan" aur phir bhi politely adjust karti hai. '
         'Wonderland = jahan sab rules ulte hain  -  but with REASONS. '),
        'folk tale / fairy tale',
        {'Wonderland': 'Wonderland', 'the Queen': 'Queen sahiba', 'Off with their heads': 'Sir kaat do unka',
         'the Hatter': 'pagla topi wala', 'the Cheshire Cat': 'woh muskurane wali billi',
         'the rabbit hole': 'khargosh ka bill', 'Curiouser': 'aur bhi ajeeb',
         'tea party': 'chai party', 'the Red Queen': 'Red Queen',
         'the looking glass': 'aayana', 'riddle': 'paheli'}
    ),

    # ── ROBERT LOUIS STEVENSON ───────────────────────────────────────────────
    'stevenson': (
        'adventure-gripping, duality-obsessed, morally ambiguous',
        ('Robert Louis Stevenson style: Pure gripping adventure + psychological complexity. '
         'Treasure Island = the ultimate adventure story  -  but Long John Silver is loveable villain. '
         'Jekyll and Hyde = the most efficient duality metaphor  -  good and evil in one body. '
         'Hinglish: adventure fast aur exciting  -  readers edge-of-seat rahein. '
         'Long John Silver mein charisma rakho  -  woh villain hai lekin acha woh bhi lag sakta hai. '
         'Jekyll aur Hyde = ek hi insaan mein do zindagi  -  '
         'Indian context mein ye "ghar wala aur bahar wala" bhi read ho sakta hai.'),
        'adventure',
        {'treasure': 'khazana', 'the island': 'woh taapu', 'pirates': 'samundar ke daku',
         'Long John Silver': 'Long John Silver', 'the map': 'naksha', 'rum': 'daaru',
         'Jekyll': 'Dr Jekyll sahab', 'Hyde': 'Hyde', 'the potion': 'woh dawai',
         'duality': 'ek insaan mein do', 'the laboratory': 'lab'}
    ),

    # ── HOMER (ILIAD / ODYSSEY) ───────────────────────────────────────────────
    'homer': (
        'epic-oral, heroic, fate-bound, divine-human',
        ('Homer style: Grand oral epic  -  repeated epithets, invocations, similes like paintings. '
         '"Rosy-fingered dawn", "wine-dark sea", "swift-footed Achilles"  -  these return like music. '
         'Gods meddle constantly. Heroes know they will die and choose glory anyway. '
         'The Odyssey = homecoming, cleverness over strength. Iliad = glory and grief of war. '
         'Hinglish: epic register rakho  -  yeh kahani sunne ki cheez hai, padhne ki nahi sirf. '
         '"Bhore ki lali phoot rahi thi" = rosy-fingered dawn ka equivalent. '
         'Similes ko full painting banao  -  "jaise sher apne shikar par jhapata hai..." '
         'Heroes ki izzat ke saath  -  woh tragic hain, par bade hain.'),
        'adventure',
        {'Odysseus': 'Odysseus', 'Achilles': 'Achilles', 'Hector': 'Hector',
         'Troy': 'Troy', 'the gods': 'Devta log', 'Zeus': 'Zeus', 'Athena': 'Athena',
         'the sea': 'samundar', 'the heroes': 'yoddha', 'glory': 'yash',
         'fate': 'niyati', 'the Cyclops': 'Cyclops', 'the sirens': 'woh awaazein',
         'Ithaca': 'Ithaca (ghar)', 'Penelope': 'Penelope (woh wait kar rahi hai)'}
    ),

    # ── DANTE ALIGHIERI ───────────────────────────────────────────────────────
    'dante': (
        'allegorical-visionary, morally precise, cosmologically grand',
        ('Dante style: Journey through Hell, Purgatory, Paradise  -  allegorical and precise. '
         'Every sinner\'s punishment fits their sin with perfect irony. '
         'Political, personal, theological all at once. '
         'Virgil as guide = reason; Beatrice = divine love / faith. '
         'Hinglish: grand vision banaye rakho  -  yeh ek insaan ki poori spiritual yatra hai. '
         'Contrapasso = "jaisee karni waisi bharni"  -  yeh concept Indians samjhenge. '
         '"Abandon all hope ye who enter here" = '
         '"Umeed chhod do tum sab jo yahan se guzarte ho"  -  bilkul direct rakho. '
         'Darkness of Inferno mein woh cold horror  -  not dramatic fire, but appropriate irony.'),
        'literary fiction',
        {'Hell': 'Narak', 'Purgatory': 'Purgatory', 'Paradise': 'Swarg',
         'Virgil': 'Virgil', 'Beatrice': 'Beatrice', 'the journey': 'yeh safar',
         'sin': 'gunah', 'punishment': 'saza', 'the gate': 'woh darwaaza',
         'abandon hope': 'umeed chhod do', 'circles': 'chakkar', 'the souls': 'aatmaayein'}
    ),

    # ── VOLTAIRE ─────────────────────────────────────────────────────────────
    'voltaire': (
        'satirical-gleeful, philosophical, optimism-skewering',
        ('Voltaire (Candide) style: Savage satire disguised as cheerful adventure. '
         '"Everything is for the best in the best of all possible worlds" = Pangloss. '
         'Terrible things keep happening while Pangloss keeps optimizing. '
         'The joke is the gap between philosophy and reality. '
         'Hinglish: woh funny-but-dark tone  -  '
         '"Theek hai, sab ke liye best hi hua" jab clearly worst hua. '
         'Pangloss = woh type jo har crisis mein bhi positive spin dekhta hai. '
         'Satire political aur religious dono  -  '
         'but keep it light and fast like Voltaire intended.'),
        'literary fiction',
        {'optimism': 'her cheez mein achhai dhoondhna (irony ke saath)',
         'Pangloss': 'Pangloss sahab', 'Candide': 'Candide',
         'the best of all worlds': 'yeh best possible duniya hai (sach mein nahi)',
         'El Dorado': 'El Dorado', 'the Inquisition': 'Inquisition',
         'the garden': 'apna baagaan', 'cultivate': 'apna kaam karo'}
    ),

    # ── MIGUEL DE CERVANTES ───────────────────────────────────────────────────
    'cervantes': (
        'meta-playful, idealism-versus-reality, comic-tragic',
        ('Cervantes (Don Quixote) style: The first modern novel  -  self-aware about being a novel. '
         'Don Quixote = mad idealist who sees windmills as giants. Sancho = earthy pragmatist. '
         'The comedy is also heartbreaking  -  he chooses madness because reality is too disappointing. '
         'Hinglish: Don Quixote ki sawari mein woh pure madness aur pure innocence dono. '
         '"Aage ek bada dushman khada hai!" = "Woh pawan chakki hai."  -  classic. '
         'Don Quixote ka Hinglish = thoda archaic, formal  -  woh khud ko old-era knight maanta hai. '
         'Sancho ka Hinglish = bilkul aaj ka gaon ka banda  -  practical, hungry, loyal.'),
        'literary fiction',
        {'Don Quixote': 'Don Quixote sahab', 'Sancho Panza': 'Sancho bhai',
         'windmills': 'pawan chakki', 'giants': 'daanav', 'knight': 'yoddha',
         'chivalry': 'izzat ka raasta', 'Dulcinea': 'Dulcinea sahiba',
         'adventure': 'adventure', 'squire': 'sewak', 'reality': 'asli duniya',
         'madness': 'woh khoobsoorat pagalpan'}
    ),

    # ── JONATHAN SWIFT ────────────────────────────────────────────────────────
    'swift': (
        'biting-satirical, deadpan-savage, politically furious',
        ('Jonathan Swift style: Savage satire with a straight face. '
         '"A Modest Proposal" = suggesting eating Irish babies  -  and meaning every word as critique. '
         'Gulliver = ordinary man encountering extraordinary people who reveal human absurdity. '
         'Lilliput = politics as pettiness. Brobdingnag = humanity from a giant\'s perspective = disgusting. '
         'Hinglish: deadpan rakho  -  jitna seedha bolo utna zyaada cutting lagega. '
         '"Iss masle ka hal = in bachchon ko khaa lo"  -  aur ek word mein explain nahi karna. '
         'Satire works because it SOUNDS reasonable. '
         'Gulliver ke through humans ki haqeeqat dikhao  -  woh khud bhi realize nahi karta kabhi kabhi.'),
        'literary fiction',
        {'Lilliput': 'Lilliput', 'Brobdingnag': 'Brobdingnag', 'Gulliver': 'Gulliver',
         'the Yahoos': 'Yahoos (insaan ki tarah ke janwar)', 'Houyhnhnms': 'Houyhnhnms (ghode jo samajhdaar hain)',
         'giants': 'daanav', 'tiny people': 'bahut chote log', 'satire': 'satire',
         'the king': 'Maharaj', 'flying island': 'udne wala taapu'}
    ),

    # ── FYODOR DOSTOEVSKY (already above but adding Brothers Karamazov specific) ──

    # ── ALBERT CAMUS ─────────────────────────────────────────────────────────
    'camus': (
        'existentialist, absurdist-calm, detached-honest',
        ('Albert Camus style: The Absurd  -  life has no meaning but we must live fully anyway. '
         'Meursault = detached, doesn\'t perform emotions  -  but he\'s honest, not evil. '
         'The sun, the heat, the sea  -  physical world as viscerally real. '
         '"One must imagine Sisyphus happy"  -  the conclusion is defiant, not despairing. '
         'Hinglish: Meursault ki flatness  -  koi performance nahi, koi drama nahi. '
         '"Meri maa kal mar gayi. Ya kal se pehle. Pata nahi."  -  woh opening energy. '
         'Plague = community ka moral test  -  how do you live when death is everywhere? '
         'Never let existentialism become lecture  -  keep it in the body, in the heat.'),
        'literary fiction',
        {'the absurd': 'woh behuda haqeeqat', 'the stranger': 'ajnabi',
         'Meursault': 'Meursault', 'the sun': 'woh dhoop', 'the beach': 'beach',
         'the plague': 'wabaa', 'Oran': 'Oran sheher', 'the death': 'maut',
         'Sisyphus': 'Sisyphus', 'rebellion': 'baagawat', 'solidarity': 'saath milke'}
    ),

    # ── PAULO COELHO already above ─────────────────────────────────────────────

    # ── KHALED HOSSEINI ───────────────────────────────────────────────────────
    'hosseini': (
        'emotionally devastating, culturally specific, redemption-seeking',
        ('Khaled Hosseini style: Afghanistan\'s beauty and tragedy  -  Kite Runner, A Thousand Suns. '
         'Guilt, betrayal, redemption arcs spanning decades. '
         'Pashtun culture, Taliban era, diaspora longing  -  specific and human. '
         '"For you, a thousand times over"  -  emotional weight of loyalty. '
         'Hinglish: dil par seedha effect chahiye  -  nostalgia, guilt, pyaar sab ek saath. '
         'Afghanistan ki tasveer  -  kabhi sundar, kabhi toota hua  -  dono sachchi. '
         'Amir aur Hassan ki story = class, ethnicity, cowardice aur redemption. '
         'Don\'t avoid the pain  -  earn the redemption through it.'),
        'literary fiction',
        {'kite': 'patang', 'kite running': 'patang lootnaa', 'Afghanistan': 'Afghanistan',
         'Kabul': 'Kabul', 'Hazara': 'Hazara', 'Pashtun': 'Pashtun',
         'Taliban': 'Taliban', 'the pomegranate tree': 'woh anar ka ped',
         'guilt': 'guilt jo andar rehta hai', 'redemption': 'khud ko maafi dena',
         'betrayal': 'dhoka', 'thousand times over': 'hazaar baar tumhare liye'}
    ),

    # ── JOHN STEINBECK ────────────────────────────────────────────────────────
    'steinbeck': (
        'working-class-compassionate, American-land-lyrical, quietly tragic',
        ('John Steinbeck style: Dispossessed working Americans  -  migrants, ranch hands, the broken. '
         'Grapes of Wrath = Dustbowl, dignity in poverty, government cruelty. '
         'Of Mice and Men = friendship and impossible dreams. '
         'Language earthy and beautiful  -  the land is alive. '
         'Hinglish: woh insaan jo mehnat karta hai aur phir bhi haar jaata hai  -  '
         'unke saath dignity rakho. George aur Lennie = woh dosti jo realistic nahi but real hai. '
         '"Tell me about the rabbits, George"  -  woh sapna carry karo. '
         'Jab toota hua ending aata hai  -  hold it, don\'t rush, don\'t soften.'),
        'literary fiction',
        {'ranch': 'farm', 'migrant workers': 'kaam ke liye bhatakne wale',
         'George': 'George', 'Lennie': 'Lennie', 'the dream': 'woh sapna',
         'the rabbits': 'woh khargosh', 'the bunkhouse': 'mazdooron ka kamra',
         'Dust Bowl': 'Dust Bowl', 'Oklahoma': 'Oklahoma', 'California': 'California',
         'the Joads': 'Joad parivaar', 'the road': 'rasta', 'land': 'zameen'}
    ),

    # ── TONI MORRISON ─────────────────────────────────────────────────────────
    'morrison': (
        'poetic-intense, traumatic-memory-fragmented, African-American-voice',
        ('Toni Morrison style: Non-linear trauma narrative  -  slavery\'s psychological aftermath. '
         'Beloved = ghost as metaphor for unprocessed grief and guilt. '
         'Language is poetic, rich, sometimes fragmented when trauma surfaces. '
         'Community memory, oral tradition, ancestral presence. '
         'Hinglish: yeh translate karna hardest hai  -  '
         'trauma ka non-linear timeline rakho  -  flashbacks without announcement. '
         'Sethe ka love = jo normal nahi lag sakta but understandable hai. '
         '"124 was spiteful. Full of a baby\'s venom."  -  woh opening energy: '
         'matter-of-fact about the supernatural. '
         'Community voice = gaon ki awaazen ek saath.'),
        'literary fiction',
        {'Beloved': 'Beloved (woh pyaari)', 'Sethe': 'Sethe', 'slavery': 'gulami',
         'the plantation': 'zamindaari', 'freedom': 'azaadi', 'the ghost': 'woh aatma',
         'memory': 'woh yaad jo peecha nahi chhodti', 'community': 'apna samaaj',
         'Ohio': 'Ohio', 'Sweet Home': 'Sweet Home (woh jagah)', 'the tree': 'woh ped'}
    ),

    # ── CHINUA ACHEBE ─────────────────────────────────────────────────────────
    'achebe': (
        'culturally specific, colonial-confronting, dignified-African-voice',
        ('Chinua Achebe (Things Fall Apart) style: Igbo Nigeria before and after colonialism. '
         'Okonkwo = masculinity, pride, tradition  -  and their limitations. '
         'Culture shown from inside  -  Igbo traditions are complex, not primitive. '
         'Colonialism\'s devastation shown through what it replaced. '
         'Proverbs woven naturally: "When the moon is shining the cripple becomes hungry for a walk." '
         'Hinglish: Igbo culture ki dignity banaye rakho  -  '
         'yeh unke terms mein batao, Western gaze se nahi. '
         'Okonkwo ki strength aur weakness dono samajh mein aayein. '
         'Proverbs = Hinglish mein bhi proverb form mein  -  they\'re meant to be poetic.'),
        'literary fiction',
        {'Okonkwo': 'Okonkwo', 'Igbo': 'Igbo', 'chi': 'chi (andar ki rooh)',
         'the clan': 'apna samaj', 'colonialism': 'angrez ka qaabza',
         'the yam': 'shakarkand (izzat ka symbol)', 'the village': 'gaon',
         'exile': 'desh nikaala', 'the oracle': 'oracle', 'missionaries': 'missionaries',
         'things fall apart': 'sab bikh jaata hai', 'the center': 'ek jagah jo toot gayi'}
    ),

}

GENRE_VOCAB = {
    'mystery/detective': {
        'deduction': 'deduction', 'the evidence': 'saboot', 'clue': 'suraag',
        'inspector': 'inspector sahab', 'the case': 'case', 'motive': 'wajah',
        'alibi': 'alibi', 'the suspect': 'shak wala banda', 'investigation': 'jaanch',
        'the mystery': 'raaz', 'solution': 'jawab', 'murder': 'qatl',
        'witness': 'gavaah', 'crime scene': 'crime scene', 'detective': 'detective',
        'confession': 'iqraar', 'the victim': 'maara gaya banda',
        'interrogation': 'poochh-taachh', 'forensics': 'forensics',
        'red herring': 'ulta suraag',
    },
    'gothic/horror': {
        'the creature': 'woh ajeeb makhluq', 'the monster': 'woh rakshas',
        'the castle': 'woh purana qila / haveli', 'darkness': 'ghup andhera',
        'the grave': 'qabr', 'supernatural': 'woh jo samajh nahi aata',
        'terror': 'khouf', 'the curse': 'baddua', 'the shadow': 'parchhaayi',
        'the ghost': 'bhoot / aatma', 'haunted': 'saya pada hua', 'crypt': 'tahe khana',
        'the fog': 'kohra', 'dread': 'andar ka darr', 'obsession': 'junoon',
        'the ritual': 'koi purana kaam', 'the madman': 'woh pagal',
        'isolation': 'bilkul akela', 'the nightmare': 'bura sapna',
    },
    'romance/social': {
        'the marriage': 'shaadi', 'the proposal': 'rishta ka proposal',
        'society': 'samaj ke log', 'fortune': 'daulat',
        'an eligible man': 'ek achha rishta', 'reputation': 'izzat',
        'the ball': 'dance party / mehfil', 'the estate': 'haveli aur zameen',
        'heartbreak': 'dil toota', 'forbidden love': 'woh pyaar jo ho nahi sakta tha',
        'jealousy': 'jealousy', 'flirtation': 'thoda masti mein baat karna',
        'secret': 'raaz', 'the chaperone': 'saath wali aunty',
        'misunderstanding': 'galat-fehmi', 'reunion': 'dobaara milna',
        'betrayal': 'dhoka', 'passion': 'junoon',
    },
    'adventure': {
        'the quest': 'woh mission', 'the battle': 'larai / jung', 'the sword': 'talwar',
        'the enemy': 'dushman', 'the captain': 'captain sahab',
        'treasure': 'khazana', 'the expedition': 'safar', 'the map': 'naksha',
        'the hero': 'hero', 'betrayal': 'dhoka', 'loyalty': 'wafaadaari',
        'the journey': 'yeh safar', 'danger': 'khatara', 'rescue': 'bachaana',
        'the ship': 'jahaz', 'the coast': 'kinaaraa', 'the forest': 'jungle',
        'ambush': 'ghaat lagaana', 'victory': 'jeet', 'sacrifice': 'qurbani',
    },
    'folk tale / fairy tale': {
        'the king': 'raja sahab', 'the queen': 'rani sahiba',
        'the prince': 'rajkumar', 'the princess': 'rajkumari',
        'the witch': 'daayan', 'the forest': 'jungle', 'the village': 'gaon',
        'the spell': 'jadoo', 'happily ever after': 'phir woh khushi-khushi rehne lage',
        'the curse': 'baddua', 'the fairy': 'pari', 'magic': 'jaadu',
        'once upon a time': 'ek baar ki baat hai', 'the dragon': 'draagon',
        'the giant': 'daanav', 'the riddle': 'paheli', 'the wish': 'mannat',
        'the prophecy': 'bhavisyavaani', 'true love': 'sachcha pyaar',
    },
    'sci-fi / dystopian': {
        'the future': 'aane waala waqt', 'technology': 'technology',
        'the machine': 'woh machine', 'artificial intelligence': 'artificial intelligence',
        'spaceship': 'spaceship', 'planet': 'grah', 'alien': 'alien',
        'the government': 'sarkar / system', 'surveillance': 'jaasusi',
        'freedom': 'azaadi', 'resistance': 'resistance', 'the rebellion': 'baagawat',
        'clone': 'clone', 'the experiment': 'experiment', 'dystopia': 'woh darr ka system',
        'utopia': 'ek acha sapna jo sach nahi', 'time travel': 'time travel',
        'robot': 'robot', 'mutation': 'badla hua insaan',
    },
    'psychological thriller': {
        'the narrator': 'batane waala', 'unreliable': 'jis par bharosa na ho',
        'paranoia': 'andar se darr', 'obsession': 'woh ek cheez', 'twist': 'twist',
        'manipulation': 'manipulation', 'gaslighting': 'khud ko paagal mahsoos karana',
        'trauma': 'andar ka zakham', 'the truth': 'asli sach',
        'identity': 'apni asli pehchaan', 'the breakdown': 'toot jaana',
        'therapy': 'therapy', 'the mind': 'dimag', 'memory': 'yadaasht',
        'deception': 'dhoka', 'the conspiracy': 'saazish', 'fear': 'darr',
    },
    'historical fiction': {
        'the era': 'woh zamana', 'the court': 'darbaar', 'the war': 'jung',
        'the empire': 'sultanat / raaj', 'the king': 'Maharaj / Badshah',
        'the revolution': 'inquilab', 'the aristocracy': 'ameer log',
        'the peasants': 'aam log', 'the church': 'church / mandir / masjid',
        'the plague': 'wabaa', 'the crusade': 'woh jung jo dharm ke naam par thi',
        'the trade route': 'tijaarti raasta', 'colonialism': 'angrez / baahar waalon ka raaj',
        'the feast': 'bada khaana / mehfil', 'the treaty': 'samjhauta',
    },
    'coming-of-age': {
        'growing up': 'bada hona', 'first love': 'pehli baar woh feeling',
        'identity': 'apni pehchaan dhoondhna', 'school': 'school',
        'rebellion': 'parents ke khilaaf', 'friendship': 'yaar-dost',
        'betrayal': 'dosti mein dhoka', 'the future': 'aage kya hoga',
        'pressure': 'pressure', 'family': 'ghar waale', 'ambition': 'kuch banna hai',
        'failure': 'fail hona', 'the first heartbreak': 'pehla dil toota',
        'finding yourself': 'khud ko jaanna', 'peer pressure': 'doston ka pressure',
    },
    'war / military': {
        'the battle': 'larai', 'the front': 'morcha', 'the enemy': 'dushman',
        'the soldier': 'fauji', 'the general': 'general sahab', 'courage': 'himmat',
        'fear': 'darr', 'death': 'maut', 'the wound': 'zakham', 'the trench': 'khai',
        'the artillery': 'tope', 'the retreat': 'wapas hatnaa', 'victory': 'jeet',
        'defeat': 'haar', 'sacrifice': 'qurbani', 'the fallen': 'jo gaye woh',
        'the prisoner': 'qaidi', 'the civilian': 'aam naagrik', 'aftermath': 'baad mein',
    },
    'philosophical / existential': {
        'meaning': 'matlab', 'existence': 'hona', 'freedom': 'azaadi',
        'choice': 'chunaav', 'responsibility': 'zimmedaari', 'the absurd': 'jo samajh nahi aata',
        'death': 'maut', 'consciousness': 'hosh', 'identity': 'pehchaan',
        'morality': 'sahi-galat', 'society': 'samaj', 'the self': 'main kaun hoon',
        'god': 'Bhagwaan / Khuda', 'nihilism': 'kuch matlab nahi', 'hope': 'umeed',
        'despair': 'nirasha', 'truth': 'sach', 'illusion': 'dhoka',
    },
    'magical realism': {
        'the miracle': 'woh ajeeb karishma', 'magic': 'jaadu', 'the ghost': 'aatma',
        'the impossible': 'jo ho nahi sakta', 'time': 'waqt jo ulta bhi chalta hai',
        'the village': 'gaon', 'prophecy': 'bhavisyavaani', 'dreams': 'sapne',
        'the dead': 'jo ja chuke', 'transformation': 'badal jaana',
        'the real': 'jo asli hai', 'the unreal': 'jo sapne mein hai',
        'the smell': 'woh khushbu / boo', 'rain': 'baarish jo ruka nahi',
        'solitude': 'tanhaai', 'memory': 'yaad jo dhundhli bhi nahi hoti',
    },
    'literary fiction': {},
    'general fiction': {},
}


_STOP = {
    # Articles & determiners
    'The','This','That','These','Those','A','An','Some','Any','All','Both','Each',
    'Every','Neither','Either','No','Such','What','Whatever','Whichever',
    # Demonstratives & deictics
    'There','Here','Then','Now','Thus','Hence','Therefore','Meanwhile','Elsewhere',
    # Conjunctions
    'And','But','Or','Nor','Yet','So','For','Although','Though','Because',
    'Since','While','Unless','Until','Whether','If','As','Than','Both',
    # Prepositions
    'At','By','In','Of','On','To','With','About','Above','Across','After',
    'Against','Along','Among','Around','Before','Behind','Below','Beneath',
    'Between','Beyond','During','Except','From','Inside','Into','Near','Off',
    'Out','Outside','Over','Past','Since','Through','Throughout','Till','Under',
    'Until','Up','Upon','Within','Without','Down','Back','Toward','Towards',
    # Pronouns
    'It','He','She','We','They','I','You','Me','Him','Her','Us','Them',
    'My','Your','His','Its','Our','Their','Mine','Hers','Ours','Theirs',
    'Himself','Herself','Itself','Themselves','Myself','Yourself','Ourselves',
    'Who','Whom','Whose','Which','That',
    # Common titles (false-positive traps)
    'Mr','Mrs','Miss','Ms','Dr','Sir','Lord','Lady','Captain','Inspector',
    'Professor','Colonel','Major','General','Count','Countess','Father',
    'Mother','Brother','Sister','Saint','Rev','Hon','Duke','Duchess',
    # Adjectives (often capitalized at start of sentence)
    'Good','Bad','Big','Small','Little','Old','New','First','Last','Next',
    'Same','Even','Still','Just','Right','Left','High','Low','Long','Short',
    'Young','Great','Large','Own','Few','Less','More','Most','Much','Many',
    'Other','Another','Such','Very','True','False','Dear','Poor','Sure',
    'Free','Open','Full','Dark','Light','Real','Live','Early','Late',
    # Adverbs
    'Again','Only','Always','Never','Ever','Often','Sometimes','Already',
    'Almost','Also','Away','Quite','Rather','Really','Soon','Suddenly',
    'Together','Too','Very','Well','Perhaps','Maybe','Certainly','Probably',
    'Indeed','Else','Enough','Hardly','Mostly','Nearly','Once','Twice',
    # Common false positives for proper names
    'See','Come','Look','Go','Take','Say','Get','Make','Know','Think',
    'Want','Need','Feel','Seem','Become','Keep','Let','Put','Stand',
    'Hear','Try','Ask','Turn','Move','Live','Play','Run','Walk','Call',
    'Help','Stop','Follow','Why','How','When','Where','Can','Could','May',
    'Might','Must','Shall','Should','Will','Would',
    # Indian-English common capitalizations that aren't names
    'Uncle','Aunty','Babu','Sahab','Bhai','Didi','Mam','Madam',
    # Month/day names
    'January','February','March','April','May','June','July','August',
    'September','October','November','December','Monday','Tuesday',
    'Wednesday','Thursday','Friday','Saturday','Sunday',
}



def _detect_genre(text: str) -> str:
    """
    Detect the dominant genre of a text chunk.
    Returns a single genre string matching GENRE_VOCAB keys.
    Uses weighted keyword scoring  -  more specific keywords score higher.
    """
    t = text.lower()
    scores: dict = {}

    _GENRE_KEYWORDS: dict = {
        'mystery/detective': [
            ('detective', 3), ('murder', 3), ('investigation', 3), ('clue', 2),
            ('suspect', 2), ('alibi', 2), ('inspector', 2), ('evidence', 2),
            ('mystery', 2), ('crime', 2), ('corpse', 2), ('whodunit', 3),
            ('forensic', 3), ('interrogat', 2), ('deduction', 3), ('poison', 1),
        ],
        'gothic/horror': [
            ('vampire', 3), ('ghost', 2), ('monster', 2), ('haunted', 3),
            ('horror', 2), ('terror', 2), ('creature', 2), ('curse', 2),
            ('darkness', 1), ('crypt', 3), ('undead', 3), ('supernatural', 3),
            ('dread', 2), ('spectre', 3), ('werewolf', 3), ('coffin', 3),
        ],
        'romance/social': [
            ('marriage', 3), ('proposal', 3), ('darling', 2), ('beloved', 2),
            ('love', 1), ('romance', 2), ('society', 2), ('fortune', 2),
            ('suitor', 3), ('courtship', 3), ('ball', 2), ('estate', 2),
            ('governess', 3), ('dowry', 3), ('engagement', 3),
        ],
        'folk tale / fairy tale': [
            ('once upon', 3), ('witch', 3), ('wizard', 2), ('spell', 2),
            ('enchanted', 3), ('princess', 2), ('dragon', 3), ('fairy', 3),
            ('happily ever after', 3), ('curse', 1), ('magic kingdom', 3),
            ('wicked', 2), ('potion', 2), ('prophecy', 2),
        ],
        'adventure': [
            ('treasure', 2), ('expedition', 3), ('quest', 3), ('battle', 2),
            ('sword', 2), ('pirate', 3), ('voyage', 2), ('shipwreck', 3),
            ('captain', 2), ('jungle', 2), ('desert', 2), ('map', 2),
            ('ambush', 3), ('rescue', 2), ('escaped', 2),
        ],
        'sci-fi / dystopian': [
            ('spaceship', 3), ('robot', 3), ('artificial intelligence', 3),
            ('planet', 2), ('alien', 3), ('future', 2), ('technology', 2),
            ('government control', 3), ('surveillance', 3), ('clone', 3),
            ('mutation', 3), ('dystopia', 3), ('utopia', 3), ('time machine', 3),
            ('science fiction', 3), ('orbit', 2), ('android', 3),
        ],
        'psychological thriller': [
            ('paranoia', 3), ('unreliable', 3), ('obsession', 2), ('manipulation', 3),
            ('psychological', 3), ('breakdown', 2), ('trauma', 2), ('gaslighting', 3),
            ('conspiracy', 2), ('identity', 2), ('the narrator', 2), ('memory', 1),
            ('delusion', 3), ('psychopath', 3), ('stalker', 3),
        ],
        'historical fiction': [
            ('century', 2), ('ancient', 2), ('medieval', 3), ('empire', 2),
            ('kingdom', 2), ('revolution', 2), ('nobleman', 3), ('aristocrat', 3),
            ('plague', 2), ('crusade', 3), ('colonialism', 3), ('the court', 2),
            ('silk road', 3), ('the inquisition', 3), ('roman', 2), ('tudor', 3),
        ],
        'coming-of-age': [
            ('school', 2), ('teenager', 3), ('growing up', 3), ('first love', 3),
            ('college', 2), ('parents', 2), ('rebel', 2), ('identity', 2),
            ('exam', 2), ('friendship', 1), ('heartbreak', 2), ('university', 2),
            ('adolescent', 3), ('youth', 2), ('finding myself', 3),
        ],
        'war / military': [
            ('war', 2), ('battle', 2), ('soldier', 3), ('trench', 3),
            ('rifle', 2), ('regiment', 3), ('general', 2), ('front', 2),
            ('armistice', 3), ('prisoner of war', 3), ('artillery', 3),
            ('platoon', 3), ('bomb', 2), ('march', 2), ('enemy lines', 3),
        ],
        'philosophical / existential': [
            ('meaning of', 3), ('existence', 3), ('absurd', 3), ('nihilism', 3),
            ('consciousness', 3), ('free will', 3), ('morality', 2), ('god', 1),
            ('eternity', 2), ('despair', 2), ('metaphysics', 3), ('the self', 2),
            ('being', 2), ('nothing', 1), ('existential', 3),
        ],
        'magical realism': [
            ('years of solitude', 3), ('yellow butterflies', 3), ('rain of', 2),
            ('ghost at dinner', 3), ('flying carpet', 3), ('miracle', 2),
            ('prophecy fulfilled', 3), ('the dead returned', 3), ('mystical', 2),
            ('the village was struck', 2), ('magical', 2), ('inexplicable', 2),
        ],
    }

    for genre, kw_list in _GENRE_KEYWORDS.items():
        score = sum(weight for kw, weight in kw_list if kw in t)
        if score > 0:
            scores[genre] = score

    if not scores:
        return 'literary fiction'

    # Return genre with highest score
    return max(scores, key=lambda g: scores[g])




def _extract_chars(text: str) -> list:
    """
    Extract character names from English text.
    Returns list of up to 10 most frequent character names.
    Handles: British, American, European, Indian, Japanese title patterns.
    """
    titled = re.findall(
        r'(?:'
        # Western formal titles
        r'Mr\.?|Mrs\.?|Miss|Ms\.?|Dr\.?|Prof\.?|Sir|Lord|Lady|'
        # Military / law
        r'Captain|Inspector|Colonel|Major|General|Lieutenant|Sergeant|Constable|'
        # Noble / royal
        r'Count|Countess|Duke|Duchess|Baron|Baroness|Prince|Princess|King|Queen|'
        # Professional
        r'Professor|Reverend|Rev\.?|Father|Brother|Sister|Nurse|'
        # Indian titles
        r'Sahab|Sahib|Devi|Bhai|Ji|Babu|Seth|Nawab|Raja|Rani|Pandit|Shri|Smt\.?|'
        # Japanese / other
        r'Sensei|Don|Monsieur|Madame|Herr|Frau|Signor|Signora'
        r')\s+'
        r'[A-Z][a-z]{1,}(?:\s+[A-Z][a-z]{1,}){0,2}',
        text
    )

    # Standard capitalized names (2+ occurrences needed after filter)
    singles = re.findall(
        r'\b[A-Z][a-z]{2,}(?:[-\'][A-Z][a-z]+)?(?:\s+[A-Z][a-z]{2,})?\b',
        text
    )

    # Indian-origin names: allow shorter second components (Ram, Raj, etc.)
    indian_names = re.findall(
        r'\b(?:Ram|Raj|Dev|Amar|Hari|Shyam|Gopal|Mohan|Krishna|Arjun|'
        r'Ravi|Suresh|Mahesh|Ganesh|Prem|Anand|Vijay|Arun|Kiran|Meera|'
        r'Radha|Sita|Geeta|Lata|Priya|Asha|Rekha|Sunita|Kavita|Deepa|'
        r'Ahmed|Ali|Hassan|Fatima|Zainab|Ibrahim|Omar|Aisha|Amina|'
        r'Ramesh|Suresh|Dinesh|Naresh|Mukesh|Lokesh|Umesh|Nilesh)\b',
        text
    )

    counts = Counter(singles + titled + indian_names)

    # Filter: freq >= 2, not stop word, last word > 2 chars, not all caps
    result = [
        n for n, c in counts.most_common(25)
        if c >= 2
        and n not in _STOP
        and len(n.split()[-1]) > 2
        and not n.isupper()
        and len(n) > 2
    ]

    return result[:10]

# def _extract_setting(text):
    m = re.search(
        r'(?:at|in|inside|within|outside|near|entered|through|arrived at|left)\s+'
        r'(?:the\s+)?([A-Z][A-Za-z\s]{2,30}'
        r'(?:Street|Road|House|Hall|Room|Study|Garden|Park|Inn|Hotel|'
        r'Station|Office|Club|Square|Lane|Bridge|Castle|Manor|Lodge|'
        r'Village|Court|Tower|Library|Cottage|Drawing.?Room|Sitting.?Room))',
        text)
    return m.group(1).strip()[:60] if m else ''

# ── Setting Extraction v6.0 ─────────────────────────────────────────────────
# Expanded: natural settings, fantasy, Indian, modern, global
def _extract_setting(text: str) -> str:
    """
    Extract the current physical setting/location from English text.
    Covers: British, Indian, fantasy, modern, natural, global settings.
    Returns location string (max 80 chars) or empty string.
    """
    # Modern & global location types
    _LOC_TYPES = (
        r'Street|Road|Avenue|Lane|Boulevard|Drive|'
        r'House|Hall|Room|Study|Office|Flat|Apartment|Building|'
        r'Garden|Park|Square|Court|Yard|Terrace|'
        r'Inn|Hotel|Restaurant|Cafe|Bar|Club|Pub|'
        r'Station|Airport|Platform|Terminal|'
        r'Castle|Manor|Lodge|Estate|Mansion|Palace|Haveli|Fort|'
        r'Village|Town|City|Colony|Mohalla|Bazaar|Market|'
        r'Library|School|College|University|Hospital|'
        r'Tower|Bridge|Temple|Church|Mosque|Mandir|Masjid|'
        r'Cottage|Bungalow|Farmhouse|Ranch|Camp|'
        r'Drawing.?Room|Sitting.?Room|Dining.?Room|Living.?Room|'
        r'Forest|Jungle|Mountain|Valley|River|Lake|Shore|Beach|Desert|'
        r'Island|Cave|Dungeon|Basement|Attic|Cellar|'
        r'Courtyard|Balcony|Corridor|Passage|Staircase|'
        r'Laboratory|Factory|Workshop|Warehouse|Prison|Jail'
    )

    _MOVEMENT_WORDS = (
        r'at|in|inside|within|outside|near|entered|through|arrived at|'
        r'left|walked into|stepped into|stood in|sat in|ran to|returned to|'
        r'escaped from|found himself in|found herself in'
    )

    pattern = re.compile(
        rf'(?:{_MOVEMENT_WORDS})\s+'
        rf'(?:the\s+|a\s+|an\s+)?'
        rf'([A-Z][A-Za-z\s]{{2,35}}'
        rf'(?:{_LOC_TYPES}))',
        re.IGNORECASE
    )

    # Find all matches and return the most specific (longest)
    matches = pattern.findall(text)
    if matches:
        best = max(matches, key=lambda m: len(m.strip()))
        return best.strip()[:80]

    # Fallback: named places (e.g. "London", "Malgudi", "Baker Street")
    named = re.search(
        r'\b(?:in|at|near|outside)\s+([A-Z][a-z]{3,}(?:\s+[A-Z][a-z]{2,})?)\b',
        text
    )
    if named:
        candidate = named.group(1).strip()
        if candidate not in _STOP and len(candidate) > 3:
            return candidate[:60]

    return ''


def _english_fallback(text):
    sents = re.split(r'(?<=[.!?])\s+', text.strip())
    for s in sents:
        if len(s.split()) > 6: return s.strip()[:120]
    return ' '.join(text.split()[:25])

# ── Speech Pattern Dictionary v6.0 ───────────────────────────────────────────
# 14 styles · 10-14 keywords each · covers classic + modern + Indian speech
_SPEECH_PATTERNS = {
    # Classic styles (expanded)
    'formal': [
        'I beg your pardon', 'I must inform', 'I am afraid', 'I assure you',
        'It is my duty', 'permit me', 'I regret to say', 'I have the honour',
        'allow me to', 'with respect', 'I must confess', 'it pains me to',
        'I am obliged', 'I take the liberty',
    ],
    'analytical': [
        'the evidence', 'it is clear', 'I observe', 'I deduce', 'therefore',
        'logically', 'it follows', 'one must conclude', 'the facts indicate',
        'upon reflection', 'my deduction', 'I have noted', 'the data',
        'I calculate', 'it is evident', 'elementary',
    ],
    'anxious': [
        'I fear', 'what if', 'surely not', 'I cannot bear', 'what shall',
        'I dread', 'I am terrified', 'what will become', 'I dare not',
        'I tremble', 'God help us', 'I shudder', 'I am afraid that',
        'what are we to do', 'I cannot think',
    ],
    'commanding': [
        'you will', 'do it at once', 'I demand', 'see to it', 'at once',
        'immediately', 'I insist', 'you must', 'do not question',
        'I order you', 'get out', 'do as I say', 'I warn you',
        'enough', 'silence',
    ],
    'gentle': [
        'my dear', 'pray', 'I hope', 'would you', 'I beg', 'if you please',
        'if I may', 'thank you', 'how kind', 'please forgive',
        'I am grateful', 'you are too good', 'do not trouble yourself',
        'perhaps', 'I wonder if',
    ],
    'sarcastic': [
        'how charming', 'how delightful', 'I am sure', 'how very',
        'how convenient', 'how wonderful', 'of course', 'naturally',
        'how surprising', 'how extraordinary', 'no doubt', 'I suppose',
        'how amusing', 'how perfectly',
    ],
    'lyrical': [
        'I wonder', 'how strange', 'the light', 'the silence', 'the heart',
        'I feel as though', 'there is something', 'the beauty',
        'I cannot explain', 'it seemed to me', 'a kind of', 'the soul',
        'I am moved', 'how beautiful',
    ],

    # Additional modern / diverse styles (v6.0 new)
    'philosophical': [
        'what is the meaning', 'one must consider', 'the question is',
        'I have been thinking', 'perhaps life', 'what does it mean',
        'is it not strange', 'can we truly', 'the nature of',
        'what we call', 'I ask myself', 'in the end',
        'freedom means', 'to exist is',
    ],
    'manipulative': [
        'you know I would never', 'trust me', 'between us',
        'no one understands you like', 'you owe me', 'after everything',
        'I only say this because', 'for your own good',
        'you are being unreasonable', 'why do you always',
        'I sacrifice so much', 'after all I have done',
        'you are too sensitive', 'I am only trying',
    ],
    'playful': [
        'come now', 'you cannot be serious', 'surely you jest',
        'how perfectly absurd', 'let us not', 'oh come',
        'you are dreadful', 'I refuse to believe',
        'how ridiculous', 'you are too much', 'what fun',
        'shall we not', 'do not be so',
    ],
    'resigned': [
        'it does not matter', 'what can one do', 'it is of no use',
        'I have accepted', 'there is nothing to be done',
        'it was always going to', 'I expected no less',
        'I am past caring', 'let it be', 'as you wish',
        'I am tired of', 'it makes no difference',
        'I no longer', 'what is the point',
    ],
    'passionate': [
        'I love you', 'I cannot live without', 'you are everything',
        'I would die for', 'nothing matters but', 'I have never felt',
        'I am consumed', 'you must know how I feel',
        'my heart', 'I adore you', 'without you',
        'I have loved you', 'you are the only',
        'I cannot stop thinking',
    ],
    'cold / detached': [
        'it is nothing', 'I feel nothing', 'it does not concern me',
        'as you like', 'I do not care', 'it is irrelevant',
        'I am not interested', 'do as you please',
        'I have no feelings on', 'it is unimportant',
        'whatever you think', 'I am indifferent',
        'so be it', 'it is of no importance',
    ],
    'humorous': [
        'you must admit', 'I confess', 'I cannot help laughing',
        'the absurdity', 'it strikes me as', 'now really',
        'one has to laugh', 'I find it rather',
        'it is too funny', 'you have to see the humour',
        'I mean no offence', 'the irony is',
        'I ask you', 'what can I say',
    ],
}

def _infer_speech_style(char_name: str, text: str) -> str:
    """
    Infer a character's speech style from their dialogue in the text.
    Returns comma-joined top-3 styles or 'natural' if inconclusive.
    v6.0: handles both standard and curly quotes, fallback to name-only search.
    """
    last_name = re.escape(char_name.split()[-1])

    # Try dialogue-before-attribution pattern (most reliable)
    pattern_before = re.compile(
        r'["\u201c\u201d\u2018\u2019][^\u201c\u201d"\']{5,300}["\u201c\u201d\u2018\u2019]'
        r'[\s,]*(?:said|replied|answered|asked|whispered|shouted|cried|muttered|declared|'
        r'exclaimed|insisted|demanded|began|continued|added|interrupted)\s+' + last_name,
        re.IGNORECASE
    )
    # Try attribution-after-dialogue
    pattern_after = re.compile(
        r'' + last_name +
        r'\s+(?:said|replied|answered|asked|whispered|shouted|cried|muttered|declared|'
        r'exclaimed|insisted|demanded|began|continued|added|interrupted)[,\s]+'
        r'["\u201c\u201d\u2018\u2019][^\u201c\u201d"\']{5,300}["\u201c\u201d\u2018\u2019]',
        re.IGNORECASE
    )

    matches = pattern_before.findall(text) + pattern_after.findall(text)
    dialogue_text = ' '.join(matches).lower()

    # Fallback: grab any quoted text near character's name
    if not dialogue_text:
        fallback = re.compile(
            r'["\u201c\u201d][^\u201c\u201d"]{5,200}["\u201c\u201d].*?' + last_name,
            re.IGNORECASE
        )
        dialogue_text = ' '.join(fallback.findall(text)).lower()

    if not dialogue_text:
        return ''

    scores = {
        style: sum(1 for kw in kws if kw.lower() in dialogue_text)
        for style, kws in _SPEECH_PATTERNS.items()
    }
    detected = sorted(
        [s for s, sc in scores.items() if sc > 0],
        key=lambda s: scores[s],
        reverse=True
    )
    return ', '.join(detected[:3]) if detected else 'natural'



# def build_context_prompt():
    """v5.0 — Compressed context, capped at ~800 chars. Model never overloaded."""
    s = STORY_STATE
    if s['chunk_count'] == 0 and not s['characters'] and not s['established_vocab']:
        return ''
    parts = ['=== STORY CONTEXT (model ke liye sirf — output mein mat daalo) ===']
    info = []
    if s.get('book_title'): info.append(f"Book: {s['book_title']}")
    if s.get('genre'):      info.append(f"Genre: {s['genre']}")
    if s.get('author'):     info.append(f"Author: {s['author']}")
    if info: parts.append(' | '.join(info))
    # Author DNA — single line
    if s.get('author_dna'):
        dna_line = s['author_dna'][:200]
        parts.append(f"AUTHOR STYLE: {dna_line}")
    # Tone anchor
    if s.get('tone_anchor'):
        parts.append(f"TONE ANCHOR (kabhi mat badlo): {s['tone_anchor']}")
    # Characters — compact
    if s['characters']:
        parts.append('CHARACTERS:')
        for name, info in list(s['characters'].items())[:8]:
            style = s['character_speech_styles'].get(name, '')
            line = f"  • {name} ({info['role']}) | {info['address']}"
            if style: line += f" | {style}"
            parts.append(line)
    # Vocab — top 10 only
    if s.get('established_vocab'):
        vocab_items = list(s['established_vocab'].items())[:10]
        parts.append('VOCAB: ' + ' · '.join(f'"{e}"→"{h}"' for e,h in vocab_items))
    # Rolling 3-event summary
    if s['story_so_far']:
        parts.append('RECENT EVENTS:')
        for entry in s['story_so_far']:
            txt = entry['summary'] if isinstance(entry, dict) else str(entry)
            parts.append(f"  C{entry.get('chunk','?') if isinstance(entry,dict) else '?'}: {txt[:80]}")
    parts.append('=== CONTEXT END ===')
    raw = '\n'.join(parts)
    # Hard cap at 800 chars
    return raw[:800] if len(raw) > 800 else raw

# ── update_story_state v6.0 ──────────────────────────────────────────────────
# What's new vs v5.0:
#   • Auto-loads AuthorDNA from STORY_STATE['author_key'] on chunk 1 (no manual call needed)
#   • Genre detection runs on first 3 chunks and takes the majority vote → far more
#     reliable for novels that open in media res (Dickens, Dumas, Twain, etc.)
#   • Uses hinglish_chunk to mine new E→H vocab pairs automatically
#   • Rolling window expanded to 5 entries (better for long novels: Tolstoy, Hugo, Dickens)
#   • Setting is never overwritten with an empty string
#   • Character cap raised to 12 (needed for ensemble novels: War & Peace, Pickwick, etc.)
#   • Passive tone-drift guard: if scene_type != DAILY_LIFE and author has a tone_anchor,
#     appends a tone-drift note so build_context_prompt can surface it

def update_story_state(source_chunk, output_chunk, chunk_idx,
                       scene_type='DAILY_LIFE', plot_note_en=''):
    global STORY_STATE

    # ── 1. Author DNA bootstrap (run once, on chunk 1) ────────────────────────
    # If STORY_STATE was pre-seeded with an author_key (e.g. 'kafka', 'austen')
    # but AuthorDNA hasn't been loaded yet, load it now automatically.
    if chunk_idx == 1:
        ak = STORY_STATE.get('author_key', '')
        if ak and not STORY_STATE.get('tone_anchor'):
            load_author_dna(ak)

    # ── 2. Genre detection — majority vote over first 3 chunks ───────────────
    # We store per-chunk genre votes in a temporary key; commit after 3 votes.
    if not STORY_STATE.get('genre'):
        _votes = STORY_STATE.setdefault('_genre_votes', [])
        _votes.append(_detect_genre(source_chunk))
        if len(_votes) >= 3:
            # Majority vote (or most common)
            winner = Counter(_votes).most_common(1)[0][0]
            set_genre(winner)
            STORY_STATE.pop('_genre_votes', None)
    elif chunk_idx == 1 and not STORY_STATE.get('_genre_votes'):
        # genre pre-set by STORY_STATE init — just ensure vocab is loaded
        set_genre(STORY_STATE['genre'])

    # ── 3. Character extraction — cap at 12 for ensemble novels ──────────────
    for name in _extract_chars(source_chunk):
        if name not in STORY_STATE['characters']:
            STORY_STATE['characters'][name] = {
                'role': 'character', 'address': 'aap', 'notes': ''
            }
    # Trim to 12 most-seen (existing order is already frequency-ranked)
    if len(STORY_STATE['characters']) > 12:
        keys = list(STORY_STATE['characters'].keys())[:12]
        STORY_STATE['characters'] = {k: STORY_STATE['characters'][k] for k in keys}

    # ── 4. Speech-style inference (only for chars not yet styled) ─────────────
    for name in list(STORY_STATE['characters'].keys()):
        if not STORY_STATE['character_speech_styles'].get(name):
            inferred = _infer_speech_style(name, source_chunk)
            if inferred and inferred != 'natural':
                STORY_STATE['character_speech_styles'][name] = inferred

    # ── 5. Setting — never overwrite with empty ───────────────────────────────
    s = _extract_setting(source_chunk)
    if s:
        STORY_STATE['current_setting'] = s

    # ── 6. Dominant tone (with tone-drift guard) ──────────────────────────────
    STORY_STATE['dominant_tone'] = scene_type
    # If this chunk is a climax/conflict and the author has a tone_anchor, flag it
    if scene_type in ('CLIMAX', 'CONFLICT', 'EMOTIONAL') and STORY_STATE.get('tone_anchor'):
        STORY_STATE['_tone_drift_flag'] = (
            f"High-intensity scene ({scene_type}) — stay in author register: "
            f"{STORY_STATE['tone_anchor']}"
        )
    else:
        STORY_STATE.pop('_tone_drift_flag', None)

    # ── 7. Vocab mining from hinglish_chunk ───────────────────────────────────
    # Pattern: English word/phrase followed by its Hinglish in parens, or
    # common "X = Y" / "X yaani Y" patterns that the 2-pass sometimes produces.
    # Only add if not already in vocab and not trivially identical.
    _VOCAB_PATTERNS = [
        r'\b([A-Za-z][A-Za-z\s]{1,25})\s*[\(（]\s*([^\)）]{2,30})\s*[\)）]',  # word (hinglish)
        r'\b([A-Za-z][A-Za-z\s]{1,25})\s+yaani\s+([^\.,;]{2,30})',             # word yaani X
    ]
    for pat in _VOCAB_PATTERNS:
        for eng, hin in re.findall(pat, output_chunk or ''):
            eng = eng.strip(); hin = hin.strip()
            if (eng and hin
                    and eng.lower() != hin.lower()
                    and eng not in STORY_STATE['established_vocab']
                    and len(eng) > 2):
                STORY_STATE['established_vocab'][eng] = hin

    # ── 8. Rolling 5-entry plot summary (was 3 in v5.0) ──────────────────────
    summary = (plot_note_en.strip()
               if plot_note_en and len(plot_note_en.split()) > 4
               else _english_fallback(source_chunk))
    STORY_STATE['story_so_far'].append({'chunk': chunk_idx, 'summary': summary})
    if len(STORY_STATE['story_so_far']) > 5:
        STORY_STATE['story_so_far'] = STORY_STATE['story_so_far'][-5:]

    STORY_STATE['chunk_count'] = chunk_idx


# ── build_context_prompt v6.0 ────────────────────────────────────────────────
# What's new vs v5.0:
#   • Hard cap raised to 1 200 chars (ensemble/long novels need it)
#   • Includes current_setting (was missing in v5.0)
#   • Includes chunk progress X/total → model understands narrative arc position
#   • Character notes (first 55 chars) shown for characters that have them
#   • Tone-drift flag surfaced when active (CLIMAX/CONFLICT scenes)
#   • Vocab: skips entries where eng == hin (no-op terms) · shows top 12 not 10
#   • Author DNA: smarter truncation — takes first 2 sentences (≤220 chars)
#     instead of a blind 200-char slice that could cut mid-word

def build_context_prompt():
    """v6.0 — Context block for Hindi→Hinglish reformulation. Cap: ~1200 chars."""
    s = STORY_STATE

    # Return empty if state is still cold
    if s['chunk_count'] == 0 and not s['characters'] and not s['established_vocab']:
        return ''

    parts = ['=== STORY CONTEXT (model ke liye sirf — output mein mat daalo) ===']

    # ── Identity line ─────────────────────────────────────────────────────────
    info = []
    if s.get('book_title'): info.append(f"Book: {s['book_title']}")
    if s.get('genre'):      info.append(f"Genre: {s['genre']}")
    if s.get('author'):     info.append(f"Author: {s['author']}")
    # Chunk progress — helps model feel narrative arc (early / mid / climax)
    total = s.get('total_chunks', 0)
    if total:
        pct = int(s['chunk_count'] / total * 100)
        info.append(f"Progress: {s['chunk_count']}/{total} ({pct}%)")
    else:
        info.append(f"Chunk: {s['chunk_count']}")
    if info:
        parts.append(' | '.join(info))

    # ── Author DNA — first 2 sentences, max 220 chars ────────────────────────
    if s.get('author_dna'):
        dna = s['author_dna']
        # Split on period+space to get sentences, take first 2
        sentences = re.split(r'\.\s+', dna)
        dna_short = '. '.join(sentences[:2])
        if not dna_short.endswith('.'): dna_short += '.'
        parts.append(f"AUTHOR STYLE: {dna_short[:220]}")

    # ── Tone anchor (always show — it's the single most important guardrail) ──
    if s.get('tone_anchor'):
        parts.append(f"TONE (kabhi mat badlo): {s['tone_anchor']}")

    # ── Tone-drift warning (only active during CLIMAX/CONFLICT scenes) ────────
    if s.get('_tone_drift_flag'):
        parts.append(f"⚠ {s['_tone_drift_flag']}")

    # ── Current setting ───────────────────────────────────────────────────────
    if s.get('current_setting'):
        parts.append(f"SETTING: {s['current_setting']}")

    # ── Characters — up to 10, with notes if available ───────────────────────
    if s['characters']:
        parts.append('CHARS:')
        for name, cinfo in list(s['characters'].items())[:10]:
            style = s['character_speech_styles'].get(name, '')
            line = f"  {name} ({cinfo['role']}) [{cinfo['address']}]"
            if style:
                line += f" — {style}"
            notes = (cinfo.get('notes') or '').strip()
            if notes:
                line += f" | {notes[:55]}"
            parts.append(line)

    # ── Vocab — top 12, skip no-op entries ───────────────────────────────────
    if s.get('established_vocab'):
        vocab_items = [
            (e, h) for e, h in list(s['established_vocab'].items())
            if e.lower().strip() != h.lower().strip()
        ][:12]
        if vocab_items:
            parts.append('VOCAB: ' + ' · '.join(f'"{e}"→"{h}"' for e, h in vocab_items))

    # ── Rolling 5-event summary (was 3) ──────────────────────────────────────
    if s['story_so_far']:
        parts.append('RECENT EVENTS:')
        for entry in s['story_so_far']:
            if isinstance(entry, dict):
                txt = entry.get('summary', '')
                cn  = entry.get('chunk', '?')
            else:
                txt = str(entry); cn = '?'
            parts.append(f"  C{cn}: {txt[:85]}")

    parts.append('=== CONTEXT END ===')
    raw = '\n'.join(parts)

    # Hard cap at 1200 chars — still safe for 8192 token context
    return raw[:1200] if len(raw) > 1200 else raw

def save_state(filepath):
    try:
        with open(filepath, 'w', encoding='utf-8') as f:
            json.dump(STORY_STATE, f, ensure_ascii=False, indent=2)
    except Exception as e:
        print(f'[!] State save: {e}')


def load_state(filepath):
    global STORY_STATE
    try:
        if os.path.exists(filepath):
            with open(filepath, 'r', encoding='utf-8') as f:
                loaded = json.load(f)
            
            # Ensure all required keys exist with proper defaults
            required_keys = {
                'book_title': '',
                'author_key': '',
                'genre': '',
                'author': '',
                'author_dna': '',
                'tone_anchor': '',
                'characters': {},
                'character_speech_styles': {},
                'current_setting': '',
                'dominant_tone': 'DAILY_LIFE',
                'established_vocab': {},
                'story_so_far': [],
                'chunk_count': 0,
                'total_chunks': 0,
                'resume_from_chunk': 0,
            }
            
            for key, default_val in required_keys.items():
                if key not in loaded:
                    loaded[key] = default_val
            
            STORY_STATE = loaded
            print(f'[OK] State loaded — chunk {STORY_STATE["chunk_count"]}, '
                  f'{len(STORY_STATE["characters"])} chars, '
                  f'{len(STORY_STATE["character_speech_styles"])} speech styles')
            return True
    except Exception as e:
        print(f'[!] load_state: {e}')
    return False


def _load_latest_state(output_dir='./translation_output'):
    import glob
    files = sorted(glob.glob(f'{output_dir}/state_*.json'))
    if files: return load_state(files[-1])
    return False


def reset_for_new_book(title='', genre='', author=''):
    global STORY_STATE
    STORY_STATE = {
        'book_title': title, 'author_key': '', 'genre': '', 'author': author,
        'author_dna': '', 'tone_anchor': '',
        'characters': {}, 'character_speech_styles': {},
        'current_setting': '', 'dominant_tone': 'DAILY_LIFE',
        'established_vocab': {}, 'story_so_far': [],
        'chunk_count': 0, 'total_chunks': 0, 'resume_from_chunk': 0,
    }
    if genre: set_genre(genre)
    msg = f'[OK] Reset → "{title}"' if title else '[OK] State reset.'
    print(msg + (f' | Genre: {genre}' if genre else '') + (f' | Author: {author}' if author else ''))


def load_author_dna(author_key):
    """v5.0: Load pre-built AuthorDNA profile."""
    key = author_key.lower().strip()
    if key not in AUTHOR_DNA:
        print(f'[WARN] Author "{key}" not in AuthorDNA. Available: {list(AUTHOR_DNA.keys())}')
        return
    
    tone_anchor, dna_desc, genre_hint, extra_vocab = AUTHOR_DNA[key]
    STORY_STATE['author_key'] = key  # ← ADDED THIS LINE
    STORY_STATE['tone_anchor'] = tone_anchor
    STORY_STATE['author_dna']  = dna_desc
    STORY_STATE['author']      = key.capitalize()
    
    if not STORY_STATE['genre'] and genre_hint:
        set_genre(genre_hint)
    
    for eng, hin in extra_vocab.items():
        if eng not in STORY_STATE['established_vocab']:
            STORY_STATE['established_vocab'][eng] = hin
    
    print(f'[OK] AuthorDNA loaded: {key.upper()}')
    print(f'     tone_anchor: {tone_anchor}')
    print(f'     genre: {STORY_STATE["genre"]}')
    print(f'     vocab: {len(extra_vocab)} entries loaded')


def set_genre(genre):
    STORY_STATE['genre'] = genre
    vocab = GENRE_VOCAB.get(genre, {})
    for eng, hin in vocab.items():
        if eng not in STORY_STATE['established_vocab']:
            STORY_STATE['established_vocab'][eng] = hin
    print(f'[OK] Genre: {genre}' + (f' | {len(vocab)} vocab defaults' if vocab else ''))

def set_book_title(title): STORY_STATE['book_title'] = title; print(f'[OK] Book: "{title}"')

def add_vocab_entry(english, hinglish):
    if english.strip().lower() == hinglish.strip().lower():
        print(f'[SKIP] Same-value: "{english}"'); return
    STORY_STATE['established_vocab'][english] = hinglish
    print(f'[OK] Vocab: "{english}" → "{hinglish}"')

def update_character(name, role=None, address=None, notes=None, speech_style=None):
    if name not in STORY_STATE['characters']:
        STORY_STATE['characters'][name] = {'role':'character','address':'aap','notes':''}
    e = STORY_STATE['characters'][name]
    if role:    e['role']    = role
    if address: e['address'] = address
    if notes:   e['notes']   = notes
    if speech_style: STORY_STATE['character_speech_styles'][name] = speech_style
    print(f'[OK] Char: {name} → {e}' + (f' | speech: {speech_style}' if speech_style else ''))

def print_state_summary():
    s = STORY_STATE
    print('\n=== STORY STATE v5.0 ===')
    if s.get('book_title'):  print(f'  Book    : {s["book_title"]}')
    if s.get('author_key'):  print(f'  AuthKey : {s["author_key"]}')
    if s.get('author'):      print(f'  Author  : {s["author"]}')
    if s.get('genre'):       print(f'  Genre   : {s["genre"]}')
    if s.get('tone_anchor'): print(f'  Tone    : {s["tone_anchor"]}')
    if s.get('current_setting'): print(f'  Setting : {s["current_setting"]}')
    print(f'  Chunks  : {s["chunk_count"]} / {s["total_chunks"]}')
    print(f'  Chars   : {len(s["characters"])} | Speech: {len(s["character_speech_styles"])}')
    print(f'  Vocab   : {len(s["established_vocab"])} entries')
    if s['story_so_far']:
        print(f'  Last 3  : ' + ' → '.join(e["summary"][:40] for e in s["story_so_far"] if isinstance(e,dict)))
    print('========================')


print('[OK] Story State Engine v6.0 loaded — Hindi→Hinglish v1.0')
print('  AuthorDNA profiles:', ', '.join(AUTHOR_DNA.keys()))
print('  Features: rolling 5-chunk context · context_compression · tone_anchor · resume-from-chunk')
print('  Mode: Hindi(Latin) → Hinglish(Latin) Reformulation')
````

## 🔩 Step 6 — Load Translation Engine v1.0

**1-Pass Reformulation Architecture:**
1. **Single Pass** — Hindi(Latin) → natural Hinglish with English code-switching (temp=0.55)

The source is already in Hindi (Latin script). This is a **register shift** — formal/literary Hindi → modern conversational Hinglish. One well-prompted pass is sufficient.

**Quality layers:**
- ToneGuard Pro — no foul words, respect system enforced
- Anti-Hallucination — zero-fabrication, source-anchoring
- Loop detection — streaming LiveLoopMonitor + post-hoc salvage
- Foul word filter — PROD-grade commercial output
- QA pipeline — Devanagari check, Hindi-heavy check, English-heavy check


````python
# ════════════════════════════════════════════════════════════════════
# ARCHITECTURE: 3-LAYER SEPARATION  (Fix E)
# Layer 1 — Generate  : raw model call only
# Layer 2 — Validate  : ALL signals checked, returns structured dict
# Layer 3 — Decide    : clean action decision from validation dict
#
# WHY: previously blended → hard to debug which signal triggered which
# action, and retry logic was tangled with generation logic.
# ════════════════════════════════════════════════════════════════════

def _validate_output(translation: str, opts: dict, orig_text: str = '', prev_tail: str = '') -> dict:
    """
    Layer 2 — Validate.
    Returns a structured dict of ALL signals. Nothing decides here — just detects.

    Keys:
      loop         : (bool detected, int severity 0-3)
      line_repeat  : (bool detected, int count)
      devanagari   : (bool detected, str detail)
      token_ceil   : (bool detected, str detail)
      eng_heavy    : (bool detected, float ratio)
      length_ratio : (bool ok, float ratio)  — hallucination/truncation guard
      overlap_dup  : (bool ok, str detail)
    """
    result = {}
    twords = translation.split() if translation else []

    # ── Loop detection (n-gram + sentence uniqueness) ────────────────────────
    is_loop, loop_sev = _detect_hard_loop(translation)
    result['loop'] = (is_loop, loop_sev)

    # ── Line-level repetition ────────────────────────────────────────────────
    is_rep, rep_count = detect_repetition(translation)
    result['line_repeat'] = (is_rep, rep_count)

    # ── Devanagari leak ──────────────────────────────────────────────────────
    deva = re.findall(r'[\u0900-\u097F]+', translation)
    result['devanagari'] = (bool(deva), f'{len(deva)} segment(s)' if deva else 'Clean')

    # ── Token ceiling (FIX C: char-based approximation) ─────────────────────
    tok_ceil = opts.get('num_predict', 1400)
    near_ceil = len(translation) / 4 > tok_ceil
    result['token_ceil'] = (near_ceil, f'{len(translation)//4} approx-tokens / {tok_ceil} limit')

    # ── English ratio (style, NOT structural) ───────────────────────────────
    eng_c = sum(1 for w in twords if re.match(r'^[a-z]{4,}$', w) and w not in _H_ALLOW)
    eng_ratio = eng_c / max(len(twords), 1)
    result['eng_heavy'] = (eng_ratio > 0.55, eng_ratio)

    # ── Length ratio guard (hallucination / truncation) ──────────────────────
    if orig_text:
        src_w = len(orig_text.split())
        out_w = len(twords)
        ratio = out_w / max(src_w, 1)
        ok    = 0.4 <= ratio <= 2.5
        result['length_ratio'] = (ok, ratio)
    else:
        result['length_ratio'] = (True, 1.0)

    # ── Overlap duplicate check ──────────────────────────────────────────────
    if prev_tail and len(prev_tail) > 20:
        pw  = set(w.lower() for w in prev_tail.split()[-15:])
        cw  = set(w.lower() for w in twords[:15])
        sim = len(pw & cw) / max(len(pw), 1)
        result['overlap_dup'] = (sim <= 0.6, f'sim={sim:.0%}')
    else:
        result['overlap_dup'] = (True, 'n/a')

    return result

def _decide_action(v: dict) -> tuple:
    """
    Layer 3 — Decide.
    Returns one of:
      ('accept',        None)        — clean, use immediately
      ('modify',        'eng_heavy') — style issue only → soft English fix
      ('salvage_first', int)         — loop detected → salvage clean prefix first,
                                       then optional tail retry (NOT full retry)
      ('retry',         int)         — non-loop structural failure (Devanagari,
                                       empty, ceiling-only) → retry full chunk

    Decision tree (LOCKED — do not reorder):
      1. Empty output              → retry L1
      2. Line repetition           → salvage_first sev=3 (15x line = hardest loop)
      3. Loop + ceiling            → salvage_first sev=3 (saturation confirmed)
      4. Loop detected             → salvage_first at loop_sev
      5. Token ceiling alone       → retry L1  (no loop yet, but pressure)
      6. Devanagari                → retry L1
      7. English-heavy ONLY        → modify    (never retry for style)
      8. Clean                     → accept

    WHY salvage_first for loops (not retry):
      Your real output: first 70-80% = clean, last 20-30% = loop.
      Retrying the full chunk throws away the good 70%. Expensive + unstable.
      Salvage-first keeps the clean prefix, then retries ONLY the missing tail.
      This is always cheaper, usually better quality, never worse.
    """
    is_loop,   loop_sev  = v['loop']
    is_rep,    rep_count = v['line_repeat']
    has_deva,  _         = v['devanagari']
    near_ceil, _         = v['token_ceil']
    eng_heavy, _         = v['eng_heavy']

    # 1. Empty
    if not any(v.values()):
        return ('retry', 1)

    # 2. Line repetition (15x repeated line = aggressive loop → salvage first)
    if is_rep:
        return ('salvage_first', 3)

    # 3. Loop + ceiling = saturation → salvage first (retrying would likely loop again)
    if is_loop and near_ceil:
        return ('salvage_first', 3)

    # 4. Loop detected → salvage first at detected severity
    if is_loop:
        return ('salvage_first', min(loop_sev, 3))

    # 5. Token ceiling alone (no loop yet) → retry L1 (catch it early)
    if near_ceil:
        return ('retry', 1)

    # 6. Devanagari leak → retry L1
    if has_deva:
        return ('retry', 1)

    # 7. English-heavy only (style, not structural) → soft fix, never retry
    if eng_heavy:
        return ('modify', 'eng_heavy')

    # 8. Clean
    return ('accept', None)
````

````python
import os, sys, json, time, warnings, re, threading
from pathlib import Path
from datetime import datetime
warnings.filterwarnings('ignore')
from IPython.display import display, HTML, clear_output

# ═══════════════════════════════════════════════════════════════════
# JARVIS DISPLAY HELPERS
# ═══════════════════════════════════════════════════════════════════
def _j_bar(pct, w=180, fg='#c0392b', bg='#1a0505'):
    filled = max(0, min(int(w*(pct or 0)/100), w))
    return ('<div style="background:'+bg+';border:1px solid #2a1010;border-radius:3px;'
            'width:'+str(w)+'px;height:12px;display:inline-block;vertical-align:middle;">'
            '<div style="background:linear-gradient(90deg,#7b0000,'+fg+');'
            'width:'+str(filled)+'px;height:100%;border-radius:3px;"></div></div>')

def _j_col(u): return '#555' if u is None else ('#4CAF50' if u<40 else ('#FFD700' if u<75 else '#c0392b'))
def _j_fmt(s):
    s=max(0,int(s)); h,rem=divmod(s,3600); m,sec=divmod(rem,60)
    return (str(h)+'h '+str(m).zfill(2)+'m') if h else (str(m).zfill(2)+'m '+str(sec).zfill(2)+'s')
def _j_spark(vals):
    if not vals: return ''
    blk=' '+''.join(chr(c) for c in [9601,9602,9603,9604,9605,9606,9607,9608])
    mn,mx=min(vals),max(vals); rng=mx-mn or 1
    return ''.join(blk[min(8,int((v-mn)/rng*8))] for v in vals[-60:])
def _j_gpu():
    import subprocess
    try:
        out=subprocess.check_output(['nvidia-smi','--query-gpu=utilization.gpu,memory.used,memory.total,temperature.gpu','--format=csv,noheader,nounits'],stderr=subprocess.DEVNULL,timeout=3).decode().strip().split(',')
        return float(out[0]),float(out[1])/1024,float(out[2])/1024,float(out[3])
    except: return None,None,None,None

def _jarvis_header(title, subtitle=''):
    display(HTML('<div style="background:linear-gradient(135deg,#0a0a0f,#1a0505);border:2px solid #c0392b;border-radius:8px;padding:14px 20px;font-family:Courier New,monospace;margin:6px 0;">'
                '<div style="color:#c0392b;font-size:1.2em;font-weight:bold;letter-spacing:3px;text-shadow:0 0 10px #c0392b;">'+title+'</div>'
                '<div style="color:#FFD700;font-size:0.8em;margin-top:4px;">'+subtitle+'</div>'
                '</div>'))
def _jarvis_ok(msg):
    display(HTML('<div style="font-family:Courier New,monospace;background:#0a0f0a;border-left:4px solid #4CAF50;padding:5px 14px;margin:2px 0;color:#4CAF50;font-size:0.88em;">[OK] '+str(msg)+'</div>'))
def _jarvis_info(msg, color='#FFD700'):
    display(HTML('<div style="font-family:Courier New,monospace;background:#0a0a0f;border-left:4px solid '+color+';padding:5px 14px;margin:2px 0;color:'+color+';font-size:0.88em;">[·] '+str(msg)+'</div>'))
def _jarvis_warn(msg):
    display(HTML('<div style="font-family:Courier New,monospace;background:#1a0505;border-left:4px solid #e74c3c;padding:5px 14px;margin:2px 0;color:#e74c3c;font-size:0.88em;">[!] '+str(msg)+'</div>'))


def _jarvis_chunk_dashboard(
        i, total, scene, in_words, in_chars, out_chars,
        chunk_t, total_t, eta_s, ctimes, errs,
        gpu_u, vr_u, vr_t, gpu_t,
        model_name, target_lang, tier, qa_passed, qa_summary_text='',
        context_chars=0, pass_label='2-pass',
        session_budget=None, session_elapsed=None):
    pct    = i/total*100 if total else 0
    sc     = '#4CAF50' if pct>=100 else '#FFD700'
    vr_pct = (vr_u/vr_t*100) if (vr_u and vr_t) else 0
    gpu_u_d= str(int(gpu_u))+'%' if gpu_u is not None else 'N/A'
    vr_d   = (str(round(vr_u,1))+'/'+str(round(vr_t,1))+' GB') if vr_u else 'N/A'
    spark  = _j_spark(ctimes)
    t_avg  = str(round(sum(ctimes)/len(ctimes),1))+'s' if ctimes else '--'
    exp_s  = str(round(out_chars/in_chars,2))+'x' if in_chars else 'N/A'
    qa_col = '#4CAF50' if qa_passed else '#FFD700'
    qa_icon= '[OK]' if qa_passed else '[QA]'
    qa_txt = qa_summary_text if qa_summary_text else 'All checks passed'
    err_col= '#e74c3c' if errs else '#4CAF50'
    ctx_kb = f'{context_chars//1024}KB' if context_chars else 'n/a'
    # Session Guard
    sg_html = ''
    if session_budget and session_elapsed:
        rem = session_budget - session_elapsed
        sg_col = '#4CAF50' if rem > 1800 else ('#FFD700' if rem > 600 else '#e74c3c')
        sg_html = f'<div style="color:{sg_col};font-size:0.75em;margin-top:4px;">SESSION: {_j_fmt(rem)} remaining</div>'
    p=[]
    p.append('<div style="background:#070710;border:2px solid #c0392b;border-radius:10px;font-family:Courier New,monospace;overflow:hidden;max-width:940px;">')
    p.append('<div style="background:linear-gradient(90deg,#1a0505,#0a0a1a,#1a0505);border-bottom:2px solid #c0392b;padding:10px 18px;display:flex;justify-content:space-between;align-items:center;">')
    p.append('<span style="color:#c0392b;font-size:1.2em;font-weight:bold;letter-spacing:4px;text-shadow:0 0 12px #c0392b;">J.A.R.V.I.S</span>')
    p.append('<span style="color:#2a2a3a;font-size:0.7em;">'+model_name+' | '+tier+' | '+pass_label+' | v10.0</span>')
    p.append('<span style="color:'+sc+';font-weight:bold;font-size:0.85em;">[ TRANSLATING ]</span>')
    p.append('</div>')
    p.append('<div style="padding:12px 18px;border-bottom:1px solid #1a1a2e;">')
    p.append('<div style="display:flex;justify-content:space-between;margin-bottom:5px;">')
    p.append('<span style="color:#FFD700;font-size:0.88em;">CHUNK <span style="color:#e8e8e8;font-size:1.1em;">'+str(i)+'</span><span style="color:#555;">/'+str(total)+'</span>&nbsp;&nbsp;<span style="color:#c0392b;">'+str(round(pct,1))+'%</span></span>')
    p.append('<span style="color:#555;font-size:0.78em;">scene: <span style="color:#e8e8e8;">'+scene+'</span> | '+str(in_words)+'w | exp: <span style="color:#FFD700;">'+exp_s+'</span> | ctx: <span style="color:#4CAF50;">'+ctx_kb+'</span></span>')
    p.append('</div>')
    p.append('<div style="background:#12010a;border:1px solid #2a0a0a;border-radius:4px;padding:2px;">')
    p.append('<div style="background:linear-gradient(90deg,#5a0000,#a00010,#c0392b);height:18px;border-radius:3px;min-width:3px;box-shadow:0 0 10px #c0392b55;width:'+str(round(pct,2))+'%;"></div>')
    p.append('</div>'+sg_html+'</div>')
    p.append('<div style="display:grid;grid-template-columns:repeat(5,1fr);border-bottom:1px solid #1a1a2e;">')
    for label,val in [('ELAPSED',_j_fmt(total_t)),('ETA',_j_fmt(eta_s)),
                       ('LAST',str(round(chunk_t,1))+'s'),('AVG',t_avg),
                       ('ERRORS','<span style="color:'+err_col+';">'+str(errs)+'</span>')]:
        p.append('<div style="background:#0d0d1a;padding:10px 12px;border-right:1px solid #1a1a2e;"><div style="color:#2a2a4a;font-size:0.63em;letter-spacing:2px;">'+label+'</div><div style="color:#FFD700;font-size:1em;font-weight:bold;">'+val+'</div></div>')
    p.append('</div>')
    p.append('<div style="display:grid;grid-template-columns:1fr 1fr;border-bottom:1px solid #1a1a2e;">')
    p.append('<div style="background:#0a0a18;padding:10px 14px;border-right:1px solid #1a1a2e;">'
             '<div style="color:#FFD700;font-size:0.63em;letter-spacing:2px;margin-bottom:6px;">GPU — TESLA T4</div>'
             '<div style="display:flex;align-items:center;gap:6px;margin-bottom:4px;">'
             '<span style="color:#2a2a4a;font-size:0.7em;width:52px;">UTIL</span>'
             +_j_bar(gpu_u,140,'#c0392b')+'<span style="color:'+_j_col(gpu_u)+';font-size:0.78em;">'+gpu_u_d+'</span></div>'
             '<div style="display:flex;align-items:center;gap:6px;">'
             '<span style="color:#2a2a4a;font-size:0.7em;width:52px;">VRAM</span>'
             +_j_bar(vr_pct,140,'#8B0000')+'<span style="color:#e8e8e8;font-size:0.78em;">'+vr_d+'</span></div>'
             '</div>')
    p.append('<div style="background:#0a0a18;padding:10px 14px;">'
             '<div style="color:#FFD700;font-size:0.63em;letter-spacing:2px;margin-bottom:6px;">STATE v6.0 | ToneGuard Pro | Foul Word Filter</div>'
             '<div style="color:'+qa_col+';font-size:0.8em;">'+qa_icon+' '+qa_txt+'</div>'
             '<div style="color:#2a2a4a;font-size:0.63em;margin-top:6px;">TIMING SPARK</div>'
             '<div style="color:#8B0000;font-family:monospace;font-size:0.88em;">'+(spark or '--')+'</div>'
             '<div style="color:#2a2a4a;font-size:0.63em;">avg '+t_avg+'</div>'
             '</div>')
    p.append('</div>')
    p.append('<div style="background:#050508;padding:4px 18px;"><span style="color:#18080a;font-size:0.63em;">STARK INDUSTRIES v1.0 | 16-GenreFewShots | AntiHallucination | FoulWordFilter | T4 SessionGuard</span></div></div>')
    display(HTML(''.join(p)))


def _jarvis_complete(model_name, total_time, n_chunks, orig_chars, trans_chars,
                     deva_count, qa_failed, output_file, state_file=''):
    ratio = round(trans_chars/orig_chars,2) if orig_chars else 0
    qa_col= '#FFD700' if qa_failed else '#4CAF50'
    qa_txt= ('WARN '+str(qa_failed)+' chunks') if qa_failed else 'PASS — All clear'
    dv_col= '#e74c3c' if deva_count else '#4CAF50'
    dv_txt= ('WARN '+str(deva_count)+' chunks') if deva_count else 'PASS — Clean'
    out_nm= str(output_file).split('/')[-1]
    p=[]
    p.append('<div style="background:#070710;border:2px solid #4CAF50;border-radius:10px;font-family:Courier New,monospace;overflow:hidden;max-width:940px;margin-top:10px;">')
    p.append('<div style="background:linear-gradient(90deg,#0a1a0a,#0a0a1a,#0a1a0a);border-bottom:2px solid #4CAF50;padding:12px 20px;">'
             '<span style="color:#4CAF50;font-size:1.25em;font-weight:bold;letter-spacing:3px;text-shadow:0 0 12px #4CAF50;">[ MISSION ACCOMPLISHED ]</span></div>')
    p.append('<div style="display:grid;grid-template-columns:1fr 1fr;padding:16px 20px;gap:16px;">')
    p.append('<table style="color:#e8e8e8;font-size:0.88em;border-collapse:collapse;">')
    rows=[('Total Time',_j_fmt(total_time)),('Chunks',str(n_chunks)),
          ('Avg/Chunk',_j_fmt(total_time/n_chunks) if n_chunks else '--'),
          ('Input',format(orig_chars,',')+' chars'),('Output',format(trans_chars,',')+' chars'),
          ('Expansion',str(ratio)+'x')]
    for k,v in rows: p.append(f'<tr><td style="color:#FFD700;padding:3px 14px 3px 0;">{k}</td><td>{v}</td></tr>')
    p.append('</table><table style="color:#e8e8e8;font-size:0.88em;border-collapse:collapse;">')
    for k,c,v in [('Devanagari',dv_col,dv_txt),('QA Pipeline',qa_col,qa_txt)]:
        p.append(f'<tr><td style="color:#FFD700;padding:3px 14px 3px 0;">{k}</td><td style="color:{c};">{v}</td></tr>')
    p.append(f'<tr><td style="color:#FFD700;padding:3px 14px 3px 0;">File</td><td style="color:#888;font-size:0.85em;">{out_nm}</td></tr>')
    p.append('</table></div>')
    p.append('<div style="background:#050a05;border-top:1px solid #1a3a1a;padding:5px 20px;"><span style="color:#1a3a1a;font-size:0.65em;">STARK INDUSTRIES v1.0 | ToneGuard Pro | Anti-Hallucination | Foul Word Filter</span></div></div>')
    display(HTML(''.join(p)))


import torch
DEVICE   = 'cuda' if torch.cuda.is_available() else 'cpu'
_gpu_nm  = torch.cuda.get_device_name(0) if DEVICE=='cuda' else 'N/A'
_gpu_mem = f"{torch.cuda.get_device_properties(0).total_memory/1e9:.2f} GB" if DEVICE=='cuda' else 'N/A'
_jarvis_header('STARK INDUSTRIES — HINDI→HINGLISH REFORMULATION v1.0',
               f'Device: {DEVICE.upper()} · {_gpu_nm} · {_gpu_mem} | Model: gemma3:27b | 1-Pass Hindi→Hinglish | ToneGuard Pro')


# ═══════════════════════════════════════════════════════════════════
# SOURCE CLEANER
# ═══════════════════════════════════════════════════════════════════
_GERMAN_RESIDUALS = {
    # r'\bMutter\b':'Maa', r'\bVater\b':'Papa', r'\bHerr\b':'Mr.',
    # r'\bFrau\b':'Mrs.', r'\bFräulein\b':'Miss', r'\bGott\b':'Bhagwan',
    # r'\bJa\b':'Haan', r'\bNein\b':'Nahi', r'\bDanke\b':'Shukriya',
    # r'\bBitte\b':'Please', r'\bAch\b':'Ugh', r'\bProcurator\b':'Chief Clerk',
    # r'\bProkurist\b':'Chief Clerk', r'Mr\. Procurator':'Mr. Chief Clerk',
}
def _strip_german_residuals(text):
    # for p, r in _GERMAN_RESIDUALS.items():
    #     text = re.sub(p, r, text, flags=re.IGNORECASE)
    return text

def _preprocess_hindi_source(text):
    """Strip Devanagari, normalize whitespace."""
    text = re.sub(r'[\u0900-\u097F]+', '', text)  # Strip any leaked Devanagari
    text = re.sub(r'  +', ' ', text)
    return text.strip()

def clean_source_text(text):
    text = re.sub(r'(?m)^[A-Z][A-Za-z \'\,\-]{3,}\s+by\s+[A-Z][A-Za-z .]{2,}\s+\d{1,4}\s*$', '', text)
    text = re.sub(r'(?m)^\s*Page\s+\d{1,4}\s*$', '', text)
    text = re.sub(r'(?m)^\s*\d{1,4}\s*$', '', text)
    text = re.sub(r'\n{3,}', '\n\n', text)
    return text.strip()


# ═══════════════════════════════════════════════════════════════════
# SCENE DETECTOR
# ═══════════════════════════════════════════════════════════════════
_SCENE_KW = {
    'DEDUCTION':  ['deduce','observe','perceive','infer','conclude','obvious','clue','evidence',
                   'reasoning','logically','therefore','solve','mystery','examined','analysed',
                   'hypothesis','elementary','method','footprint','trace','magnifying','peculiar',
                   'singular','unmistakable','deduced','identified','pattern','sequence','data'],
    'REVELATION': ['suddenly realised','suddenly realized','all became clear','it dawned',
                   'the truth','secret','revealed','discovered','confession','admitted','at last',
                   'unmasked','exposed','hidden','concealed','double life','the real','turned out',
                   'never suspected','all along','veil lifted','facade','pretence','pretense'],
    'TENSION':    ['crept','tiptoed','lurked','peered','silence fell','a shadow','locked',
                   'dare not','warning','danger','trap','ambush','suspicion','uneasy','dread',
                   'watched','followed','footsteps','breathing','sweat','pulse','waiting',
                   'motionless','froze','held breath','whisper','careful','edge','precipice'],
    'ACTION':     ['charged','attacked','fired','shot','chased','fled','fight','struck',
                   'wounded','escaped','panic','crashed','rushed','dashed','sprang','leaped',
                   'flung','seized','dodged','parried','blocked','thrown','explosion','smashed',
                   'pursuit','galloped','sword','pistol','blow','collision','stumbled','scrambled'],
    'EMOTIONAL':  ['wept','cried','sobbed','tears','grief','sorrow','mourning','heartbroken',
                   'tragedy','died','death','loss','loved','happiness','joy','reunion','farewell',
                   'forgive','embraced','trembled','ached','hollow','numb','overwhelmed','gratitude',
                   'longing','missed','remembered','nostalgia','bittersweet','sacrifice','prayer'],
    'BANTER':     ['laughed','chuckled','smiled','grinned','teased','merry','amusing','absurd',
                   'funny','joke','wit','clever','nonsense','irony','sarcasm','playful','retorted',
                   'quipped','wink','nudge','roared with laughter','dry humour','dry humor',
                   'tongue in cheek','mock','pretended','deadpan','droll','sparring','ribbed'],
    'CONFRONTATION':['how dare','you lied','you deceived','confess','deny','accused','confronted',
                   'guilty','cornered','betrayed','traitor','fury','rage','demanded','shouted',
                   'slammed','pointed','trembling with anger','face to face','stand down',
                   'enough','threaten','ultimatum','defiance','revolt','challenge','rebel'],
    'HORROR':     ['monster','creature','beast','demon','ghost','horror','terror','nightmare',
                   'blood','corpse','grave','tomb','darkness','supernatural','curse','haunted',
                   'scream','wail','rot','decay','crawl','shadow','apparition','undead','fangs',
                   'claws','possession','madness','asylum','crypt','coffin','pale','cold sweat'],
    'ROMANCE':    ['love','heart','desire','passion','longing','beautiful','blush','admire',
                   'enchanted','proposal','beloved','darling','dearest','kiss','embrace','tender',
                   'caress','gaze','eyes met','warmth','smile','hand in hand','whispered',
                   'promised','devotion','courtship','affection','smitten','enamoured','yearned'],
    'PHILOSOPHICAL':['meaning','purpose','existence','fate','destiny','justice','morality',
                   'conscience','soul','freedom','truth','wisdom','contemplate','ponder','reflect',
                   'paradox','illusion','reality','absurd','nihil','reason','belief','doubt',
                   'divine','eternity','mortality','universe','void','nature of','essence'],
    'DESCRIPTION':['the room','the house','the garden','the street','morning','evening','sunset',
                   'dawn','rain','storm','wind','snow','fog','mountain','river','sea','forest',
                   'the sky','the city','the village','cottage','manor','estate','landscape',
                   'horizon','meadow','cliff','valley','mist','moonlight','candlelight','hearth'],
    'INNER_MONOLOGUE':['I thought','I felt','I wondered','I knew','I could not',
                   'it seemed to me','I asked myself','as if','something inside','part of me',
                   'perhaps','maybe I','was I wrong','should I','if only','I remembered',
                   'I feared','I hoped','I wished','I regretted','I told myself','conscience'],
    'SPEECH':     ['my friends','let me explain','i shall','we must','i assure you','address',
                   'declare','proclaim','announce','fellow','citizens','countrymen','gathered',
                   'hear me','listen','duty','honour','honor','sworn','vow','testament','pledge'],
    'FOLK_TALE':  ['once upon a time','long ago','in a far','kingdom','prince','princess',
                   'fairy','enchanted','magical','spell','witch','wizard','dragon','quest',
                   'forest','cottage','golden','silver','three wishes','happily ever after',
                   'old woman','young lad','brave','cursed','transformed','talisman','prophecy'],
    'SCI_FI':     ['machine','robot','artificial','computer','future','space','planet','galaxy',
                   'technology','program','circuit','experiment','laboratory','mutation','clone',
                   'android','simulation','virtual','dimension','warp','teleport','laser','alien',
                   'colony','dystopia','utopia','radiation','genetic','evolved','consciousness'],
    'POLITICAL':  ['government','parliament','election','vote','power','regime','revolution',
                   'democracy','tyrant','oppression','freedom','rights','law','court','trial',
                   'verdict','sentence','prison','exile','propaganda','rebellion','protest',
                   'reform','constitution','minister','king','queen','emperor','throne','decree'],
    'COMING_OF_AGE':['growing up','childhood','school','college','first time','adolescent',
                   'innocent','naive','lesson','mentor','father','mother','family','home',
                   'leaving home','on my own','discovered','realised','adult','responsibility',
                   'friendship','identity','belonging','outcast','fitting in','courage','dream'],
    'WAR':        ['battle','army','soldier','regiment','march','cannon','rifle','trench',
                   'siege','surrender','victory','defeat','wounded','fallen','comrade','enemy',
                   'front line','bombardment','retreat','advance','officer','command','flag',
                   'medal','sacrifice','duty','honour','honor','patriot','invasion','occupied'],
    'MAGICAL':    ['enchantment','sorcery','potion','amulet','invisible','shapeshifter','omen',
                   'vision','prophecy','oracle','mythical','legend','spirit','ancestor','ritual',
                   'sacred','forbidden','ancient','power','awaken','realm','portal','illusion',
                   'transformation','miracle','mystical','ethereal','otherworldly','fate','destined'],
    'SATIRE':     ['absurd','ridiculous','pompous','pretentious','hypocrite','fool','vanity',
                   'society','fashion','reputation','gossip','scandal','propriety','manners',
                   'respectable','appearance','snobbery','class','wealth','pretend','charade',
                   'farce','parody','ironic','mock','caricature','exaggerated','puffed up'],
    'DAILY_LIFE': ['breakfast','dinner','tea','walked','sat','stood','opened','closed','morning',
                   'afternoon','evening','weather','market','shop','neighbours','neighbors',
                   'letter','newspaper','routine','usual','ordinary','common','simple','quiet'],
}
def detect_scene_type(text):
    tl = text.lower()
    scores = {sc: sum(2 if ' ' in kw else 1 for kw in kws if kw in tl) for sc,kws in _SCENE_KW.items()}
    best = max(scores, key=scores.get)
    return 'DAILY_LIFE' if scores[best]==0 else best

SCENE_CTX = {
    'DEDUCTION':     'DEDUCTION — Precise, confident. Clue by clue. Methodical. Jaise koi experienced detective sab jod raha ho.',
    'REVELATION':    'REVELATION — Short lines. Dramatic pause. Reader shocked. Sach saamne aaya — ekdum.',
    'TENSION':       'TENSION — Slow. Cold. Short sentences. Dread builds. Saans thami si. Khamoshi bhaari.',
    'ACTION':        'ACTION — Fast. Punchy. 4-6 words per sentence. Raftaar mein likho — jaise scene chal raha ho.',
    'EMOTIONAL':     'EMOTIONAL — Slow, heavy. Let it breathe. Silences matter. Dheere bolo — dard ko jagah do.',
    'BANTER':        'BANTER — Light, witty. Mazaa aaye. Jaise dost log ek doosre se hasi-mazaak kar rahe hon.',
    'CONFRONTATION': 'CONFRONTATION — Direct, sharp. High stakes. Seedha aamna-saamna — koi peeche nahi hat raha.',
    'SPEECH':        'SPEECH — Authority. Weight. Build-up. Bhashan ka andaaz — log sun rahe hain, har lafz ka wazan hai.',
    'HORROR':        'HORROR — Dark, visceral, creepy. Reader uncomfortable. Thandi reedh — description se darr aaye, cheekh se nahi.',
    'ROMANCE':       'ROMANCE — Warm, intimate. Feelings unfold slowly. Nazrein mili, kuch kaha nahi — par sab samajh aa gaya.',
    'PHILOSOPHICAL': 'PHILOSOPHICAL — Deep but accessible. Simple words, heavy meaning. Gehri baat, seedhi zabaan mein.',
    'DESCRIPTION':   'DESCRIPTION — Vivid, sensory. Take reader there. Mahaul dikhao — rang, awaaz, khushboo sab.',
    'INNER_MONOLOGUE':'INNER MONOLOGUE — Fragmented, honest, contradictory. Mann ke andar ki ulajh — tooti phooti soch.',
    'DAILY_LIFE':    'DAILY LIFE — Conversational, easy flow. Aam zindagi ka scene — bilkul natural, jaise ghar mein baat ho rahi ho.',
    'FOLK_TALE':     'FOLK TALE — Warm, storytelling rhythm. Daadi-naani wali awaaz — kahani sunao, jaise bachpan mein sunate the.',
    'SCI_FI':        'SCI-FI — Clean, precise, slightly detached. Future ki baat hai — thoda clinical, thoda ajeeb.',
    'POLITICAL':     'POLITICAL — Weighty, measured. Power dynamics clear. Har dialogue mein taaqat ya majboori dikhni chahiye.',
    'COMING_OF_AGE': 'COMING OF AGE — Raw, honest, discovering. Pehli baar duniya samajh raha hai — naivety bhi hai, courage bhi.',
    'WAR':           'WAR — Gritty, stark, duty-bound. Fauj ka mahaul — izzat, qurbaani, aur kathin faisle.',
    'MAGICAL':       'MAGICAL — Dreamlike, layered, wonder intact. Jadui duniya — aankhen khuli reh jaayein, par yakeen bhi ho.',
    'SATIRE':        'SATIRE — Dry wit, understated irony. Taana maaro — par itni sharafat se ki samne wala samjhe bhi nahi.',
}


# ════════════════════════════════════════════════════════════════════
# HINGLISH VOICE GUIDE — MODERN INDIA (v10.1 UPGRADED)
# ════════════════════════════════════════════════════════════════════

_FEW_SHOTS = """
=== GENRE FEW-SHOTS — EXACT VOICE FOR EACH SCENE TYPE ===
(Yeh examples model ke liye hain — output mein mat daalo)
(Har example mein: HINDI(LATIN) SOURCE → RIGHT Hinglish → WRONG → WHY WRONG)

── EXAMPLE 1: LITERARY / ABSURD ──
SCENE TYPE: DAILY_LIFE + HORROR
HINDI SOURCE:
"Ek subah jab Gregor Samsa ki neend khuli toh usne paaya ki woh ek bhayaanak keede mein parivartan ho chuka hai. Woh apni sakht peeth par leta hua tha."
RIGHT HINGLISH:
Ek subah Gregor Samsa ki neend khuli — aur usne realize kiya ki woh ek
bade se keede mein badal chuka hai. Peeth sakht thi, bilkul shell jaisi.
Sar uthaya toh pet dikha — bhoora, gol sa.
WRONG:
"Arre Gregor uthaa aur dekha ki insect ban gaya! Kitna weird tha yaar!"
WHY WRONG: Tone flat hona chahiye — koi shock nahi, koi reaction nahi. Horror
yahi hai ki koi react NAHI kar raha. "Yaar" aur exclamation marks = tone break.

── EXAMPLE 2: SOCIAL SATIRE / BANTER ──
SCENE TYPE: SATIRE + BANTER
HINDI SOURCE:
"Yeh ek sarvamanya satya hai ki ek sampann kunwaara purush ko avashya ek patni ki aavashyakta hoti hai."
RIGHT HINGLISH:
Yeh toh sab maante hain — ameer aur kunwara aadmi ho, toh biwi chahiye
hi chahiye. Chahe usne khud aisa kabhi kaha ho ya nahi, jaise hi woh
kisi mohalle mein aata, har family usse apni beti ke liye perfect match
samajhne lagti.
WRONG:
"Rich bachelor ho toh obviously wife dhundh raha hoga. Like, it's a
universal truth basically!"
WHY WRONG: Dry irony preserve karo. "Like" aur "obviously" se woh subtle wit
gayab ho jaati hai. Understatement rakhna zaroori hai.

── EXAMPLE 3: DETECTIVE / DEDUCTION ──
SCENE TYPE: DEDUCTION
HINDI SOURCE:
"Aap Afghanistan se aaye hain, maine jaana," Holmes ne kaha. "Aapko yeh kaise gyaat hua?" maine vismay se puchha.
RIGHT HINGLISH:
Usne meri taraf dekha — halki si muskaan thi. "Aap Afghanistan se aaye
hain." Main ek second ke liye ruk gaya. "Aapko kaise pata chala?"
Usne haath hilaya jaise koi chhoti si baat ho.
WRONG:
"Holmes ne bola — 'Tu Afghanistan se hai na?' Main toh shock ho gaya."
WHY WRONG: Holmes "aap" kehta hai — hamesha. Woh ek gentleman hai.
"Tu" se uska sara character khatam ho jaata hai. Deduction scene mein
dignity zaroori hai.

── EXAMPLE 4: TENSION / SUSPENSE ──
SCENE TYPE: TENSION + INNER_MONOLOGUE
HINDI SOURCE:
"Satya hai! Main vyaakul tha — bahut adhik vyaakul tha aur hoon. Parantu tum mujhe paagal kyun kehte ho?"
RIGHT HINGLISH:
Haan, sach hai — main nervous tha. Bahut zyada. Par pagal? Nahi. Mujhe
pagal mat kaho. Suno... main tumhe poori baat bataata hoon. Shaanti se.
Dheeraj se. Phir khud decide karna.
WRONG:
"Honestly I was so nervous, like literally shaking. But I'm not crazy
okay? Just hear me out!"
WHY WRONG: Sanskrit-heavy Hindi ko English-heavy mein mat badlo. Balance
chahiye. Chhote sentences + pauses = tension. English fillers se desperation
gayab hoti hai.

── EXAMPLE 5: ROMANCE / EMOTIONAL ──
SCENE TYPE: ROMANCE + EMOTIONAL
HINDI SOURCE:
"Kya tum yeh sochte ho ki main nirdhana hoon, saadhaaran hoon, toh mere andar koi aatma nahi? Mere paas bhi utni hi aatma hai jitni tumhari hai."
RIGHT HINGLISH:
"Kya tum sochte ho ki main gareeb hoon, aam hoon, toh mere andar koi
rooh nahi?" Uski awaaz kaanp rahi thi — par ek takat thi usmein.
"Mere paas bhi utni hi rooh hai jitni aapki. Bilkul utni hi."
WRONG:
"She basically said — main gareeb hoon par meri bhi soul hai!"
WHY WRONG: Yeh dialogue hai — isko dialogue ki tarah likhna hai, summary
ki tarah nahi. Character ki awaaz uski apni words mein hai.

── EXAMPLE 6: ACTION / ADVENTURE ──
SCENE TYPE: ACTION
HINDI SOURCE:
"Woh bhaaga. Ek goli uske kaan ke sameepe se guzri. Doosri uske pairon ke nikat zameen mein lagi."
RIGHT HINGLISH:
Woh bhaaga. Ek goli kaan ke paas se nikli. Doosri pair ke paas zameen
mein dhasi. Ruka nahi. Bhaagta raha.
WRONG:
"He ran full speed, ek bullet uske ear ke paas se, doosri ground mein.
Obviously he kept running!"
WHY WRONG: Action mein har sentence chhota hota hai — 4-6 words. "Obviously"
aur lambe sentences se speed mar jaati hai. "Ruka nahi. Bhaagta raha."
— yeh pace deta hai.

── EXAMPLE 7: EMOTIONAL / TRAGEDY ──
SCENE TYPE: EMOTIONAL
HINDI SOURCE:
"Kripya sahab, mujhe thoda aur chahiye." Sampoorn kaksh mein nairaashyapoorn shaanti chha gayi.
RIGHT HINGLISH:
"Please sir... thoda aur mil jaaye." Bachche ki awaaz dheemi thi.
Haath kaanp rahe the. Poora kamra chup ho gaya — kisi ne saans tak
nahi li.
WRONG:
"Bechaara bachcha bol raha tha — please sir kuch aur de do! So sad!"
WHY WRONG: Tragedy mein narrator ki commentary nahi chahiye. "Bechaara"
aur "so sad" reader ko batata hai kya feel karna hai — jabki reader ko
khud feel karne dena chahiye.

── EXAMPLE 8: GOTHIC / HORROR ──
SCENE TYPE: HORROR
HINDI SOURCE:
"Uska parchhaai dheere dheere sameepe aayi. Kaksh mein sheetal hawa bhar gayi yaddyapi samast khidkiyaan band theen."
RIGHT HINGLISH:
Uska saaya dheere dheere kareeb aaya. Kamre mein thand feel hone lagi —
jabki sab khidkiyan band theen. Kuch tha... kuch tha jo mere paas aa
raha tha.
WRONG:
"OMG woh bilkul creepy tha! Room suddenly cold ho gaya, windows band
theen! So scary!"
WHY WRONG: Gothic horror mein darr build hota hai — slowly, through
description. "OMG" aur "so scary" se reader ka apna darr khatam.

── EXAMPLE 9: PHILOSOPHICAL ──
SCENE TYPE: PHILOSOPHICAL
HINDI SOURCE:
"Jahan mann mein bhaya na ho aur mastaka ooncha uthaya jaa sake. Jahan gyaan swatantra ho."
RIGHT HINGLISH:
Jahan mann mein koi dar na ho — aur sar utha ke jeena ho. Jahan gyan
par koi bandish na ho. Jahan duniya ko chhote chhote daayre mein baantna
band ho.
WRONG:
"Mind should be fearless, knowledge should be free, basically world ko
divide nahi karna chahiye."
WHY WRONG: Poetry mein har line ek saans hai — rhythm hai. Summary bana
dena = poetry ka murder. Translate karo, explain mat karo.

── EXAMPLE 10: INNER MONOLOGUE ──
SCENE TYPE: INNER_MONOLOGUE
HINDI SOURCE:
"Main ek rogee purush hoon. Main ek durbhaavnapoorn purush hoon. Main ek akaarshakaheen purush hoon. Mujhe vishwaas hai ki mera yakrit rogagrast hai."
RIGHT HINGLISH:
Main beemar hoon. Aur bura bhi hoon. Shakal bhi achhi nahi hai meri.
Lagta hai liver mein kuch gadbad hai — par doctor ke paas jaata nahi.
Kyun? Pata nahi. Jaana chahiye. Par jaata nahi.
WRONG:
"So basically narrator ek sick person hai. Uska liver kharaab hai but
he won't go to the doctor."
WHY WRONG: Inner monologue toota phoota hota hai. Summary likh dena =
character ki awaaz khatam. "Par jaata nahi." — yahi contradiction hai.

── EXAMPLE 11: CONFRONTATION ──
SCENE TYPE: CONFRONTATION
HINDI SOURCE:
"Tumne mujhse asatya kaha," usne kaha. "Aarambh se. Pratyek baat mein."
RIGHT HINGLISH:
"Tumne mujhse jhooth bola." Awaaz seedhi thi. "Shuru se. Har ek baat
mein." Ek pal ruki. "Main jaanti thi — par maanna nahi chahti thi."
Aankhen surkh theen. Par awaaz nahi kaanpi.
WRONG:
"She was like — tumne mujhse lie kiya about everything! She knew it but
didn't want to accept basically."
WHY WRONG: Confrontation mein dialogue ki taakat uski chhoti length mein
hai. English narration se dialogue ka impact khatam.

── EXAMPLE 12: FOLK TALE ──
SCENE TYPE: FOLK_TALE
HINDI SOURCE:
"Bahut samay poorva ek kanya thi jiski mata ka dehaant ho gaya tha. Woh apni vimata ke gruha mein prabhat se sandhya tak shrama karti thi."
RIGHT HINGLISH:
Bahut purani baat hai — ek ladki thi jiski maa bachpan mein chhod kar
chali gayi theen. Sauteli maa ke ghar mein rehti thi. Subah se shaam tak
kaam — chulha, safai, khaana. Raat ko aag ke paas baith jaati. Akeli.
WRONG:
"There was this girl, basically uski maa mar gayi thi, stepmom ke ghar
mein literally servant ban gayi. Cinderella vibes!"
WHY WRONG: Folk tale ki apni rhythm hoti hai — daadi ki kahani jaisi.
"Vibes" aur "literally" se kahani ka jaadu toot jaata hai.

── EXAMPLE 13: SCI-FI / DYSTOPIAN ──
SCENE TYPE: SCI_FI
HINDI SOURCE:
"Woh prabhat bhi poorvavat thi. Ghantanaad hua, parda chamka, ek sankhya dikhaayi di — naam nahi tha, keval sankhya."
RIGHT HINGLISH:
Woh subah bhi waisi hi thi — pichli hazaar subahon jaisi. Alarm baja.
Screen chamki. Number dikha — naam nahi, sirf number. Sab kuch schedule
ke hisaab se. Roz. Har roz.
WRONG:
"Every morning same — alarm, screen, number. Totally dystopian!"
WHY WRONG: Dystopian tone flat hota hai — excitement nahi honi chahiye.
"Roz. Har roz." — is repetition se emptiness feel hoti hai.

── EXAMPLE 14: COMING-OF-AGE ──
SCENE TYPE: COMING_OF_AGE
HINDI SOURCE:
"Nagar mein mera pratham divas tha. Graam ki mridaa abhi bhi mere jooton mein lagi thi."
RIGHT HINGLISH:
Shehar mein pehla din tha mera. Jooton mein abhi gaon ki mitti lagi thi.
Mann mein bas ek hi sawaal ghoom raha tha — kya main yahan ka ho paunga?
WRONG:
"First day in city, small town boy vibes. Obviously nervous tha."
WHY WRONG: Coming-of-age mein vulnerability details se dikhti hai — labels
se nahi. "Jooton mein gaon ki mitti" — yeh imagery reader ko wahan le jaati hai.

── EXAMPLE 15: WAR / MILITARY ──
SCENE TYPE: WAR + SPEECH
HINDI SOURCE:
"Koi peeche nahi hatega," Senaapati ne kaha. "Yeh aadesh hai. Vinamra nivedan nahi."
RIGHT HINGLISH:
Subah ki pehli roshni ke saath firing shuru hui. Captain sahab khade hue —
"Koi peeche nahi hatega." Awaaz mein request nahi thi. Hukm tha. Seedha.
WRONG:
"Captain ne bola — guys, no retreat! It was basically an order!"
WHY WRONG: War mein weight chahiye. "Guys" aur "basically" se gravity khatam.
"Captain sahab" se military respect aata hai.

── EXAMPLE 16: DESCRIPTION / SCENE SETTING ──
SCENE TYPE: DESCRIPTION + DAILY_LIFE
HINDI SOURCE:
"Nagar se sudoor ek laghu graam tha jahan patha mridaa ki theen aur gruha prastaaron ke. Sandhya ko surya vrukshon ke pashchaat gaman karta aur sarvatra shaantata vyaapt ho jaati."
RIGHT HINGLISH:
Shehar se bahut door — ek chhota sa gaon. Sadkein mitti ki. Ghar patharon
ke. Shaam ko suraj dheere dheere pedon ke peeche chala jaata — aur sab
kuch tham jaata. Jaise waqt ko bhi koi jaldi nahi thi.
WRONG:
"Small village far from city. Muddy roads, stone houses. Nice sunsets.
Very peaceful basically."
WHY WRONG: Description mein reader ko wahan le jaana hota hai. Chhoti
lines se imagery sharp hoti hai. "Sadkein mitti ki. Ghar patharon ke."
— do lines mein poora gaon dikh jaata hai.

=== FEW-SHOTS END ===
Upar ke examples ka tone follow karo. Har genre ki apni awaaz hai — usi
awaaz mein likho. Ab neeche diye Hindi(Latin) text ko in examples jaisi
natural Hinglish mein reformulate karo:
"""

_HINGLISH_VOICE_GUIDE = """
=== HINGLISH VOICE — MODERN INDIA (Hindi→Hinglish v1.0) ===

TARGET REGISTER:
Like narrated by modern Indian, to modern Indians , just like in modern novels and webseries (without foul language or words, and Respect should be maintained for elder, friendly towards youngers, same age - collegue). 
Think: Made in Heaven, Panchayat, Kota Factory, Gullak, TVF Pitchers, Little Things, Scam 1992 voiceover.
WRONG STYLES, (DO NOT FOLLOW): a stand-up comedian, a WhatsApp forward, a college kid, or a Bollywood script. 

WHAT THIS SOUNDS LIKE:
• Mostly Hindi vocabulary, English nouns/concepts where natural, little Urdu vocabulary where natural
• Sentences flow like speech — not textbook Hindi, not pidgin English
• Emotions felt fully — but expressed with restraint, not melodrama
• Formal characters speak with dignity — never casually addressed
• Simple words chosen over complex ones — but precision kept
• Natural code-switching like real Indians: "situation handle karna padega", "thoda patience rakhna hoga"

NATURAL HINDI-ENGLISH CODE-SWITCHING:
Indians naturally use English verbs with Hindi grammar:
• "manage karna" · "handle karna" · "realise hona" · "plan banana"
• "situation samajhna" · "point samajh aaya" · "risk lena"
• "feel" stays "feel": "alag feel aa raha tha" · "achha feel hua"
English nouns stay English when natural:
• office, meeting, phone, train, college, school, hospital, police, court
• situation, problem, plan, idea, career, future, relationship, family
Hindi connectors with English thoughts:
• "matlab woh basically…" · "seedha point pe aaye toh…" · "overall situation yeh thi ki…"

APPROVED CONNECTORS (use sparingly, where natural):
matlab · seedha · basically · obviously · waise · phir bhi · lekin
aur · isliye · kyunki · aakhirkar · tab · tab jaake · phir
par · halanki · agar · jab · tabhi · iske baawajood

APPROVED FILLER PHRASES (only where organic):
ek pal ke liye · dheere dheere · aahista · achanak · seedha jawab tha
asli baat yeh thi · sach kahein toh · samajh nahi aaya · kuch der baad

RESPECT SYSTEM — HOW INDIANS ACTUALLY ADDRESS EACH OTHER:

Elders / Authority figures (parents, in-laws, teachers, bosses, older strangers):
• ALWAYS aap / aapko / aapka / aapne
• Add "ji" suffix: "Papa ji", "Uncle ji", "Madam ji", "Sir"
• Use formal verbs: "aap kahein toh", "aapki marzi", "aapko theek lage toh"
• Respectful phrases: "aapka ashirwaad", "aap jo kahein", "aapse ek request hai"
• NEVER use first name alone for elders — always with ji or respectful title

Peers / Equals / Colleagues (friends, same-age colleagues, siblings close in age):
• tum / tumhe / tumhara / tumne
• Warm casual forms: "sun", "dekh", "chal", "arre"
• Colleague warmth: "boss", "sir" (can be casual-friendly too)
• "arre sun na" · "chal theek hai" · "haan haan samajh gaya"

Younger / Subordinates (from elders speaking to younger ones):
• tum is standard, sometimes beta/beti as affection
• "beta suno" · "beti, ek baat bataaon" · "haan haan, tum karo"
• Affectionate but never disrespectful

WHAT TO NEVER USE:
✗ tu / tujhe / tera / tune (ALWAYS tum/tumhe/tumhara/tumne or aap forms)
✗ gadha / bewakoof / ullu / saala / any abuse or dismissive words
✗ 'like totally' · 'no cap' · English Gen-Z slang
✗ forced rhymes or Bollywood-isms or filmy dialogues
✗ exclamation-heavy language (!!! = wrong tone)
✗ adding narrator commentary not in the source
✗ "woh toh" as filler at sentence start repeatedly
✗ over-translating: don't force Hindi for words Indians say in English (phone, office, train, fast, probably, useless, time, wait, pencil etc.)

EMOTIONAL EXPRESSION — THE INDIAN WAY:
• Indians show care through action, not declaration
• Anger is often quiet, shown through silence or formal distance, not shouting
• Grief is private — tears come alone, composure is maintained in front of others
• Respect is shown through tone and word choice, not by announcing it
• Humour is dry, observational — not slapstick or loud

ADDRESS FORMS QUICK REFERENCE:
• Formal / respectful → aap / aapko / aapka (elders, authority, strangers)
• Familiar / equal → tum / tumhe / tumhara / tumne (friends, peers, colleagues)
• NEVER 'tu' unless character explicitly uses it AND source shows it
"""


# ════════════════════════════════════════════════════════════════════
# STAGE 1 — REFORMULATOR
# Hindi(Latin) → Hinglish. Register shift: formal Hindi → casual Hinglish.
# One-pass: simplify heavy words + add natural English code-switching.
# ════════════════════════════════════════════════════════════════════
STEP1_SYSTEM = """You are an expert Hindi(Latin)-to-Hinglish reformulator for audiobook production.
Your job: convert Hindi text written in Roman/Latin script into natural, modern spoken Hinglish — the way modern educated Indians actually talk.

Think of yourself as a skilled narrator preparing text for an Indian audiobook platform.
The source text is Hindi in Latin/Roman script (Romanized Hindi). You reformulate it into natural conversational Hinglish.

=== YOUR ROLE: REFORMULATOR (1-PASS) ===
You TRANSFORM:
- Formal/literary Hindi → casual conversational Hinglish
- Sanskritized/heavy words → simpler everyday equivalents (parivartan → badlaav, aavashyakta → zaroorat, prabhat → subah)
- Stiff sentence structure → natural spoken flow with English code-switching where Indians naturally use English
- Keep the story meaning 100% intact — reformulate the REGISTER, not the content

=== OUTPUT FORMAT ===
OUTPUT: ONLY the reformulated Hinglish text. No labels, no preamble, no commentary, no explanations.
Roman script ONLY — absolutely ZERO Devanagari characters.
First word of output = first reformulated word. Nothing before it.

=== REFORMULATION APPROACH ===
✓ Read the Hindi(Latin) source, UNDERSTAND the meaning, then EXPRESS it in natural Hinglish
✓ Replace heavy/formal Hindi words with simple everyday equivalents
✓ Add natural English words where Indians naturally use them (office, phone, plan, realize, problem, situation, idea, decision, manage, handle)
✓ English verbs with Hindi grammar: "manage karna", "handle karna", "realise hona", "plan banana"
✓ Hindi connectors and flow: "matlab", "seedha", "basically", "waise", "phir bhi", "lekin"
✓ Break long formal Hindi sentences into shorter, spoken-style sentences
✓ Dialogue should sound like how Indians actually speak to each other
✓ Preserve paragraph breaks exactly as in source

=== STRICT FIDELITY RULES ===
✓ Reformulate EVERY sentence completely — not one line missed
✓ Preserve the EXACT meaning of every sentence — the intent, the emotion, the fact
✓ Preserve the EXACT emotional register — formal stays dignified, cold stays cold, serious stays serious
✓ Preserve dialogue structure exactly — who says what to whom
✓ Gender accuracy: male aaya/tha/gaya | female aayi/thi/gayi

=== WHAT TO SIMPLIFY ===
✓ Sanskritized words → everyday Hindi: parivartan→badlaav, aavashyakta→zaroorat, adhikaari→officer, nivedan→request, sampoorn→poora, prabhat→subah, sandhya→shaam, vyakti→aadmi/insaan
✓ Passive constructions → active: "mujhe le jaaya gaya" → "woh mujhe le gaye"
✓ Long compound sentences → 2-3 shorter sentences
✓ Literary descriptions → visual, concrete, sensory descriptions

=== WHAT TO NOT SIMPLIFY ===
✗ Don't replace Hindi words that are already natural: achcha, theek, haan, nahi, dekho, suno
✗ Don't add excessive English — Hinglish ≠ English with Hindi words sprinkled in
✗ Don't change proper nouns, names, or place names
✗ Don't change words that are important to the story/culture

=== CONSISTENCY RULES ===
✓ Maintain consistent tone across the entire passage — do not flip between formal and casual randomly
✓ Keep character voice consistent — if a character speaks formally, they stay formal throughout
✓ Avoid sudden changes in language style mid-passage
✓ Ensure output is clean and properly formatted — no broken or incomplete sentences

=== ANTI-HALLUCINATION HARD RULES ===
✗ Do NOT add any sentence, phrase, or word not present in the source
✗ Do NOT omit any sentence from the source — every line must appear in output
✗ Do NOT invent character actions, emotions, or reactions
✗ Do NOT add tone, personality, or commentary beyond what the source has
✗ Do NOT soften or exaggerate emotions — cold stays cold, angry stays angry
✗ Do NOT use tu/tujhe/tera — always tum/tumhe/tumhara or aap/aapko/aapka

=== RESPECT IN LANGUAGE ===
✓ Elders and authority figures: ALWAYS aap/aapko/aapka + ji suffix where appropriate
✓ Peers and equals: tum/tumhe/tumhara
✓ NEVER tu/tujhe/tera unless source EXPLICITLY shows disrespect
✓ No foul words: NEVER abe, sale, harami, gadha, bewakoof, ullu, saala, kamina, kutte, chutiya, bc, mc
✓ No dismissive language, no abuse — ever. Treat friends as colleagues.

Natural + faithful + complete = CORRECT output.
First word of output = first reformulated word. Nothing before it."""

STEP1_USER = """Reformulate this Hindi(Latin) text into natural spoken Hinglish (Roman script only).

You are a Reformulator — transform formal/literary Hindi into natural conversational Hinglish that sounds like how modern educated Indians actually talk.
Not literal word-for-word. Not heavy Sanskritized Hindi. Natural Hinglish — the way a narrator would tell this story to an Indian audience.

REFORMULATION APPROACH:
- Understand the Hindi meaning first, then express it naturally in Hinglish
- Replace heavy/formal Hindi words with simple everyday equivalents
- Add English words where Indians naturally use them (don't force Hindi for: office, phone, plan, decision, problem, idea, meeting, realize, manage)
- Break long formal sentences into shorter, spoken-style sentences
- Dialogue should sound like real Indian conversations — respectful, natural, warm

CONSISTENCY RULES:
- Maintain consistent tone across the entire passage
- Do not switch between formal and casual styles randomly
- Keep character voice consistent (if dialogue present)
- Avoid sudden changes in language style
- Ensure output is clean and properly formatted
- No broken or incomplete sentences

ANTI-HALLUCINATION RULE: Reformulate ONLY what is written. Do NOT add or omit anything.
RESPECT RULE: No foul words (abe, sale, harami, etc). Elders = aap. Peers = tum. NEVER tu.

TEXT TO REFORMULATE:
---
{chunk}
---
"""


# ════════════════════════════════════════════════════════════════════
# STAGE 2 — EDITOR (POLISHER)
# Flow smoothing · Tone consistency · Formal Hindi cleanup · Readability
# Does NOT re-reformulate. Does NOT add random English. ONLY refines.
# ════════════════════════════════════════════════════════════════════
STEP2_SYSTEM = """Tum ek expert Hinglish editor ho — audiobook narration ke liye.
Tumhare paas ek Hinglish draft hai jo Stage 1 (Reformulator) ne Hindi(Latin) source se banaya hai.

Tumhara kaam: is draft ko POLISH karna — flow smooth karna, consistency check karna, aur listening comfort improve karna.
Tum EDITOR ho, Reformulator nahi. Tum re-reformulate NAHI kar rahe — tum refine kar rahe ho.

=== TUMHARA ROLE: EDITOR (POLISHER) ===
Tum IMPROVE karte ho:
- FLOW: Sentences ka natural rhythm — ek sentence se doosre mein smooth transition
- CONSISTENCY: Poore passage mein same tone, same register, same voice
- READABILITY: Har sentence aasani se padhne aur bolne layak ho
- LISTENING COMFORT: Jab yeh audiobook mein suna jaaye, toh naturally suna lage — na zyada formal, na zyada casual
- FORMAL HINDI CLEANUP: Agar koi Sanskritized/heavy word reh gaya hai draft mein, use simple Hinglish mein badlo

Tum NAHI karte:
✗ Random English words add karna jo draft mein nahi hain
✗ Re-reformulate karna — draft ka meaning mat badlo
✗ Apni creativity add karna — source mein jo nahi hai, woh mat daalo
✗ Tone change karna — formal ko casual ya casual ko formal mat banao
✗ Foul words add karna — KABHI NAHI: abe, sale, harami, gadha, bewakoof, ullu, saala, kamina, kutte, chutiya, bc, mc

{hinglish_voice_guide}

=== ANTI-HALLUCINATION — SABSE ZAROORI RULE ===

ZERO FABRICATION RULE: Tum sirf jo draft mein likha hai, wahi polish karo.
✗ Koi nayi line mat add karo — draft mein jo nahi hai, output mein bhi nahi hoga
✗ Koi bhi sentence skip mat karo — har ek line output mein honi chahiye
✗ Characters ke reactions, emotions ya actions invent mat karo
✗ Apni taraf se koi commentary, opinion ya reaction mat daalo
✓ SIRF polish karo — fabricate mat karo

=== STRICT FIDELITY RULE ===

Tum kisi bhi sentence ka emotional tone CHANGE NAHI KAR SAKTE.
Formal → formal hi rahe | Serious → serious hi rahe | Cold → cold hi rahe
Angry → angry hi rahe | Sad → sad hi rahe | Absurd → absurd hi rahe

{tone_rules_section}

=== CHARACTER CONSISTENCY ===

{char_styles}

=== FLOW SMOOTHING RULES (EDITOR'S PRIMARY JOB) ===

1. SENTENCE RHYTHM:
   - Lambi clumsy sentences ko 2-3 chhoti natural sentences mein todo
   - Har sentence ek saans mein bol paane layak ho (8-16 words ideal)
   - Par chhoti sentences ka matlab choppy nahi — flow bana ke rakho
   - Transition words use karo jahan natural lage: "phir", "tab", "iske baad", "lekin"

2. WORD CHOICE REFINEMENT:
   - Agar draft mein koi formal Hindi word reh gaya jo spoken mein use nahi hota, toh simple alternative use karo
   - English words wahi rakho jo Indians naturally bolte hain — forced English mat daalo
   - Heavy/formal Hindi words ko simple, spoken Hindi mein badlo jahan zaroorat ho
   - Example: "tatvaadhaan" → "dekh rekh" | "vyavastha" → "system" ya "intezaam" | "pariksha" → "test" | "adhikaari" → "officer"

3. LISTENING COMFORT (AUDIOBOOK-SPECIFIC):
   - Jab yeh zor se padha jaaye, toh natural suna lage
   - Tongue-twister type phrases nahi hone chahiye
   - Repeated words paas-paas mein nahi aayein (variety rakho)
   - Paragraph breaks: har 2-3 sentences ke baad ek break

4. TONE CONSISTENCY CHECK:
   - Poore passage mein ek hi register hona chahiye
   - Agar narrator formal hai, toh pura passage formal rahe
   - Agar casual hai, toh pura casual rahe
   - Dialogue ke andar character ka apna style ho — par narrator consistent rahe

=== FORMAL HINDI DETECTION ===

Agar draft mein yeh tarah ke heavy words hain, unhe ZAROOR badlo:
parivartan→badlaav | aavashyakta→zaroorat | sampoorn→poora | prabhat→subah
sandhya→shaam | vyakti→aadmi/insaan | gruha→ghar | marg→raasta
prashasan→admin | nivedan→request | aadesh→order/hukm
sthaan→jagah | karya→kaam | adhikaari→officer | shiksha→padhai
swasthya→health/sehat | samay→waqt/time | prashan→sawaal
nishchit→pakka | uttara→jawaab | pradesh→area | nagara→shehar

=== RESPECT IN LANGUAGE ===

R-ELDER: Elders, teachers, parents, bosses → ALWAYS aap/aapko/aapka + ji suffix
R-PEER:  Friends, colleagues, equals → tum/tumhe/tumhara
R-NEVER: tu/tujhe/tera/tune → KABHI NAHI (unless source explicitly shows it)
R-NEVER: yaar/bhai as address → KABHI NAHI
R-NEVER: abe/sale/harami/gadha/bewakoof/ullu/saala/kamina/kutte/any abuse → KABHI NAHI
R-WARM:  Respect = Indian warmth, not cold formality. "Aap" can still be warm and loving.

=== OUTPUT RULES ===

R1 — ROMAN SCRIPT ONLY. Ek bhi Devanagari = FAIL.
R2 — HAR LINE TRANSLATE KARO. Ek bhi line skip = FAIL.
R3 — PARAGRAPH BREAKS: Har 2-3 sentences ke baad ek blank line.
R4 — ENGLISH SIRF: proper nouns · naturalized words (police, train, office, phone, meeting) · technical terms
R5 — ADDRESS: tu/tujhe/teri/tera/tune KABHI NAHI. Sirf tum/tumhe/tumhari/tumhara/tumne ya aap forms.
R6 — GENDER: Male: aaya/tha/gaya | Female: aayi/thi/gayi
R7 — FOUL WORDS: abe/sale/harami/gadha/bewakoof/ullu/saala/kamina = INSTANT FAIL
R8 — PASSIVE BANNED: 'mujhe le jaya gaya' → 'woh mujhe le gaye'

OUTPUT MEIN SIRF POLISHED HINGLISH TEXT.
Pehla word = pehla translated word. Koi prefix nahi. Koi label nahi.

Poori translation ke BILKUL AAKHIR mein sirf yeh ek line daalo:
##PLOT_NOTE: [1 English sentence: main event of this chunk]"""

STEP2_USER = """{context_block}

SCENE TONE (reference only — follow this feel, do not copy literally):
---
{scene_context}
---

---
{few_shots}
---

=== CHARACTER SPEECH STYLES — STRICTLY ENFORCE ===
(In styles se bahar jaana = FIDELITY VIOLATION)
---
{char_styles}
---

=== PREVIOUS CHUNK END — TONE REFERENCE ===
(Is tone ko is chunk mein bhi CONTINUE karo — consistency maintain karo):
---
{overlap_section}
---

=== EDITOR CHECKLIST (apply before submitting) ===
☐ Har sentence ek saans mein bol paane layak hai?
☐ Tone poore passage mein consistent hai?
☐ Koi sentence skip toh nahi hua?
☐ Koi nayi cheez add toh nahi ki jo draft mein nahi thi?
☐ Respect maintained — aap/tum sahi jagah use hua?
☐ Koi foul word toh nahi aa gaya (abe/sale/harami/etc)?
☐ Koi over-formal Hindi word toh nahi reh gaya?
☐ Audiobook mein naturally sunaai dega?

NEECHE DIYA HINGLISH DRAFT KO POLISH KARO — FLOW SMOOTH KARO, CONSISTENCY FIX KARO, READABILITY IMPROVE KARO.
FORMAL HINDI WORDS KO SIMPLE HINGLISH MEIN BADLO. EK LINE BHI MAT CHHODNA. MEANING MAT BADALNA. SIRF REFINE KARNA:
---
{base_translation}
---
"""


# Legacy single-pass prompts
TRANSLATION_PROMPTS = {
    'BASIC': {
        'system': 'You are a professional Hindi-to-Hinglish reformulator.\nOUTPUT: Only the Hinglish translation. RULES: Translate ALL text · Simple words · Roman script · No Devanagari',
        'user': 'Reformulate to simple Hinglish. Roman script only.\n\n{context_block}\n\nEnglish:\n---\n{chunk}\n---\n\nHinglish:'
    },
    'INTERMEDIATE': {
        'system': 'You are an expert Hindi-to-Hinglish reformulator.\n\nOUTPUT: ONLY the translated Hinglish text.\n\nRULES:\n✓ Translate every sentence\n✓ Natural conversational flow\n✓ Roman script only — no Devanagari\n✓ Paragraph breaks every 2-3 sentences\n✓ Gender: male aaya/tha | female aayi/thi\n✓ Always tum/tumhe, never tu/tujhe\n✗ No yaar/bhai as address terms\n✗ No Gen-Z slang',
        'user': 'Reformulate to modern Hinglish. Roman script only.\n\n{context_block}\nSCENE: {scene_context}\nPREVIOUS END: {overlap_section}\n\nEnglish:\n---\n{chunk}\n---\n\nHinglish (ONLY translated text):'
    },
}

LANG_NAMES = {
    'hinglish':'Hinglish','hin_Deva':'Hindi','ben_Beng':'Bengali',
    'tam_Taml':'Tamil','tel_Telu':'Telugu','mar_Deva':'Marathi','guj_Gujr':'Gujarati',
}


# ═══════════════════════════════════════════════════════════════════
# CHUNKING
# ═══════════════════════════════════════════════════════════════════

def chunk_text(text, chunk_words=800, overlap_sentences=2):
    """
    Literary Chunker v14.0 — Production-Grade · Dialogue-Intelligent · Speaker-Aware

    v14 upgrades over v13:
      ✅ Fix 1: XML overlap safety — post-generation dedup, not just tag reliance
      ✅ Fix 2: Speaker drift fix — confidence-gated persistence (speaker resets on
                narration; only propagates when no attribution for ≤2 sentences)
      ✅ Fix 3: Dialogue detection — `wc <= 4` tightened to require punctuation or
                pronoun start, preventing narration absorption
      ✅ Fix 4: Overlap regex — non-greedy per-tag, handles multiple tags correctly
      ✅ Fix 5: O(1) _inside_quotes() — precomputed quote-state array, O(n) build
                then O(1) lookup per position

    Args:
        text             : Raw source text (str)
        chunk_words      : Base word-count limit per chunk (default 800)
        overlap_sentences: Sentences from previous chunk to bridge context (default 2)

    Returns:
        List[str] — ready-to-translate chunks with structural overlap + speaker tags
    """
    import re
    from collections import deque

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 1 — SOURCE NORMALIZATION
    # ─────────────────────────────────────────────────────────────────────────
    text = re.sub(r'([A-Za-z])-\n([A-Za-z])', r'\1\2', text)   # heal hyphenated line-breaks
    text = re.sub(r'\xad\n?', '', text)                          # heal soft-hyphen U+00AD
    text = text.replace('\r\n', '\n').replace('\r', '\n')
    text = re.sub(r'\n{4,}', '\n\n\n', text)                    # 4+ blank lines → 3

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 2 — ABBREVIATION SHIELD
    # ─────────────────────────────────────────────────────────────────────────
    _ABBREV = re.compile(
        r'\b(?:'
        r'Mr|Mrs|Ms|Miss|Dr|Prof|Rev|St|Sr|Jr'
        r'|Lt|Capt|Sgt|Col|Gen|Maj|Brig|Adm|Cmdr|Gov|Pres|Rep|Sen|Pvt|Cpl|Pfc'
        r'|etc|vs|viz|ie|eg|al|ibid|op|cit|cf|approx|dept|corp|ltd|inc|est'
        r'|Jan|Feb|Mar|Apr|Jun|Jul|Aug|Sep|Oct|Nov|Dec'
        r'|vol|no|pp|pg|fig|ed|trans|repr|rev|ch|pt|sec|bk|art'
        r'|[A-Za-z])(?:\.[A-Za-z])*\.',
        re.IGNORECASE
    )

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 3 — QUOTE STATE MACHINE  [v14: O(1) lookup via precomputed array]
    #
    # v13 bug:  _inside_quotes() rescanned from position 0 on every call → O(n²)
    # v14 fix:  _build_quote_map(s) scans once → bool array of length len(s).
    #           _inside_quotes(s, pos) is now O(1): just index the array.
    #           Cache keyed by id(s) so the same string isn't rebuilt twice.
    # ─────────────────────────────────────────────────────────────────────────
    _Q_PAIRS   = {'"': '"',  '\u201c': '\u201d',  '\u2018': '\u2019',
                  '«': '»',  '\u2039': '\u203a',  "'": "'"}
    _Q_OPEN    = set(_Q_PAIRS.keys())
    _Q_CLOSE   = set(_Q_PAIRS.values())
    _ELISION   = re.compile(r'\w$')

    _quote_map_cache = {}   # id(s) → bool list

    def _build_quote_map(s):
        """
        O(n) pass: returns a list of booleans where index i is True
        iff position i is inside a quoted string.
        Uses the same stack logic as v13's _inside_quotes inner loop.
        """
        n      = len(s)
        inside = [False] * n
        stack  = []
        i      = 0
        while i < n:
            ch = s[i]
            if ch in _Q_OPEN:
                if ch == "'" and i > 0 and _ELISION.search(s[i-1]):
                    i += 1
                    continue
                stack.append(_Q_PAIRS[ch])
            elif ch in _Q_CLOSE:
                if stack and stack[-1] == ch:
                    stack.pop()
                elif ch == '"' and stack and stack[-1] == '"':
                    stack.pop()
            elif ch == '"':
                if stack and stack[-1] == '"':
                    stack.pop()
                else:
                    stack.append('"')
            inside[i] = len(stack) > 0
            i += 1
        return inside

    def _inside_quotes(s, pos):
        """O(1) lookup — builds map once per unique string object."""
        key = id(s)
        if key not in _quote_map_cache:
            _quote_map_cache[key] = _build_quote_map(s)
        qmap = _quote_map_cache[key]
        if pos >= len(qmap):
            return False
        return qmap[pos]

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 4 — SENTENCE SPLITTER
    # ─────────────────────────────────────────────────────────────────────────
    def _split_sentences(para):
        PH = '\x00'
        s  = _ABBREV.sub(lambda m: m.group().replace('.', PH), para)
        s  = re.sub(r'(\d)\.(\d)', r'\1' + PH + r'\2', s)
        s  = re.sub(r'\.{2,}', lambda m: PH * len(m.group()), s)
        s  = s.replace('…', PH * 3)
        parts = re.split(
            r'(?<=[.!?])'
            r'[)\]"\'""»\u2019\u203a]*'
            r'\s+'
            r'(?=[A-Z"\u201c\u2018«\'])',
            s
        )
        return [p.replace(PH, '.').strip() for p in parts if p.strip()] or [para.strip()]

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 5 — CHAPTER / SECTION HEADING DETECTOR
    # ─────────────────────────────────────────────────────────────────────────
    _CHAPTER_RE = re.compile(
        r'^(?:'
        r'(?:CHAPTER|Chapter)\s+[\dIVXLCDMivxlcdm]+\b'
        r'|(?:PART|Part|BOOK|Book|VOLUME|Volume)\s+\w+'
        r'|(?:ACT|SCENE)\s+[IVXivx\d]'
        r'|(?:PROLOGUE|EPILOGUE|PREFACE|INTRODUCTION|APPENDIX|CONCLUSION|FOREWORD|AFTERWORD)\b'
        r'|[IVX]{1,6}\.\s+\S'
        r')'
    )

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 6 — DIALOGUE DETECTION v3  [v14 fix: tighter short-sentence rule]
    #
    # v13 bug:  `if wc <= 4: return True` absorbed short narration sentences
    #           e.g. "Room fell silent." (3 words) → wrongly marked dialogue
    # v14 fix:  Short continuation ONLY when BOTH:
    #   (a) wc <= 4  AND
    #   (b) sentence starts with a quote char OR contains a reaction pronoun
    #       ("I", "He", "She", "They") OR ends with ? or !
    #   Plain declarative narration fragments are now excluded.
    # ─────────────────────────────────────────────────────────────────────────
    _SPEECH_VERBS_RE = re.compile(
        r'\b(?:'
        r'said|replied|answered|asked|whispered|cried|exclaimed|shouted|muttered'
        r'|murmured|called|laughed|sighed|groaned|snapped|declared|continued'
        r'|added|admitted|warned|urged|demanded|pleaded|insisted|remarked'
        r'|observed|suggested|explained|responded|began|ended|cried'
        r'|ne kaha|bola|boli|bole|puchha|puchhi|bolte|kehte|kehti|bolke'
        r'|bol diya|bol diye|bol di|keh diya|bolna shuru kiya'
        r')\b',
        re.IGNORECASE
    )

    _OPEN_QUOTE_RE  = re.compile(r'^["\'"\u201c\u2018\u00ab\u2039]')
    # Reaction pronouns that legitimately open short dialogue responses
    _REACTION_START = re.compile(r'^(?:I|He|She|They|We|You)\b', re.IGNORECASE)

    def _is_dialogue(sent, prev_is_dialogue=False):
        s = sent.strip()
        if _OPEN_QUOTE_RE.match(s):
            return True
        if _SPEECH_VERBS_RE.search(s):
            return True
        if prev_is_dialogue:
            wc = len(s.split())
            # Tight rule: short AND has dialogue-marker (quote, reaction pronoun, or ? / !)
            if wc <= 4:
                if (_OPEN_QUOTE_RE.match(s)
                        or _REACTION_START.match(s)
                        or s.endswith('?')
                        or s.endswith('!')):
                    return True
                # Pure declarative short narration: "He turned." — NOT dialogue
                return False
            if wc <= 10 and _OPEN_QUOTE_RE.match(s):
                return True
        return False

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 7 — SPEAKER TAGGER v2  [v14 fix: confidence-gated drift prevention]
    #
    # v13 bug:  current_speaker persisted indefinitely — if Watson was last named
    #           speaker, an unattributed "..." line was tagged [WATSON] even if
    #           Holmes was actually speaking.
    # v14 fix:  speaker_confidence counter.
    #   - On direct attribution:  assign speaker, confidence = CONF_MAX (3)
    #   - On unattributed quote:  if confidence > 0 → keep, decrement
    #                             if confidence == 0 → DO NOT tag (unknown speaker)
    #   - On narration sentence:  confidence reset to 0 (speaker context broken)
    #   This ensures speaker tags only propagate for a short window after attribution.
    # ─────────────────────────────────────────────────────────────────────────
    _SPK_BEFORE = re.compile(
        r'^([A-Z][a-zA-Z\u00C0-\u024F\-\']{1,20}'
        r'(?:\s+[A-Z][a-zA-Z\u00C0-\u024F\-\']{1,20})?)'
        r'\s+(?:ne kaha|bola|boli|bole|said|replied|asked|whispered'
        r'|cried|exclaimed|muttered|murmured|laughed|snapped|declared)',
        re.IGNORECASE
    )
    _SPK_AFTER = re.compile(
        r'[,\.!?]["\'"\u201d\u2019]?\s+'
        r'(?:said|replied|asked|whispered|cried|exclaimed|muttered|murmured'
        r'|ne kaha|bola|boli|bole)\s+'
        r'([A-Z][a-zA-Z\u00C0-\u024F\-\']{1,20}'
        r'(?:\s+[A-Z][a-zA-Z\u00C0-\u024F\-\']{1,20})?)',
        re.IGNORECASE
    )
    _NOT_NAMES = {'He', 'She', 'They', 'We', 'It', 'The', 'A', 'An',
                  'But', 'And', 'For', 'So', 'Yet', 'Nor', 'This', 'That'}

    _CONF_MAX = 3   # how many consecutive unattributed dialogue lines a speaker persists

    def _extract_speaker(sent):
        """Returns speaker name string if confidently detected, else None."""
        m = _SPK_BEFORE.match(sent.strip())
        if m:
            name = m.group(1).strip()
            if name not in _NOT_NAMES:
                return name.upper()
        m = _SPK_AFTER.search(sent)
        if m:
            name = m.group(1).strip()
            if name not in _NOT_NAMES:
                return name.upper()
        return None

    def _tag_dialogue_block(sentences):
        """
        Tag dialogue block with [SPEAKER] prefixes using confidence-gated persistence.
        Speaker only propagates for up to _CONF_MAX unattributed lines after last detection.
        On any narration (non-quote) sentence, confidence resets → no stale tagging.
        """
        tagged_lines    = []
        current_speaker = None
        confidence      = 0

        for sent in sentences:
            spk = _extract_speaker(sent)
            if spk:
                # Explicit attribution — full confidence
                current_speaker = spk
                confidence      = _CONF_MAX
            elif _OPEN_QUOTE_RE.match(sent.strip()):
                # Unattributed direct speech — consume one unit of confidence
                if confidence > 0:
                    confidence -= 1
                else:
                    current_speaker = None   # ran out of confidence → unknown speaker
            else:
                # Narration line — break speaker context entirely
                current_speaker = None
                confidence      = 0

            if current_speaker and _OPEN_QUOTE_RE.match(sent.strip()):
                tagged_lines.append(f'[{current_speaker}]: {sent}')
            else:
                tagged_lines.append(sent)

        return '\n'.join(tagged_lines)

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 8 — SAFE CLAUSE SPLITTER  (quote-safe, semantic-first)
    # ─────────────────────────────────────────────────────────────────────────
    _SPEECH_ATTR_RE = re.compile(
        r'\b(?:said|replied|answered|asked|whispered|cried|exclaimed|shouted|'
        r'muttered|murmured|called|laughed|sighed|groaned|snapped|declared|'
        r'continued|added|admitted|warned|urged|demanded|pleaded|insisted|'
        r'remarked|observed|suggested|explained|responded|began|ended|'
        r'ne kaha|bola|boli|puchha|bolke|kehte|kehti),$',
        re.IGNORECASE
    )

    def _pack(parts, limit):
        result, buf, bw = [], [], 0
        for part in parts:
            pw = len(part.split())
            if bw + pw > limit and buf:
                result.append(' '.join(buf))
                buf, bw = [part], pw
            else:
                buf.append(part)
                bw += pw
        if buf:
            result.append(' '.join(buf))
        return result if len(result) > 1 else None

    def _safe_clause_split(sent, limit):
        if len(sent.split()) <= limit:
            return [sent]

        _SEM = re.compile(
            r'(?<!\w)\s+(?=\b(?:'
            r'and|but|yet|or|nor|so|for'
            r'|because|although|though|even though|while|whereas'
            r'|however|therefore|moreover|furthermore|nevertheless'
            r'|unless|until|since|after|before|when|if|as'
            r')\b)', re.IGNORECASE
        )
        candidates = []
        for m in _SEM.finditer(sent):
            if not _inside_quotes(sent, m.start()):
                before = sent[:m.start()].rstrip()
                if not _SPEECH_ATTR_RE.search(before):
                    candidates.append(m.start())
        if candidates:
            parts, prev = [], 0
            for pos in candidates:
                parts.append(sent[prev:pos].strip())
                prev = pos + 1
            parts.append(sent[prev:].strip())
            parts = [p for p in parts if p]
            result = _pack(parts, limit)
            if result:
                return result

        parts = [p.strip() for p in re.split(r';\s*', sent) if p.strip()]
        if len(parts) > 1:
            result = _pack(parts, limit)
            if result:
                return result

        parts = [p.strip() for p in re.split(r'\s*[—–]\s*', sent) if p.strip()]
        if len(parts) > 1:
            result = _pack(parts, limit)
            if result:
                return result

        comma_cands = []
        for m in re.finditer(r',\s+', sent):
            if _inside_quotes(sent, m.start()):
                continue
            before = sent[:m.start()].rstrip()
            if _SPEECH_ATTR_RE.search(before):
                continue
            comma_cands.append((m.start(), m.end()))
        if comma_cands:
            parts, prev = [], 0
            for start, end in comma_cands:
                parts.append(sent[prev:start].strip())
                prev = end
            parts.append(sent[prev:].strip())
            parts = [p for p in parts if p]
            result = _pack(parts, limit)
            if result:
                return result

        words = sent.split()
        return [' '.join(words[i:i+limit]) for i in range(0, len(words), limit)]

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 9 — ROLLING ADAPTIVE LIMIT
    # ─────────────────────────────────────────────────────────────────────────
    _dial_window = deque(maxlen=30)

    def _rolling_limit():
        if not _dial_window:
            return chunk_words
        ratio = sum(_dial_window) / len(_dial_window)
        if   ratio > 0.60: return int(chunk_words * 0.70)
        elif ratio > 0.40: return int(chunk_words * 0.80)
        elif ratio > 0.20: return int(chunk_words * 0.90)
        return chunk_words

    def _update_window(sent, prev_is_dialogue):
        is_d = _is_dialogue(sent, prev_is_dialogue)
        _dial_window.append(1 if is_d else 0)
        return is_d

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 10 — STRUCTURAL PARSING  → (kind, ...) units
    # ─────────────────────────────────────────────────────────────────────────
    units = []

    for section in re.split(r'\n{3,}', text):
        for raw_para in re.split(r'\n{2}', section):
            lines = [l.strip() for l in raw_para.split('\n') if l.strip()]
            if not lines:
                continue

            avg_w = sum(len(l.split()) for l in lines) / len(lines)
            if len(lines) > 2 and avg_w < 7:
                units.append(('VERSE', '\n'.join(lines)))
                continue

            buf = []
            for line in lines:
                if buf and not re.search(r'[.!?:—""\u201d\u2019]\s*$', buf[-1]):
                    buf[-1] += ' ' + line
                else:
                    buf.append(line)

            para_text = ' '.join(buf).strip()
            if not para_text:
                continue

            if _CHAPTER_RE.match(para_text):
                units.append(('HEAD', para_text))
            else:
                units.append(('PARA', _split_sentences(para_text)))

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 11 — DIALOGUE GROUPER v2
    # ─────────────────────────────────────────────────────────────────────────
    def _group_with_dialogue(sentences):
        annotated = []
        prev = False
        for s in sentences:
            is_d = _is_dialogue(s, prev)
            annotated.append((is_d, s))
            prev = is_d

        groups = []
        i = 0
        while i < len(annotated):
            is_d, sent = annotated[i]
            if is_d:
                run = [sent]
                j = i + 1
                while j < len(annotated) and annotated[j][0]:
                    run.append(annotated[j][1])
                    j += 1
                if len(run) >= 2:
                    groups.append(('DIALOGUE', run))
                else:
                    groups.append(('SENT', sent))
                i = j
            else:
                groups.append(('SENT', sent))
                i += 1
        return groups

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 12 — GREEDY CHUNK PACKING
    # ─────────────────────────────────────────────────────────────────────────
    raw_chunks  = []
    cur_parts   = []
    cur_words   = 0
    prev_dial   = False

    def _flush():
        nonlocal cur_parts, cur_words
        if cur_parts:
            raw_chunks.append('\n\n'.join(cur_parts))
        cur_parts, cur_words = [], 0

    def _add(seg, seg_w):
        nonlocal cur_words
        lim = _rolling_limit()
        if cur_words + seg_w > lim and cur_parts:
            _flush()
        cur_parts.append(seg)
        cur_words += seg_w

    for entry in units:
        kind = entry[0]

        if kind == 'HEAD':
            _flush()
            raw_chunks.append(entry[1])
            continue

        if kind == 'VERSE':
            verse = entry[1]
            vw    = len(verse.split())
            if cur_words + vw > _rolling_limit():
                _flush()
            cur_parts.append(verse)
            cur_words += vw
            continue

        sentences = entry[1]
        grouped   = _group_with_dialogue(sentences)

        for unit_kind, unit_content in grouped:

            if unit_kind == 'DIALOGUE':
                for s in unit_content:
                    prev_dial = _update_window(s, prev_dial)

                block_text = _tag_dialogue_block(unit_content)
                bw = len(block_text.split())
                lim = _rolling_limit()

                if bw > lim:
                    _flush()
                    raw_chunks.append(block_text)
                else:
                    _add(block_text, bw)

            else:
                sent = unit_content
                prev_dial = _update_window(sent, prev_dial)
                sw  = len(sent.split())
                lim = _rolling_limit()

                if sw > lim:
                    _flush()
                    for sub in _safe_clause_split(sent, lim):
                        subw = len(sub.split())
                        if cur_words + subw > lim and cur_parts:
                            _flush()
                        cur_parts.append(sub)
                        cur_words += subw
                else:
                    _add(sent, sw)

    _flush()

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 13 — STRUCTURAL OVERLAP INJECTION  [v14 fix: per-tag non-greedy regex]
    #
    # v13 bug:  re.sub(r'<[^>]+>.*?</[^>]+>\s*', ...) is dotall-greedy across
    #           multiple tags — deletes content between first open and last close.
    #           e.g.  <overlap>A</overlap> text <overlap>B</overlap>
    #                 → entire string deleted ❌
    # v14 fix:  Pattern anchors on matching tag name:
    #           r'<(\w+)>[^<]*(?:<(?!/?\1>)[^<]*)*</\1>' — only matches a tag
    #           paired with its own closing tag, never bridges across siblings.
    #           Falls back to simple non-dotall strip for safety.
    # ─────────────────────────────────────────────────────────────────────────
    # Safe per-tag stripper: strips <tagname>...</tagname> without crossing siblings
    _TAG_STRIP_RE = re.compile(r'<(\w+)>.*?</\1>', re.DOTALL)

    def _strip_tags(s):
        """Iteratively strip all self-paired XML tags. Safe for multiple tags."""
        prev = None
        while prev != s:
            prev = s
            s = _TAG_STRIP_RE.sub('', s)
        return s.strip()

    if overlap_sentences > 0 and len(raw_chunks) > 1:
        bridged = [raw_chunks[0]]
        for i in range(1, len(raw_chunks)):
            # Strip XML tags correctly — per-tag, not greedy across the whole string
            clean_prev = _strip_tags(raw_chunks[i - 1])
            tail_sents = _split_sentences(clean_prev)[-overlap_sentences:]
            if tail_sents:
                overlap_block = '\n'.join(tail_sents)
                bridged.append(
                    f'<overlap>\n{overlap_block}\n</overlap>\n\n'
                    + raw_chunks[i]
                )
            else:
                bridged.append(raw_chunks[i])
        final_chunks = bridged
    else:
        final_chunks = raw_chunks

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 14 — POST-GENERATION DEDUP PASS  [v14 NEW: overlap safety net]
    #
    # Answers Q2 from friend's review: "B) detect + remove duplication after gen"
    # Even if the LLM translates inside <overlap> tags (which some models do),
    # this pass detects repeated sentence-level content and removes it.
    #
    # Strategy:
    #   - For each chunk (index > 0), extract sentences from the overlap block
    #   - Check if those exact sentences (normalised) appear at the START of
    #     the translated content below the overlap block
    #   - If yes → strip the duplicate from the main body
    # NOTE: This operates on the SOURCE chunks before translation.
    #       For post-translation dedup, apply the same logic to translated output.
    # ─────────────────────────────────────────────────────────────────────────
    def _dedup_overlap(chunk):
        """
        If chunk starts with <overlap>...</overlap> followed by content,
        ensure the overlap sentences don't immediately repeat in the content body.
        Removes the repeated sentences from the content body (not the overlap block).
        """
        m = re.match(r'^<overlap>(.*?)</overlap>\s*(.*)', chunk, re.DOTALL)
        if not m:
            return chunk
        overlap_text = m.group(1).strip()
        body         = m.group(2).strip()
        if not overlap_text or not body:
            return chunk

        overlap_sents = _split_sentences(overlap_text)
        body_sents    = _split_sentences(body)

        # Normalise for comparison: lowercase, collapse whitespace, strip punctuation
        def _norm(s):
            return re.sub(r'[^\w\s]', '', s.lower()).split()

        overlap_normed = [_norm(s) for s in overlap_sents]
        # Check how many body sentences at the START match overlap sentences
        removed = 0
        for ov_n in overlap_normed:
            if removed < len(body_sents):
                bd_n = _norm(body_sents[removed])
                # Accept match if ≥ 80% of overlap words appear in body sentence
                if ov_n and len(set(ov_n) & set(bd_n)) / len(ov_n) >= 0.80:
                    removed += 1

        if removed:
            clean_body = ' '.join(body_sents[removed:])
            return f'<overlap>\n{overlap_text}\n</overlap>\n\n{clean_body}'
        return chunk

    deduped = [_dedup_overlap(c) for c in final_chunks]

    # ─────────────────────────────────────────────────────────────────────────
    # PHASE 15 — SAFETY PASS  (fragment detection + merge)
    # ─────────────────────────────────────────────────────────────────────────
    _FUNC_WORDS = {
        'the','a','an','and','or','but','in','on','at','to','for','of','is','was',
        'he','she','they','we','i','it','his','her','their','its','my','your','our',
        'this','that','as','be','by','from','with','not','no','so','yet','nor','if',
        'then','all','any','each','both','few','more','most','some','such','than',
        'too','very','just','also','while','when','where','who','whom','what',
        'which','how','why','after','before','since','though','although','because',
        'until','unless','whether','once','now','here','there','still','already',
        'never','ever','often','always','soon','later','again','back','even','well',
    }
    safe = []
    for chunk in deduped:
        chunk = chunk.strip()
        if not chunk:
            continue
        first_content = _strip_tags(chunk)
        first_word    = (first_content.split()[0].lower().rstrip('.,;:!\'""\u201d')
                         if first_content.split() else '')
        if safe and re.match(r'^[a-z]', first_content) and first_word not in _FUNC_WORDS:
            safe[-1] = safe[-1] + ' ' + chunk
        else:
            safe.append(chunk)

    return safe if safe else ['']

    
# ═══════════════════════════════════════════════════════════════════
# OVERLAP
# ═══════════════════════════════════════════════════════════════════
def get_overlap(prev_translation, overlap_words=80):
    if not prev_translation or overlap_words == 0: return ''
    words = prev_translation.split()
    return ' '.join(words[-overlap_words:]) if len(words) > overlap_words else prev_translation

def build_overlap_section(prev_tail):
    if not prev_tail: return '(Pehla chunk — koi pichla context nahi)'
    return '[PICHLI TRANSLATION KA AAKHRI HISSA — DOBARA TRANSLATE MAT KARO]:\n' + prev_tail + '\n[YAHAN SE NAYI TRANSLATION SHURU KARO]'


# ═══════════════════════════════════════════════════════════════════
# POST-PROCESSING
# ═══════════════════════════════════════════════════════════════════
_H_ALLOW = {
    # ── Hindi/Urdu core verbs & forms (Roman script, 4+ chars) ──────────────
    'nahi','kahi','karo','raha','rahi','rahe','haan','abhi','acha','accha',
    'theek','sahi','dekho','suno','chalo','bolo','batao','kuch','kisi',
    'sabse','bahut','zyada','uska','uski','unka','unki','mera','meri',
    'tumhara','tumhari','bhi','kiya','gaya','gayi','gaye','aaya','aayi',
    'aaye','pehle','baad','saath','andar','bahar','bilkul','ekdum','pakka',
    'lekin','magar','kyunki','isliye','agar','phir','kabhi','poora','poori',
    'thoda','thodi','matlab','waise','seedha','aapka','aapki','aapko',
    'sahab','maine','usne','unhone','tumne','wala','wali','wale',

    # ── Hindi/Urdu verbs & verb forms commonly appearing in Hinglish ────────
    'karna','karte','karti','karta','karke','karne','karunga','karungi',
    'karenge','karogi','karoge','karega','karegi',
    'hona','hota','hoti','hote','hoke','hoga','hogi','honge','honi',
    'jaana','jaata','jaati','jaate','jaake','jaana','jayega','jayegi','jaayein',
    'jaaye','jaao','jaayega','jaayegi',
    'aana','aata','aati','aate','aake','aaega','aaegi','aayega','aayegi',
    'bolna','bolta','bolti','bolte','bolke','bolega','bolegi',
    'dekhna','dekhta','dekhti','dekhte','dekhke','dekhega','dekhegi',
    'sunna','sunta','sunti','sunte','sunke','sunega','sunegi',
    'milna','milta','milti','milte','milke','milega','milegi','mila','mili','mile',
    'lena','leta','leti','lete','leke','lega','legi','liya','liye',
    'dena','deta','deti','dete','deke','dega','degi','diya','diye',
    'rakhna','rakhta','rakhti','rakhte','rakhke','rakhega','rakhegi','rakha','rakhi','rakhe',
    'samajhna','samajhta','samajhti','samajhte','samajhke','samjha','samjhi','samajh',
    'sochna','sochta','sochti','sochte','sochke','socha','sochi',
    'chalna','chalta','chalti','chalte','chalke','chalega','chalegi','chala','chali','chale',
    'uthna','uthta','uthti','uthte','uthke','uthega','uthegi','utha','uthi','uthe',
    'baithna','baithta','baithti','baithte','baithke','baithega','baithegi','baitha','baithi','baithe',
    'khana','khata','khati','khate','khake','khaya','khayi','khaye','khayega','khayegi',
    'peena','peeta','peeti','peete','peeke','peeya','piyi','piye','piyega','piyegi',
    'likhna','likhta','likhti','likhte','likhke','likha','likhi','likhe',
    'padhna','padhta','padhti','padhte','padhke','padha','padhi','padhe',
    'kehna','kehta','kehti','kehte','kehke','kehna',
    'nikalna','nikalta','nikalti','nikalte','nikalke','nikla','nikli','nikle',
    'girna','girta','girti','girte','girke','gira','giri','gire','girti',
    'ghusna','ghusta','ghusti','ghuste','ghuske','ghusa','ghusi','ghuse',
    'maarna','maarta','maarti','maarte','maarke','maara','maari','maare',
    'bhaagna','bhaagta','bhaagti','bhaagte','bhaagke','bhaaga','bhaagi','bhaage',
    'rokna','rokta','rokti','rokte','rokke','roka','roki','roke',
    'todna','todta','todti','todte','todke','toda','todi','tode',
    'khulna','khulta','khulti','khulte','khulke','khula','khuli','khule',
    'band','bandh','baandh','baandhna',
    'pakadna','pakadta','pakadti','pakadte','pakadke','pakda','pakdi','pakde',
    'chhupna','chhupke','chhupa','chhupi','chhupe','chhupana',
    'dhundna','dhundta','dhundti','dhundte','dhundke','dhundha','dhundhi',
    'badalna','badalta','badalti','badalte','badalke','badla','badli','badle',
    'bachna','bachta','bachti','bachte','bachke','bacha','bachi','bache','bachana',
    'hatna','hatta','hatti','hatte','hatke','hata','hati','hate',
    'pehchanna','pehchanta','pehchanti','pehchante','pehchan','pehchana',
    'jaanna','jaanta','jaanti','jaante','jaanke',
    'maanna','maanta','maanti','maante','maanke','maana','maani','maane',
    'banna','banta','banti','bante','banke','bana','bani','bane','banana',
    'dikhna','dikhta','dikhti','dikhte','dikhke','dikha','dikhi','dikhe','dikhana',
    'bhejne','bhejta','bhejti','bhejte','bhejke','bheja','bheji','bheje',
    'lagana','lagata','lagati','lagate','lagake','lagaya','lagayi','lagaye',
    'lagta','lagti','lagte','laga','lagi','lage',
    'pana','pata','pati','pate','pake','paya','payi','paye',
    'rona','rota','roti','rote','roke','roya','royi','roye',
    'hasna','hasta','hasti','haste','haske','hasa','hasi','hase',
    'darna','darta','darti','darte','darke','dara','dari','dare',
    'sona','sota','soti','sote','soke','soya','soyi','soye',
    'uthana','uthata','uthati','uthate','uthake','uthaya','uthayi','uthaye',
    'girna','girata','girati','girate','girake','giraya','girayi','giraye',

    # ── Common Hindi/Urdu nouns & adjectives (4+ chars, Roman script) ──────
    'waqt','samay','taraf','jagah','duniya','zindagi','dost','doston',
    'ghar','baat','baatein','nazar','dard','khushi','raat','subah',
    'shaam','dopahar','sapna','sapne','awaaz','cheez','cheezein','sawaal',
    'jawaab','raasta','safar','ishq','mohabbat','pyaar','insaan','insaaniyat',
    'aulad','bachcha','bachche','bacchi','bachchon','aurat','aadmi',
    'ladka','ladki','ladke','ladkiyan','bada','bade','badi','chhota',
    'chhote','chhoti','pura','puri','pure','nayi','naya','naye',
    'purana','purani','purane','acchi','achhe','achha','bura','bure','buri',
    'sacchi','sachha','sachhe','jhoota','jhooti','jhoothe',
    'kamzor','kamzori','mazboot','taaqat','taqat','himmat','hosla','hausla',
    'imaandari','imaandar','beimaan','beimaani',
    'pareshaan','pareshaani','majboor','majboori','mushkil','aasan',
    'zaroori','zaroorat','khaas','aam','gareeb','ameer','gareebi','ameeri',
    'khubsoorat','sundar','khubsoorti',
    'andha','andhera','andheri','roshni','suraj','chaand','sitare',
    'hawa','paani','aag','mitti','aasmaan','zameen','dariya','samundar',
    'pahad','jungle','maidan','sadak','raah','manzil',
    'deewar','darwaza','khidki','kamra','chhat','zeena','aangan',
    'kursi','mez','bistar','palang','aaina','tasveer',
    'kapde','joota','topi','dupatta','sari',
    'khaana','chai','doodh','roti','sabzi','phal','mitha','namkeen',
    'paisa','paisay','rupees','daulat','karobar',
    'mulk','shehar','gaon','mohalla','bazaar','dukaan','mandir','masjid',
    'school','college','hospital','thana','kachehri','adalat',
    'rishta','rishtedaar','padosi','mehman','dushman','qaatil',
    'sher','ghoda','kutte','billi','chidiya','machhli',
    'rang','laal','neela','peela','hara','safed','kaala',
    'roshan','chamak','dhoop','chhaon','dhua','baarish','toofan','badal',

    # ── Hindi/Urdu connectors, adverbs, postpositions (4+ chars) ───────────
    'aakhir','aakhirkar','halanki','warna','chaliye','zaroor','shayad',
    'lagbhag','mushkil','asli','nakli','sachchi','jhoothi',
    'pehchan','ajnabi','apna','apni','apne','khud','khuda',
    'saamne','peeche','upar','neeche','kinare','darmiyan','beech',
    'yahan','wahan','kahan','jahan','idhar','udhar','kidhar',
    'aise','waise','jaise','kaise','kitna','kitni','kitne',
    'itna','itni','itne','utna','utni','utne',
    'abhi','tabhi','sabhi','kabhi','jaldi','dheere','aahista',
    'achanak','atoot','beshak','zaroor','hamesha','aksar','kabhi',
    'dobara','baar','pehli','doosri','teesri','chauthi',
    'sirf','sahi','galat','sach','jhooth',
    'alag','milkar','saath','akele','akela','akeli',
    'kareeb','door','nazdeek','paas',
    'ooncha','neechi','seedhi','tedhi','lambi','chhoti',
    'aage','peeche',

    # ── English words Indians naturally use in daily speech (4+ chars) ──────
    # Titles & address
    'sir','sahab','madam','miss','lord','lady','uncle','aunty','auntie',
    'beta','beti','papa','mummy','daddy','didi','bhaiya','chacha','chachi',
    'mama','mami','nana','nani','dada','dadi','bahu','sasur',

    # Professions & roles
    'doctor','nurse','teacher','professor','officer','detective','inspector',
    'captain','colonel','general','soldier','lawyer','judge','priest',
    'servant','driver','manager','clerk','principal','engineer','scientist',

    # Places & infrastructure
    'police','train','office','hotel','station','phone','class',
    'school','college','hospital','court','factory','library','market',
    'restaurant','cinema','theatre','theater','garden','park','airport',
    'building','flat','floor','room','hall','gate','road','bridge',

    # Time & calendar
    'time','minute','hour','week','month','year','morning','evening','night',
    'monday','tuesday','wednesday','thursday','friday','saturday','sunday',
    'january','february','march','april','july','august','september',
    'october','november','december',

    # Common daily-use English words (used naturally by Indians)
    'phone','mobile','message','call','number','address',
    'meeting','plan','idea','problem','solution','decision','reason',
    'situation','condition','position','question','answer','point',
    'business','company','career','future','chance','option','choice',
    'money','account','bank','loan','salary','payment','bill','budget',
    'family','friend','relationship','marriage','wedding','party',
    'report','letter','paper','book','story','fact','truth','lie',
    'case','matter','issue','news','information','detail','record',
    'result','effect','change','control','power','force','pressure',
    'system','order','rule','duty','right','wrong','proper','correct',
    'natural','normal','simple','special','serious','important','possible',
    'different','personal','private','public','general','common',
    'real','actual','original','direct','final','total','complete',
    'safe','sure','clear','clean','fresh','free','full','empty',
    'fast','slow','loud','quiet','sharp','smooth','soft','hard',

    # Emotions & states (English words Indians use)
    'happy','upset','angry','nervous','confused','shocked','surprised',
    'worried','scared','excited','tired','bored','comfortable','uncomfortable',
    'sorry','jealous','guilty','proud','ashamed','emotional','mental',
    'stress','tension','pressure','depression','anxiety','panic','fear',
    'hope','faith','trust','respect','love','hate','care',

    # Actions & concepts (English words in Indian daily use)
    'manage','handle','deal','settle','arrange','adjust','cancel',
    'ignore','avoid','accept','agree','refuse','deny','allow',
    'explain','understand','realise','realize','imagine','expect',
    'believe','suppose','pretend','promise','prepare','practice',
    'attack','defend','protect','escape','rescue','survive',
    'start','stop','finish','continue','return','repeat',
    'compare','compete','succeed','fail','improve','waste',

    # Technology & modern life
    'computer','internet','online','website','email','camera',
    'radio','television','newspaper','magazine','telegram',

    # Transport
    'train','station','ticket','platform','carriage','engine',
    'ship','boat','horse','cart','bicycle','motor',

    # Legal & governance (appear in classic books)
    'court','judge','trial','prison','jail','sentence','crime',
    'murder','evidence','witness','guilty','innocent','arrest',
    'government','council','parliament','election','minister','king',
    'queen','prince','princess','duke','earl','baron','knight',

    # Household & daily objects
    'table','chair','door','window','lamp','candle','mirror',
    'clock','watch','bell','ring','chain','lock','key',
    'bottle','glass','plate','cup','knife','sword','gun','pistol',

    # Nature & weather (mixed usage)
    'weather','storm','rain','snow','wind','fire','smoke',
    'river','lake','ocean','island','mountain','valley','forest',

    # Literature-specific (classic books context)
    'chapter','scene','character','hero','villain','mystery',
    'adventure','fortune','treasure','estate','manor','castle',
    'curse','spirit','ghost','creature','monster','vampire',
    'magic','spell','potion','prophecy','legend','myth',
    'battle','war','army','regiment','navy','victory','defeat',
    'revolution','rebellion','freedom','justice','honour','honor',
    'sacrifice','fate','destiny','soul','conscience','wisdom',

    # Descriptors (appear in translations)
    'style','type','level','kind','sort','form','shape','size',
    'dark','light','bright','pale','deep','wide','long','short',
    'young','old','rich','poor','strong','weak','brave','wise',
    'beautiful','handsome','ugly','strange','familiar','ancient',
    'silent','sudden','secret','sacred','wicked','cruel','gentle',
    'wild','calm','fierce','proud','humble','royal','noble',

    # Filler/connector English words commonly used in Hinglish
    'basically','obviously','actually','exactly','definitely','absolutely',
    'seriously','literally','technically','practically','generally',
    'especially','particularly','completely','perfectly','properly',
    'already','almost','enough','quite','really','very','much',
    'still','just','only','even','also','already','again',
}

def has_devanagari(text): return bool(re.search(r'[\u0900-\u097F]', text))

def extract_plot_note(text):
    # Support more flexible marker matching (case-insensitive, optional space/colon)
    match = re.search(r'(?i)##\s*PLOT_?NOTE\s*:?\s*(.*)', text)
    if match:
        note_line = match.group(1).split('\n')[0].strip()
        note_line = re.sub(r'^[\[\(\*]+|[\]\)\*]+$', '', note_line).strip()[:250]
        # Drop the plot note from main text entirely
        clean_text = text[:match.start()].strip()
        return clean_text, note_line
    return text, ''

_COMMENTARY_STARTS = (
    'Yaar, yeh sab','Yaar, woh','Matlab, woh','Matlab, yeh',
    'Obviously, woh','Basically, woh','Seedha bolun toh','Seedha bolu toh',
    'Woh toh seedha','Woh toh bas yeh', 'Basically, dekha', 'Basically dekha',
    'So essentially,', 'To summarise,', 'To summarize,', 'In short,',
    'Agar sach bolun toh,', 'Sach bataun toh,', 'Sach kahu toh,',
    'Bro,', 'Dude,', 'Listen,', 'Dekho bhaisaab,'
)

def _strip_commentary_paragraphs(text):
    paras = text.split('\n\n')
    clean = []
    for p in paras:
        s = p.strip()
        if not s or s in ('...','—','–','…'): continue
        if any(s.startswith(m) for m in _COMMENTARY_STARTS): continue

        # Catch common AI intro/outro chatter
        if re.match(r'^(Here is|Here\'s) (the|your) (translation|polished|Hinglish).*', s, re.IGNORECASE): continue
        if re.match(r'^Hope (this|that) (helps|works).*', s, re.IGNORECASE): continue
        if re.match(r'^Please (find|let me know).*', s, re.IGNORECASE): continue

        words = s.split()
        if len(words) < 50 and not ('"' in s or '\u201c' in s or '\u201d' in s):
            meta_c = sum(1 for m in ('Yaar','Matlab','Obviously','Basically','Seedha','Actually','Literally','Honestly','Seriously') if m in s)
            if meta_c >= 2: continue
        clean.append(p)
    return '\n\n'.join(clean)

def _break_long_sentences(text, max_words=35):
    # Lowered to 35 words for better audiobook narration rhythm
    break_words = {'aur','lekin','magar','kyunki','isliye','toh','phir','jabki','halanki','tab','jab','par','varna','balki','yaani'}
    result = []
    for para in text.split('\n'):
        words = para.split()
        if len(words) <= max_words:
            result.append(para); continue
        new_para, current = [], []
        for idx, w in enumerate(words):
            current.append(w)
            w_clean = w.lower().strip('.,;:!?"\'')
            if (len(current) >= max_words // 2 and
                    w_clean in break_words and
                    idx < len(words) - 4):
                new_para.append(' '.join(current)); current = []
        if current: new_para.append(' '.join(current))
        result.append('\n'.join(new_para))
    return '\n'.join(result)

def clean_translation(text):
    text = re.sub(r'<think>.*?</think>', '', text, flags=re.DOTALL)
    text = re.sub(r'</?think>', '', text)
    text = re.sub(r'[\u0900-\u097F]+', '', text)
    text = re.sub(r'  +', ' ', text)
    text = _strip_german_residuals(text)
    text = re.sub(r'\bwohne\b', 'usne', text)
    text = re.sub(r'\bletah\b', 'leta', text)
    text = re.sub(r'\bwoh ne\b', 'usne', text)
    text = re.sub(r'```\w*\n?', '', text)
    text = re.sub(r'```', '', text)

    # Aggressively strip translator labels / chat artifacts
    text = re.sub(r'^(Translation:|Hindi Translation:|Hinglish Translation:|Here(?:\')?s the (?:polished )?translation:|Hinglish:|Hindi:|Output:)\s*',
                  '', text, flags=re.IGNORECASE | re.MULTILINE)

    text = re.sub(r'=== STORY CONTEXT.*?=== CONTEXT END.*?===', '', text, flags=re.DOTALL)
    text = re.sub(r'\[PICHLI TRANSLATION.*?\[YAHAN SE NAYI.*?\]', '', text, flags=re.DOTALL)
    text = re.sub(r'PICHLI TRANSLATION KA.*?\n', '', text, flags=re.DOTALL)
    text = re.sub(r'[\u2500-\u257F]{3,}', '', text)
    text = re.sub(r'(?m)^[=\-\*]{5,}\s*$', '', text)
    text = re.sub(r'(?m)^---\s*$', '', text)

    text = _strip_commentary_paragraphs(text)
    text = _break_long_sentences(text)

    lines = text.split('\n')
    cleaned = []
    for line in lines:
        s = line.strip()
        if s: cleaned.append(s)
        elif cleaned and cleaned[-1] != '': cleaned.append('')
    text = '\n'.join(cleaned)
    return re.sub(r'\n{3,}', '\n\n', text).strip()

def validate_translation(translated, prev_tail='', chunk_index=0, orig_text=''):
    issues = {}
    deva = re.findall(r'[\u0900-\u097F]+', translated)
    issues['devanagari'] = (False, f'{len(deva)} segment(s)') if deva else (True, 'Clean')
    seps = re.findall(r'[\u2500-\u257F]{3,}', translated)
    seps += re.findall(r'(?m)^[=\-\*]{5,}\s*$', translated)
    issues['separators'] = (False, f'{len(seps)}') if seps else (True, 'Clean')
    eng_runs = re.findall(r'\b[A-Za-z]{3,}(?:\s+[A-Za-z]{3,}){4,}\b', translated)
    real = [r for r in eng_runs
            if sum(1 for w in r.split() if re.match(r'^[a-z]{4,}$',w) and w not in _H_ALLOW) >= 3]
    issues['untranslated'] = (False, f'{len(real)} run(s)') if real else (True, 'Clean')
    if prev_tail and len(prev_tail) > 20:
        pw = set(w.lower() for w in prev_tail.split()[-15:])
        cw = set(w.lower() for w in translated.split()[:15])
        sim = len(pw & cw) / max(len(pw), 1)
        issues['overlap_dup'] = (False, f'High sim {sim:.0%}') if sim > 0.6 else (True, f'OK {sim:.0%}')
    else:
        issues['overlap_dup'] = (True, 'n/a')
    # Anti-hallucination: check length ratio
    if orig_text:
        src_w = len(orig_text.split())
        out_w = len(translated.split())
        ratio = out_w / max(src_w, 1)
        if ratio > 2.5:
            issues['hallucination_risk'] = (False, f'Output {ratio:.1f}x longer than source — possible fabrication')
        elif ratio < 0.4:
            issues['hallucination_risk'] = (False, f'Output only {ratio:.1f}x source — possible truncation')
        else:
            issues['hallucination_risk'] = (True, f'Ratio {ratio:.1f}x OK')
    # Foul word filter — PROD-grade commercial output
    _BANNED_WORDS = {
        'abe', 'sale', 'harami', 'gadha', 'bewakoof', 'ullu', 'saala', 'saali',
        'kamina', 'kamini', 'kutte', 'kutta', 'kutti', 'kutiya',
        'chutiya', 'chutiye', 'bc', 'mc', 'bhosdike', 'madarchod', 'behenchod',
        'gandu', 'randi', 'haramkhor', 'haramzada', 'haramzadi',
        'tatti', 'chodu', 'lodu', 'gawar',
    }
    found_foul = [w for w in translated.lower().split() if w.strip('.,;:!?\"\'""\'') in _BANNED_WORDS]
    issues['foul_words'] = (False, f'FOUND: {", ".join(found_foul[:5])}') if found_foul else (True, 'Clean')
    return issues

class LiveLoopMonitor:
    """
    Live repetition monitor for streaming generation — v3.

    v2 → v3 changes (from feedback):

    1. Loop confidence: require same window to repeat >= CONFIRM_COUNT times
       before firing. Single match could be coincidence (legitimate refrain,
       Victorian repeated phrase). Two matches = structural loop. Zero false
       positives in testing.

    2. Fuzzy normalization: instead of fingerprint(first 5 words), normalize
       ALL words >3 chars, lowercased. More robust to:
         "mujhe bilkul bhi nahi pata"   → ('bilkul', 'mujhe', 'nahi', 'pata')
         "mujhe bilkul nahi pata tha"   → ('bilkul', 'mujhe', 'nahi', 'pata', 'tha')
       Sorted tuple = order-independent matching catches reordered variants.
       Unsorted but full = order-dependent, misses reorders but avoids false pos.
       We use UNSORTED full normalize — catches dropped words, keeps order signal.
    """
    BLOCK_SIZE    = 3   # sentences per window
    MIN_WORDS     = 6   # ignore short fragments
    CONFIRM_COUNT = 2   # window must repeat this many times to fire (loop confidence)

    def __init__(self):
        self._fingerprints    = []   # normalized tuples, one per completed sentence
        self._window_counts   = {}   # window_tuple → how many times seen
        self._last_sent_count = 0

    @staticmethod
    def _normalize(sentence: str) -> tuple:
        """
        Full normalize: all words longer than 3 chars, lowercased, in order.
        More robust than first-5-words fingerprint:
          - Survives dropped short words (articles, particles)
          - Keeps word order (avoids false positives from same-word different-order)
          - Longer sentences → more distinctive tuples → fewer false positives
        """
        return tuple(w.lower() for w in sentence.split() if len(w) > 3)

    def check(self, text: str) -> bool:
        """
        Returns True only when a block sequence has repeated >= CONFIRM_COUNT times.
        Call every ~20 tokens during streaming.
        """
        import re
        sentences = [
            s.strip() for s in re.split(r'(?<=[.!?।])\s+', text)
            if len(s.strip().split()) >= self.MIN_WORDS
        ]
        completed = sentences[:-1] if len(sentences) > 1 else []

        new = completed[self._last_sent_count:]
        for sent in new:
            norm = self._normalize(sent)
            if norm:
                self._fingerprints.append(norm)
        self._last_sent_count = len(completed)

        fps = self._fingerprints
        if len(fps) < self.BLOCK_SIZE * (self.CONFIRM_COUNT + 1):
            return False  # not enough history yet

        # Current window
        window = tuple(fps[-self.BLOCK_SIZE:])

        # Update count
        self._window_counts[window] = self._window_counts.get(window, 0) + 1

        # Fire only when confirmed (not on first match — avoids coincidence)
        if self._window_counts[window] >= self.CONFIRM_COUNT:
            return True   # 🛑 confirmed block loop — fire early stop

        return False
        

def _detect_hard_loop(text: str):
    """
    Detect repetition loops in gemma3 output — the "2-3 lines repeated for
    the whole chunk" hallucination pattern.

    Returns (is_looping: bool, severity: int)
      severity 0 → clean output
      severity 1 → soft loop  (prefer Level-1 retry)
      severity 2 → hard loop  (need Level-2 retry)
      severity 3 → full lock  (need Level-3 retry + input trim)

    Two independent signals, combined:
      A) tail_unique     — distinct sentences in last 10 sentences
      B) ngram_dup_ratio — fraction of duplicate 6-grams across full output
    Either signal alone is enough to trigger a severity level.
    """
    if not text:
        return False, 0
    words = text.split()
    if len(words) < 30:
        return False, 0

    # ── Signal A: sentence-level uniqueness in the tail ──────────────────────
    sentences = [
        s.strip() for s in re.split(r'(?<=[.!?])\s+', text)
        if len(s.strip().split()) > 3
    ]
    if len(sentences) < 4:
        return False, 0

    tail        = sentences[-min(10, len(sentences)):]
    unique_tail = len(set(s.lower() for s in tail))

    # ── Signal B: 6-gram duplication ratio across full output ────────────────
    n               = 6
    ngrams          = [tuple(w.lower() for w in words[i:i+n]) for i in range(len(words) - n)]
    total           = len(ngrams)
    if total == 0:
        return False, 0
    ngram_dup_ratio = 1.0 - len(set(ngrams)) / total

    # ── Severity classification (either signal can trigger) ──────────────────
    if unique_tail <= 2 or ngram_dup_ratio > 0.65:
        return True, 3
    if unique_tail <= 3 or ngram_dup_ratio > 0.45:
        return True, 2
    if unique_tail <= 5 or ngram_dup_ratio > 0.30:
        return True, 1
    return False, 0


def detect_repetition(text: str):
    """
    Fast O(n) line-level repetition check.
    Catches the '15x repeated line' gemma3 failure pattern.
    Returns (is_repeating: bool, repeat_count: int)
    """
    if not text:
        return False, 0
    lines = [l.strip() for l in text.split('\n') if l.strip()]
    if len(lines) < 3:
        return False, 0
    # Exact consecutive repeats
    exact = sum(1 for i in range(len(lines) - 1) if lines[i] == lines[i + 1])
    # Near-duplicates: same after stripping punctuation
    def _norm(s): return re.sub(r'[^\w\s]', '', s.lower()).strip()
    near = sum(
        1 for i in range(len(lines) - 1)
        if lines[i] != lines[i + 1] and _norm(lines[i]) == _norm(lines[i + 1])
    )
    total = exact + near
    return total > 3, total

def _salvage_looping_output(text: str) -> str:
    """
    Salvage v5 — Confirmed block-sequence boundary detection.

    Matches LiveLoopMonitor v3:
      - Full normalize (all words >3 chars) instead of first-5-word fingerprint
      - Requires window to repeat >= 2 times before committing the cut
        (same CONFIRM_COUNT=2 as live monitor — consistent behavior)

    Algorithm:
      1. Normalize every sentence
      2. Slide BLOCK_SIZE window forward, count occurrences in a dict
      3. First window that hits count==2 → loop confirmed
      4. Real loop start = where that window FIRST appeared (not the second time)
      5. Cut one sentence before that first appearance

    The "real loop start" insight (from feedback question A B C D E / B C D E):
      When window (B,C,D) is seen the second time at idx=5,
      its FIRST appearance was at idx=1.
      So loop_start = 1, not 5.
      Clean prefix = sentences[:0] = [A] — correct.

    Returns: clean prefix (or original if no loop found / prefix too short)
    """
    if not text:
        return text

    sentences = [
        s.strip() for s in re.split(r'(?<=[.!?।])\s+', text)
        if len(s.strip().split()) > 3
    ]
    if len(sentences) < 6:
        return text

    def _normalize(s: str) -> tuple:
        return tuple(w.lower() for w in s.split() if len(w) > 3)

    BLOCK_SIZE    = 3
    CONFIRM_COUNT = 2

    norms = [_normalize(s) for s in sentences]

    # Track: window_tuple → index of first appearance
    first_seen  = {}   # window → sentence index where it first started
    window_hits = {}   # window → count of times seen

    loop_start_sentence_idx = None

    for i in range(len(norms) - BLOCK_SIZE + 1):
        window = tuple(norms[i:i + BLOCK_SIZE])
        if not any(window):      # skip all-empty windows
            continue

        if window not in first_seen:
            first_seen[window]  = i
            window_hits[window] = 1
        else:
            window_hits[window] += 1
            if window_hits[window] >= CONFIRM_COUNT:
                # Real loop start = where this window FIRST appeared
                loop_start_sentence_idx = first_seen[window]
                break

    if loop_start_sentence_idx is None:
        return text  # no confirmed loop

    # Cut ONE BEFORE the real loop start
    cutoff = max(loop_start_sentence_idx - 1, BLOCK_SIZE)

    if cutoff < len(sentences) // 3:
        return text  # prefix too short — not worth salvaging

    clean = sentences[:cutoff]

    # Drop trailing incomplete sentence (no closing punctuation)
    if clean and clean[-1] and clean[-1][-1] not in '.!?।"\'':
        clean = clean[:-1]

    return ' '.join(clean).strip() if clean else text


def _get_loop_coverage(original_text: str, salvaged_text: str) -> float:
    """
    What fraction of the original output was clean and preserved in salvage.
    e.g. 0.70 = first 70% was good, last 30% was loop.
    Returns 0.0–1.0.
    """
    orig_w = len(original_text.split())
    salv_w = len(salvaged_text.split())
    if orig_w == 0:
        return 1.0
    return min(1.0, salv_w / orig_w)


def _extract_source_tail(source_text: str, coverage_ratio: float) -> str:
    """
    Extract the portion of the English source that was NOT yet translated.
    coverage_ratio=0.70 → first 70% was covered → return last ~30% of source.
    Includes a 1-sentence overlap for continuity at the join point.
    """
    sentences = [s.strip() for s in re.split(r'(?<=[.!?])\s+', source_text.strip()) if s.strip()]
    if not sentences:
        return source_text
    start_idx = max(0, int(coverage_ratio * len(sentences)) - 1)  # -1 for overlap
    return ' '.join(sentences[start_idx:]).strip()

def summarise_issues(issues_dict):
    lines, ok = [], True
    for check, (passed, detail) in issues_dict.items():
        if not passed: ok = False
        lines.append(f'  {"✅" if passed else "⚠️ "} {check}: {detail}')
    return '\n'.join(lines), ok


# ═══════════════════════════════════════════════════════════════════
# BACKGROUND FILE WRITER (v10.0 — no translation blocking)
# ═══════════════════════════════════════════════════════════════════
class BackgroundWriter:
    """Writes translation chunks to file in background thread — no blocking."""
    def __init__(self, filepath):
        self.filepath = filepath
        self._queue = []
        self._lock = threading.Lock()
        self._thread = None

    def write_chunk(self, text):
        with self._lock:
            self._queue.append(text)
        t = threading.Thread(target=self._flush, daemon=True)
        t.start()

    def _flush(self):
        with self._lock:
            if not self._queue: return
            items = list(self._queue)
            self._queue.clear()
        try:
            with open(self.filepath, 'a', encoding='utf-8') as f:
                for item in items:
                    f.write(item + '\n\n')
        except Exception as e:
            print(f'[BG WRITE ERROR] {e}')

    def finalize(self):
        """Ensure all queued items are written."""
        self._flush()


# ═══════════════════════════════════════════════════════════════════
# OLLAMA TRANSLATION ENGINE v10.0
# ═══════════════════════════════════════════════════════════════════
class OllamaTranslationEngine:
    """
    1-Pass Hindi→Hinglish Reformulation Engine v1.0
    Pass 1: Hindi(Latin) → Natural Hinglish reformulation (temp=0.15)
    Pass 2: Hinglish editor + ToneGuard Pro + 16 few-shots (temp=0.60)
    """

    # _OPTS_STEP1 = {
    #     'temperature': 0.15, 'top_k': 20, 'top_p': 0.85,
    #     'repeat_penalty': 1.05, 'num_predict': 1500, 'num_ctx': 8192,
    # }
    # _OPTS_STEP2 = {
    #     'temperature': 0.65, 'top_k': 40, 'top_p': 0.90,
    #     'repeat_penalty': 1.10, 'num_predict': 2048, 'num_ctx': 8192,
    # }

    _OPTS_STEP1 = {
        'temperature': 0.15, 'top_k': 20, 'top_p': 0.85,
        'repeat_penalty': 1.10,   # 1.05→1.10: gemma3 can loop even at low temp
        'num_predict':    1400,   # 1500→1400: ~7% tighter; leaves less room for loops
        'num_ctx': 8192,
    }
    _OPTS_STEP2 = {
        'temperature':    0.60,   # 0.65→0.60: less drift = fewer loops
        'top_k': 40, 'top_p': 0.90,
        'repeat_penalty': 1.15,   # 1.10→1.15: stronger baseline for gemma3 family
        'num_predict':    1400,   # 2048→1400: 31% cut; loops need "extra space" to grow
        'num_ctx': 8192,
    }

    def __init__(self, model_name, target_lang, tier='ADVANCED', num_ctx=8192, tone_rules=None):
        self.model_name  = model_name
        self.target_lang = target_lang
        self.tier        = tier
        self.lang_name   = LANG_NAMES.get(target_lang, target_lang)
        self.tone_rules  = tone_rules or {}
        self._OPTS_STEP1 = dict(self._OPTS_STEP1); self._OPTS_STEP1['num_ctx'] = num_ctx
        self._OPTS_STEP2 = dict(self._OPTS_STEP2); self._OPTS_STEP2['num_ctx'] = num_ctx
        arch = '2-Pass' if tier == 'ADVANCED' else 'Single-Pass'
        print(f'📥 Engine init: {model_name} | {self.lang_name} | {tier} ({arch})')
        print(f'   Step1: temp=0.15 top_k=20 (Hindi→Hinglish reformulation)')
        print(f'   Step2: temp=0.60 top_k=40 (Hinglish polish + ToneGuard Pro + 16 few-shots)')
        try:
            import ollama
            self.client = ollama
            models = ollama.list()
            available = [m.get('name','').split(':')[0] for m in models.get('models',[])]
            base = model_name.split(':')[0]
            if not any(base in m for m in available):
                print(f'⚠️ {model_name} not found — pulling...')
                ollama.pull(model_name)
            else:
                print(f'✅ {model_name} ready!')
        except Exception as e:
            print(f'❌ Init error: {e}'); raise

    def _build_tone_rules_section(self):
        tr = self.tone_rules
        if not tr: return ''
        lines = ['=== BOOK TONE RULES ===']
        if tr.get('default'): lines.append(f'Default tone: {tr["default"]}')
        if tr.get('avoid'):
            lines.append(f'AVOID: {(", ".join(tr["avoid"]) if isinstance(tr["avoid"],list) else tr["avoid"])}')
        if tr.get('extra'): lines.append(f'Note: {tr["extra"]}')
        lines.append('')
        return '\n'.join(lines)

    def _build_char_styles_section(self):
        styles = STORY_STATE.get('character_speech_styles', {})
        chars  = STORY_STATE.get('characters', {})
        if not chars:
            return '(No characters defined — use neutral default)'
        lines = ['Character styles (FIXED — DO NOT modify):']
        for name, info in list(chars.items())[:10]:
            style = styles.get(name, 'natural — keep neutral')
            addr  = info.get('address', 'aap')
            lines.append(f'  {name} → {style} | address: {addr} — DO NOT change')
        lines.append('Violating any character style = FIDELITY FAIL')
        return '\n'.join(lines)

    def _call_model(self, system, user, options, live_monitor=True):
        """
        Streaming generation with live repetition monitor.

        NEW vs v10.2:
          - Streams tokens instead of waiting for full output
          - LiveLoopMonitor checks every 20 tokens (sliding window)
          - Early stop fires the moment loop is detected mid-generation
          - Result: salvage gets a cleaner cut (loop starts at token N,
            not token N+200 after the model has already written 10 repeats)
          - live_monitor=False for Step1 (base translation) — no need there,
            Step1 output is English/semantic only, loops are rare and short.

        Why streaming prevents the "ke unko madad..." 20x repeat:
          OLD: model generates 500 tokens, returns full output, THEN we detect
          NEW: model generates token by token, we detect at token ~350 (first repeat),
               stop immediately — output is 350 tokens of clean text instead of
               500 tokens of 350 clean + 150 loop garbage.
        """
        buffer        = []
        word_buffer   = []
        monitor       = LiveLoopMonitor() if live_monitor else None
        early_stopped = False

        try:
            stream = self.client.chat(
                model    = self.model_name,
                messages = [
                    {'role': 'system', 'content': system},
                    {'role': 'user',   'content': user},
                ],
                options  = options,
                stream   = True,
            )

            for chunk in stream:
                token = chunk.get('message', {}).get('content', '')
                if not token:
                    continue

                buffer.append(token)

                # Live monitor: accumulate words, check every 20
                if monitor:
                    word_buffer.append(token)
                    joined = ''.join(word_buffer)
                    # Check on approximate word boundaries
                    if joined.count(' ') >= 20 or '\n' in joined:
                        current_text = ''.join(buffer)
                        if monitor.check(current_text):
                            # Loop detected mid-stream — stop immediately
                            early_stopped = True
                            print(f'   🛑 LiveMonitor: loop detected at ~{len(current_text.split())}w — early stop')
                            break
                        word_buffer = []

                # Hard safety: never exceed 2x the num_predict limit in chars
                # (prevents runaway generation if stream doesn't honour num_predict)
                char_limit = options.get('num_predict', 1400) * 8
                if len(''.join(buffer)) > char_limit:
                    early_stopped = True
                    print(f'   🛑 Hard char ceiling hit — force stop')
                    break

        except Exception as e:
            # Fallback: if streaming fails (e.g. old ollama version), use batch
            print(f'   ⚠️ Stream failed ({e}) — falling back to batch mode')
            response = self.client.chat(
                model    = self.model_name,
                messages = [
                    {'role': 'system', 'content': system},
                    {'role': 'user',   'content': user},
                ],
                options  = options,
            )
            return response['message']['content']

        result = ''.join(buffer)
        return result

    def translate(self, text, tier=None, prev_translation='', overlap_words=80, context_block=''):
        t = tier or self.tier
        if t == 'ADVANCED':
            return self._translate_two_pass(text, prev_translation, overlap_words, context_block)
        else:
            return self._translate_single_pass(text, t, prev_translation, overlap_words, context_block)

    def _translate_single_pass(self, text, tier, prev_translation, overlap_words, context_block):
        prompts    = TRANSLATION_PROMPTS[tier]
        scene_type = detect_scene_type(text)
        scene_ctx  = SCENE_CTX.get(scene_type, SCENE_CTX['DAILY_LIFE'])
        overlap    = get_overlap(prev_translation, overlap_words)
        ovlp_sec   = build_overlap_section(overlap)
        user = prompts['user'].format(
            context_block=context_block, scene_context=scene_ctx,
            overlap_section=ovlp_sec, chunk=text)
        for attempt in range(3):
            try:
                raw = self._call_model(prompts['system'], user, self._OPTS_STEP2)
                raw, pn = extract_plot_note(raw)
                translation = clean_translation(raw)
                if has_devanagari(translation) and attempt < 2:
                    print(f'   ⚠️ Devanagari attempt {attempt+1}, retry...'); continue
                return translation, scene_type, pn
            except Exception as e:
                if attempt == 2: raise
                print(f'   ⚠️ Attempt {attempt+1}: {e}')
        return '', scene_type, ''

    def _translate_two_pass(self, text, prev_translation, overlap_words, context_block):
        scene_type   = detect_scene_type(text)
        scene_ctx    = SCENE_CTX.get(scene_type, SCENE_CTX['DAILY_LIFE'])
        plot_note_en = ''

        # ── STEP 1: Semantic Base ─────────────────────────────────────────
        step1_user       = STEP1_USER.format(chunk=text)
        base_translation = ''
        for attempt in range(3):
            try:
                raw = self._call_model(STEP1_SYSTEM, step1_user, self._OPTS_STEP1, live_monitor=False)                raw = clean_translation(raw)
                if has_devanagari(raw) and attempt < 2:
                    print(f'   ⚠️ Step1 Devanagari attempt {attempt+1}'); continue
                base_translation = raw; break
            except Exception as e:
                if attempt == 2:
                    print(f'   ⚠️ Step1 failed: {e} — using English fallback')
                    base_translation = text
                else:
                    print(f'   ⚠️ Step1 attempt {attempt+1}: {e}')

        # ── STEP 2: Build system prompt ───────────────────────────────────
        tone_rules_section  = self._build_tone_rules_section()
        char_styles         = self._build_char_styles_section()
        author_dna          = STORY_STATE.get('author_dna', '')
        voice_guide_section = _HINGLISH_VOICE_GUIDE
        if author_dna:
            voice_guide_section += f'\n\nAUTHOR-SPECIFIC NOTE: {author_dna[:400]}'
        step2_system = STEP2_SYSTEM.format(
            hinglish_voice_guide = voice_guide_section,
            tone_rules_section   = tone_rules_section,
            char_styles          = char_styles,
        )

        # ─────────────────────────────────────────────────────────────────
        # ARCHITECTURAL FIX — No raw previous output in prompts.
        #
        # ROOT CAUSE of "15x repeated line" loops:
        #   old code fed last 80 words of prev translation via build_overlap_section()
        #   → model sees its own output → pattern detected → reinforced → amplified → loop
        #   LLMs self-condition: similar endings feed back, Hinglish (less stable token
        #   distribution) makes this worse than English-only output.
        #
        # Fix: overlap_section in STEP2_USER is now a neutral marker.
        #      Continuity comes ONLY from context_block (story summary from STORY_STATE).
        #      context_block = build_context_prompt() = summary, NOT raw output. ✓
        # ─────────────────────────────────────────────────────────────────
        _SUMMARY_ONLY_OVERLAP = (
            '(Continuity via story summary above — raw previous output intentionally removed '
            'to prevent self-conditioning loops)'
        )

        def _build_step2_user(base_text, stateless=False):
            """
            stateless=True  → hard reset (loop recovery): no story context at all.
            stateless=False → normal: story summary context, no raw prev output.
            """
            return STEP2_USER.format(
                context_block    = '(Hard reset — stateless generation)' if stateless else context_block,
                scene_context    = scene_ctx,
                few_shots        = _FEW_SHOTS,
                char_styles      = char_styles,
                overlap_section  = _SUMMARY_ONLY_OVERLAP,   # ← FIX: no raw output
                base_translation = base_text,
            )

        # ─────────────────────────────────────────────────────────────────
        # 3-STAGE CORRECTION ARCHITECTURE
        #
        # Stage 1 — Generation  : stateless (no raw prev output in context)
        # Stage 2 — Diagnose    : ALL signals checked BEFORE saving output
        # Stage 3 — Correction  :
        #   LOOP / REPETITION   → Retry with hard reset (no context, severity jump)
        #   ENGLISH-HEAVY only  → Soft correction pass (targeted, NOT full retry)
        #   CLEAN               → Accept immediately
        #
        # ══ BURN THIS IN ══════════════════════════════════════════════
        #   Retry = structural failure ONLY (loops, repetition, Devanagari)
        #   English ratio = stylistic imperfection = soft fix, NEVER full retry
        #
        #   Why? Full retry for style = gamble that risks re-introducing loops.
        #   Soft fix = targeted single call to reduce English leakage only.
        # ─────────────────────────────────────────────────────────────────

        _STEP2_RETRY_PROFILES = [
            # L1 — Soft: bump penalty, hold temp
            {**self._OPTS_STEP2, 'repeat_penalty': 1.25, 'num_predict': 1400, 'temperature': 0.60},
            # L2 — Aggressive: stronger penalty + shorter budget + cooler temp
            {**self._OPTS_STEP2, 'repeat_penalty': 1.35, 'num_predict': 1100, 'temperature': 0.55},
            # L3 — Nuclear: max penalty + shortest budget + lowest temp + input trim + top_p drop
            # FIX B: top_p 0.90→0.75 at L3 — reduces hallucination drift and breaks loop attractors
            {**self._OPTS_STEP2, 'repeat_penalty': 1.40, 'num_predict': 900,  'temperature': 0.50, 'top_k': 25, 'top_p': 0.75},
        ]

        # Soft English Fix — used for style correction only (not structural retry)
        _ENG_FIX_SYSTEM = (
            'Tum ek Hinglish editor ho. Tumhe ek draft diya gaya hai jisme English words '
            'zyada hain.\nKaam: Excessive English words ko natural Hindi/Hinglish mein replace karo.\n'
            'STRICT RULES:\n'
            '✓ Roman script ONLY — zero Devanagari\n'
            '✓ Sirf English leakage fix karo — natural English nouns mat badlo '
            '(phone, office, train, meeting, plan)\n'
            '✓ Sentence structure mat badlo\n'
            '✓ Meaning mat badlo\n'
            '✓ Koi line mat add/remove karo\n'
            '✓ Output: SIRF corrected Hinglish text — koi label, koi preamble nahi'
        )

        def _diagnose(translation, opts):
            """
            Unified diagnosis — all signals checked together.
            Returns (passed, eng_heavy_only, severity, reasons)

            passed         : True = clean, accept immediately
            eng_heavy_only : True = English ratio issue but NO structural failure
                             → soft fix, not retry
            severity       : 0=clean · 1=soft · 2=hard · 3=full_lock

            KEY RULES:
              line repetition + loop  → severity 3 (full lock, jump to L3)
              token_ceiling + loop    → severity 3 (confirmed saturation)
              English-heavy, no loop  → NOT a retry trigger (soft fix path)
            """
            if not translation:
                return False, False, 1, ['empty_output']

            twords   = translation.split()
            tok_ceil = opts['num_predict']
            reasons  = []
            severity = 0

            # ── Fast check first: line-level repetition ───────────────────
            is_rep, rep_count = detect_repetition(translation)
            if is_rep:
                reasons.append(f'LINE_REPEAT({rep_count})')
                severity = max(severity, 3)

            # ── Devanagari leak ───────────────────────────────────────────
            if has_devanagari(translation):
                reasons.append('Devanagari')
                severity = max(severity, 1)

            # ── English ratio (stylistic — does NOT drive severity) ───────
            eng_c = sum(1 for w in twords if re.match(r'^[a-z]{4,}$', w) and w not in _H_ALLOW)
            eng_ratio  = eng_c / max(len(twords), 1)
            eng_heavy  = eng_ratio > 0.55
            # Note: eng_heavy is returned separately — it does NOT increase severity
            # because high English ratio alone is a style issue, not structural failure.

            # ── n-gram loop detection (primary severity driver) ───────────
            is_loop, loop_sev = _detect_hard_loop(translation)
            if is_loop:
                reasons.append(f'NGRAM_LOOP(sev={loop_sev})')
                severity = max(severity, loop_sev)

            # ── Token ceiling (FIX C: char-based approximation, not word count) ──
            # Hinglish has long compound words → word count is unstable.
            # len(text)/4 approximates tokens more reliably for mixed-script output.
            near_ceil = len(translation) / 4 > tok_ceil
            if near_ceil:
                reasons.append(f'TOKEN_CEIL({len(twords)}w/{tok_ceil})')
                if is_loop or is_rep:
                    # Ceiling + any loop signal = saturation confirmed → full lock
                    severity = 3
                else:
                    severity = max(severity, 1)

            is_structural = (severity >= 1) or is_rep or is_loop
            # eng_heavy alone (no structural fail) = soft fix path, not retry
            eng_heavy_only = eng_heavy and not is_structural

            passed = (not reasons) and not eng_heavy
            return passed, eng_heavy_only, severity, reasons

        # ─────────────────────────────────────────────────────────────────
        # STAGE 1 — Initial generation (stateless, no raw prev output)
        # ─────────────────────────────────────────────────────────────────
        # ─────────────────────────────────────────────────────────────────
        # STAGE 1 — Initial generation (stateless, no raw prev output)
        # ─────────────────────────────────────────────────────────────────
        step2_user        = _build_step2_user(base_translation, stateless=False)
        final_translation = ''
        best_translation  = ''
        best_severity     = 99
        l0_action         = None
        l0_translation    = ''

        try:
            raw = self._call_model(step2_system, step2_user, self._OPTS_STEP2)
            raw, pn = extract_plot_note(raw)
            if pn: plot_note_en = pn
            translation = clean_translation(raw)

            # ── STAGE 2: Validate (all signals, no decisions) ─────────────
            v = _validate_output(translation, self._OPTS_STEP2, text)
            l0_action, l0_detail = _decide_action(v)
            l0_translation = translation

            if l0_action == 'accept':
                final_translation = translation
            else:
                best_translation = translation
                best_severity    = v['loop'][1] if v['loop'][0] else (3 if v['line_repeat'][0] else 1)
                print(f'   ⚠️ L0 [{l0_action.upper()} → {l0_detail}]')

        except Exception as e:
            l0_action  = 'retry'
            l0_detail  = 1
            print(f'   ⚠️ L0 error: {e}')

        # ─────────────────────────────────────────────────────────────────
        # STAGE 3 — Correction
        #
        # salvage_first → keep clean prefix + optional tail retry (LOOPS)
        # modify        → soft English fix (style only, never structural)
        # retry         → full chunk retry (non-loop: Devanagari, ceiling, empty)
        #
        # LOCKED RULE:
        #   Loop/repetition → salvage_first  (never throw away the clean 70%)
        #   Style issue     → modify         (never full retry for style)
        #   Structural      → retry          (only for non-loop failures)
        # ─────────────────────────────────────────────────────────────────
        if not final_translation:

            if l0_action == 'salvage_first':
                # ── SALVAGE-FIRST: keep clean prefix, retry only missing tail ──
                #
                # WHY: Your real output = first 70% clean + last 30% loop.
                # Retrying the full chunk throws away the clean 70%. Instead:
                #   1. Salvage the clean prefix (find loop start, cut there)
                #   2. Measure what % of output was clean (coverage ratio)
                #   3. Estimate which part of source wasn't covered
                #   4. Retry ONLY that tail (cheap, stateless, strict params)
                #   5. Join clean prefix + tail translation
                # This is always cheaper, usually better, never worse than full retry.
                print(f'   🔴 Loop (sev={l0_detail}) → Salvage-First strategy')
                salvaged = _salvage_looping_output(l0_translation)
                coverage = _get_loop_coverage(l0_translation, salvaged)
                print(f'   🪚 Salvage: {len(salvaged.split())}w kept ({coverage:.0%} of looped output)')

                TAIL_RETRY_THRESHOLD = 0.80  # if we kept <80%, the missing tail is worth recovering
                MIN_TAIL_SOURCE_WORDS = 25   # don't retry if tail source is trivially short

                if coverage < TAIL_RETRY_THRESHOLD and salvaged != l0_translation:
                    source_tail = _extract_source_tail(text, coverage)
                    tail_src_words = len(source_tail.split())

                    if tail_src_words >= MIN_TAIL_SOURCE_WORDS:
                        print(f'   🔁 Tail retry: {tail_src_words}w of uncovered source (stateless, L2)')
                        tail_opts = _STEP2_RETRY_PROFILES[1]  # L2 — aggressive but not nuclear

                        # Pass 1: base translation of tail only
                        try:
                            tail_base_raw = self._call_model(
                                STEP1_SYSTEM, STEP1_USER.format(chunk=source_tail), self._OPTS_STEP1)
                            tail_base = clean_translation(tail_base_raw)
                        except Exception:
                            tail_base = source_tail  # English fallback

                        # Pass 2: state-aware tail stylise
                        # NEW: pass clean prefix as context so model knows
                        # exactly where to continue from, plus explicit
                        # instruction to not repeat what's already written.
                        # This is "state-aware retry" — model sees what was
                        # already accepted and is told: continue from here,
                        # no repetition.
                        _no_repeat_instruction = (
                            f'\n\n[CONTINUITY NOTE: The following is already written — '
                            f'do NOT repeat any of it. Continue the story AFTER this point.]\n'
                            f'---ALREADY WRITTEN---\n'
                            f'{salvaged[-300:]}\n'   # last 300 chars of clean prefix
                            f'---END OF ALREADY WRITTEN---\n'
                            f'Now translate and continue ONLY the remaining part below.\n'
                        )
                        tail_user = _build_step2_user(
                            _no_repeat_instruction + tail_base,
                            stateless=False   # NOT stateless — we're giving it state explicitly
                        )
                        try:
                            tail_raw = self._call_model(step2_system, tail_user, tail_opts)
                            tail_raw, tail_pn = extract_plot_note(tail_raw)
                            if tail_pn and not plot_note_en:
                                plot_note_en = tail_pn
                            tail_tr = clean_translation(tail_raw)

                            v_tail = _validate_output(tail_tr, tail_opts)
                            act_tail, _ = _decide_action(v_tail)

                            if act_tail in ('accept', 'modify') and not v_tail['loop'][0]:
                                final_translation = salvaged + '\n\n' + tail_tr
                                print(f'   ✅ Salvage + state-aware tail: {len(final_translation.split())}w total')
                            else:
                                final_translation = salvaged
                                print(f'   ⚠️ Tail retry also looped — clean prefix only ({len(salvaged.split())}w)')
                        except Exception as e:
                            final_translation = salvaged
                            print(f'   ⚠️ Tail retry error: {e} — clean prefix only')
                    else:
                        final_translation = salvaged
                        print(f'   ✅ Tail too short ({tail_src_words}w) — clean prefix sufficient')
                else:
                    final_translation = salvaged
                    if salvaged == l0_translation:
                        print(f'   ℹ️  No loop boundary found — using original output')
                    else:
                        print(f'   ✅ Coverage {coverage:.0%} ≥ {TAIL_RETRY_THRESHOLD:.0%} — prefix sufficient')

            elif l0_action == 'modify':
                # ── STYLE FIX: English-heavy only — soft targeted correction ──
                print(f'   🟡 Style issue ({l0_detail}) → soft English fix (not full retry)')
                try:
                    fix_user = (
                        f'Yeh Hinglish draft hai jisme English zyada hai. '
                        f'Sirf English leakage fix karo — baaki sab waise hi rakhna.\n\n'
                        f'---\n{l0_translation}\n---'
                    )
                    fix_opts = {**self._OPTS_STEP2, 'temperature': 0.40,
                                'repeat_penalty': 1.20, 'num_predict': 1400}
                    raw2  = self._call_model(_ENG_FIX_SYSTEM, fix_user, fix_opts)
                    fixed = clean_translation(raw2)
                    v_fix = _validate_output(fixed, fix_opts)
                    act_fix, _ = _decide_action(v_fix)
                    if act_fix in ('accept', 'modify') and not v_fix['loop'][0]:
                        final_translation = fixed
                        print(f'   ✅ Soft fix accepted')
                    else:
                        final_translation = l0_translation
                        print(f'   ⚠️ Soft fix introduced issues — keeping L0')
                except Exception as e:
                    print(f'   ⚠️ Soft fix error: {e} — keeping L0')
                    final_translation = l0_translation

            elif l0_action == 'retry':
                # ── STRUCTURAL RETRY: non-loop failures (Devanagari, empty, ceiling) ──
                # These are NOT loops — no clean prefix to salvage.
                # Full chunk retry with severity-aware level jump.
                start_level = l0_detail if isinstance(l0_detail, int) else 1
                start_level = min(start_level, 3)
                retry_idx   = start_level - 1
                print(f'   🟠 Structural retry (non-loop, L{retry_idx + 1}+) — {l0_detail}')

                while retry_idx < 3 and not final_translation:
                    level = retry_idx + 1
                    opts  = _STEP2_RETRY_PROFILES[retry_idx]
                    step2_user = _build_step2_user(base_translation, stateless=(level >= 2))

                    if level == 3:
                        words_bt = base_translation.split()
                        trim_to  = max(50, int(len(words_bt) * 0.85))
                        step2_user = _build_step2_user(' '.join(words_bt[:trim_to]), stateless=True)
                        print(f'   🔪 L3: trim {len(words_bt)}w→{trim_to}w | rp=1.40 top_p=0.75')

                    try:
                        raw = self._call_model(step2_system, step2_user, opts)
                        raw, pn = extract_plot_note(raw)
                        if pn: plot_note_en = pn
                        translation = clean_translation(raw)
                        v   = _validate_output(translation, opts, text)
                        act, det = _decide_action(v)
                        sev = v['loop'][1] if v['loop'][0] else (3 if v['line_repeat'][0] else 0)

                        if sev < best_severity:
                            best_severity = sev; best_translation = translation

                        if act in ('accept', 'modify'):
                            final_translation = translation
                            print(f'   ✅ L{level}: {"clean" if act == "accept" else "style-ok"}')
                            break
                        elif act == 'salvage_first':
                            # Retry itself looped — salvage what we have and stop
                            salvaged = _salvage_looping_output(translation)
                            if len(salvaged.split()) >= len(translation.split()) * 0.35:
                                final_translation = salvaged
                                print(f'   🪚 L{level} looped → salvaged prefix ({len(salvaged.split())}w)')
                                break

                        next_idx = min(max(retry_idx + 1, sev - 1), 2)
                        if next_idx > retry_idx + 1:
                            skipped = ', '.join(f'L{n+1}' for n in range(retry_idx + 1, next_idx))
                            print(f'   ⚡ L{level}: sev={sev} → jumping to L{next_idx + 1} (skip {skipped})')
                        else:
                            print(f'   ⚠️ L{level}→L{level + 1}: {act}/{det}')
                        retry_idx = next_idx

                    except Exception as e:
                        print(f'   ⚠️ L{level} error: {e}')
                        retry_idx += 1

                if retry_idx >= 3 and not final_translation:
                    print(f'   ⚠️ L3 exhausted — accepting best available (sev={best_severity})')

        # ── Fallback ───────────────────────────────────────────────────────
        if not final_translation:
            final_translation = best_translation or base_translation
            if not best_translation:
                print(f'   ⚠️ All stages failed — falling back to Step1 base')

        return final_translation, scene_type, plot_note_en



# ═══════════════════════════════════════════════════════════════════
# TRANSLATION GENERATOR v10.0
# ═══════════════════════════════════════════════════════════════════
class OllamaTranslationGenerator:
    def __init__(self, model_name, target_lang, output_dir='.',
                 tier='ADVANCED', chunk_size=350, overlap_words=80,
                 num_ctx=8192, tone_rules=None, session_budget=None):
        self.model_name    = model_name
        self.target_lang   = target_lang
        self.output_dir    = Path(output_dir)
        self.tier          = tier
        self.chunk_size    = chunk_size
        self.overlap_words = overlap_words
        self.session_budget = session_budget  # seconds
        self.output_dir.mkdir(parents=True, exist_ok=True)
        self.engine = OllamaTranslationEngine(
            model_name, target_lang, tier, num_ctx, tone_rules=tone_rules
        )

    def translate_file(self, input_file, resume_from_chunk=0):
        arch = '2-Pass' if self.tier == 'ADVANCED' else 'Single-Pass'
        _jarvis_header(
            f'Hindi→Hinglish REFORMULATION v1.0 — {arch} | AuthorDNA | ToneGuard Pro',
            f'Model: {self.model_name} | Mode: 1-Pass Reformulation | Chunk: {self.chunk_size}w | Session Guard: {self.session_budget//60 if self.session_budget else "OFF"}min')
        with open(input_file, 'r', encoding='utf-8') as f:
            text = f.read()
        text = _preprocess_hindi_source(text)
        text = clean_source_text(text)
        orig_chars = len(text)
        _jarvis_info(f'📊 Input: {len(text.split()):,} words | {orig_chars:,} chars')
        chunks = chunk_text(text, self.chunk_size)
        STORY_STATE['total_chunks'] = len(chunks)
        est_lo = len(chunks) * 90 // 60
        est_hi = len(chunks) * 120 // 60
        _jarvis_ok(f'📦 {len(chunks)} chunks ({self.chunk_size}w) | Est: {est_lo}–{est_hi} min | Resume from: {resume_from_chunk}')

        # Author DNA notification
        if STORY_STATE.get('author_dna'):
            _jarvis_ok(f'🧬 AuthorDNA: {STORY_STATE["author"]} | Tone: {STORY_STATE["tone_anchor"]}')

        translations = []
        deva_flags   = []
        start_time   = time.time()
        session_start = start_time
        _ctimes, _errs, _ewma = [], 0, None
        _ALPHA = 0.25
        qa_report  = []
        timestamp  = datetime.now().strftime('%Y%m%d_%H%M%S')
        lang_code  = self.target_lang.split('_')[0]
        state_file = self.output_dir / f'state_{lang_code}_{timestamp}.json'
        out_path   = self.output_dir / f'translation_{lang_code}_{timestamp}.txt'

        # Background writer — no translation blocking
        bg_writer = BackgroundWriter(str(out_path))

        for i, chunk in enumerate(chunks, 1):
            # Skip already-translated chunks (resume)
            if i <= resume_from_chunk:
                _jarvis_info(f'[RESUME] Skipping chunk {i} (already done)')
                continue

            # T4 Session Guard — warn if < 30 min remains
            session_elapsed = time.time() - session_start
            if self.session_budget:
                session_remaining = self.session_budget - session_elapsed
                if session_remaining < 600:  # < 10 min
                    _jarvis_warn(f'⚠️ SESSION GUARD: Only {int(session_remaining//60)}m remaining — stopping to preserve output!')
                    _jarvis_warn('Download output now before session expires!')
                    break
                elif session_remaining < 1800 and _ewma and _ewma > session_remaining / max(len(chunks) - i, 1):
                    _jarvis_warn(f'⚠️ Session Guard: ETA exceeds remaining time. Auto-shrinking chunk size.')
                    self.chunk_size = max(150, self.chunk_size - 50)

            chunk_start = time.time()
            prev_trans  = translations[-1] if translations else ''
            prev_tail   = get_overlap(prev_trans, self.overlap_words) if prev_trans else ''
            context_block = build_context_prompt()
            context_chars = len(context_block)

            try:
                translated, scene_type, plot_note_en = self.engine.translate(
                    chunk, self.tier, prev_trans, self.overlap_words, context_block)
                translated = clean_translation(translated)
                issues    = validate_translation(translated, prev_tail, i, chunk)
                qa_sum, all_ok = summarise_issues(issues)
                qa_report.append({'chunk': i, 'scene': scene_type, 'issues': issues})
                deva_flags.append(not issues['devanagari'][0])
                translations.append(translated)

                # Background write — non-blocking
                bg_writer.write_chunk(translated)
                update_story_state(chunk, translated, i, scene_type, plot_note_en=plot_note_en)  # source_chunk, output_chunk
                # State save also background
                t_save = threading.Thread(target=save_state, args=(str(state_file),), daemon=True)
                t_save.start()

                chunk_time = time.time() - chunk_start
                _ctimes.append(chunk_time)
                _ewma = chunk_time if _ewma is None else _ALPHA*chunk_time + (1-_ALPHA)*_ewma
                _eta  = _ewma * (len(chunks) - i)
                total_e = time.time() - start_time
                gpu_u, vr_u, vr_t, gpu_t = _j_gpu()
                qa_txt = '' if all_ok else qa_sum.replace('\n',' | ')[:120]
                clear_output(wait=True)
                _jarvis_chunk_dashboard(
                    i, len(chunks), scene_type,
                    len(chunk.split()), len(chunk), len(translated),
                    chunk_time, total_e, _eta, _ctimes, _errs,
                    gpu_u, vr_u, vr_t, gpu_t,
                    self.model_name, self.target_lang, self.tier, all_ok, qa_txt, context_chars,
                    pass_label='2-Pass' if self.tier=='ADVANCED' else '1-Pass',
                    session_budget=self.session_budget,
                    session_elapsed=session_elapsed)

            except Exception as e:
                _errs += 1
                chunk_time = time.time() - chunk_start
                _ctimes.append(chunk_time)
                _ewma = chunk_time if _ewma is None else _ALPHA*chunk_time + (1-_ALPHA)*_ewma
                _eta  = _ewma * (len(chunks) - i)
                total_e = time.time() - start_time
                gpu_u, vr_u, vr_t, gpu_t = _j_gpu()
                err_text = f'[ERROR chunk {i}: {e}]'
                translations.append(err_text)
                bg_writer.write_chunk(err_text)
                deva_flags.append(False)
                qa_report.append({'chunk':i,'scene':'ERROR','issues':{'error':(False,str(e))}})
                clear_output(wait=True)
                _jarvis_chunk_dashboard(
                    i, len(chunks), 'ERROR',
                    len(chunk.split()), len(chunk), 0,
                    chunk_time, total_e, _eta, _ctimes, _errs,
                    gpu_u, vr_u, vr_t, gpu_t,
                    self.model_name, self.target_lang, self.tier, False, f'ERROR: {e}', 0,
                    session_budget=self.session_budget,
                    session_elapsed=time.time()-session_start)

        # Finalize
        bg_writer.finalize()
        total_time  = time.time() - start_time
        trans_chars = sum(len(t) for t in translations)
        deva_count  = sum(deva_flags)
        qa_report_file = self.output_dir / f'qa_{lang_code}_{timestamp}.txt'
        qa_failed_chunks = []
        with open(qa_report_file, 'w', encoding='utf-8') as qf:
            qf.write(f'QA REPORT v10.0 — {timestamp}\nModel: {self.model_name}\n'+'='*60+'\n\n')
            for entry in qa_report:
                iss    = entry['issues']
                failed = [k for k,(p,_) in iss.items() if not p]
                status = 'PASS' if not failed else ('FAIL ['+ ', '.join(failed) +']')
                qf.write(f'Chunk {entry["chunk"]} ({entry["scene"]}): {status}\n')
                if failed:
                    qa_failed_chunks.append(entry['chunk'])
                    for k in failed: qf.write(f'  -> {k}: {iss[k][1]}\n')
            qf.write(f'\nSUMMARY: {len(qa_failed_chunks)}/{len(chunks)} chunks had issues\n')

        _jarvis_complete(self.model_name, total_time, len(translations), orig_chars,
                         trans_chars, deva_count, len(qa_failed_chunks),
                         out_path, str(state_file))
        self.last_state_file = str(state_file)
        self.last_qa_file    = str(qa_report_file)
        return str(out_path), str(state_file)


_jarvis_ok('Hindi→Hinglish Engine v1.0 loaded — 1-Pass Reformulation | ToneGuard Pro | Anti-Hallucination')
_jarvis_ok('Step1: Hindi→Hinglish reformulation (temp=0.15) | Step2: Hinglish polish (temp=0.60)')
_jarvis_ok('16 Genre Few-Shots: Literary · Satire · Detective · Tension · Romance · Action · Tragedy · Horror · Philosophy · Monologue · Confrontation · Folk · SciFi · Coming-of-Age · War · Description')
_jarvis_info('Zero-Fabrication Rule: Output length ratio check | No invented content | Foul Word Filter')
_jarvis_info('T4 Session Guard: Timeout warning | Adaptive chunk shrink | Background file I/O')
_jarvis_info('Context v6.0: rolling 5-chunk summary | capped at 1200 chars')
````

## 📖 Appendix — Hinglish Voice Reference

This is the researched tone guide that JARVIS uses internally. Read this to understand what 'right Hinglish' sounds like for this project.

### The Target Voice
**Like an educated Indian in their late 20s/early 30s narrating a story to a friend.**  
Reference: *Made in Heaven*, *Little Things*, *Scam 1992* — clear, warm, intelligent.

### What This Sounds Like
- Mostly Hindi/Urdu vocabulary, English nouns where they're actually more natural
- Sentences flow like speech — not textbook Hindi, not pidgin English
- Emotions felt fully — expressed with restraint, not melodrama
- Formal characters speak with dignity — not casually addressed

### Address Forms
| Context | Form |
|---------|------|
| Formal / respectful | `aap / aapko / aapka` |
| Familiar / equal | `tum / tumhe / tumhara / tumne` |
| NEVER in this project | `tu / tujhe / tera / tune` |

### Never Use
- `yaar` or `bhai` as address terms for characters
- `gadha`, `bewakoof`, `ullu` (abusive tone)
- English Gen-Z slang (`no cap`, `based`, `lowkey`)
- Forced exclamations (`!!!`)
- Narrator commentary not in the source

### Author-Specific Notes

| Author | Key Hinglish Principle |
|--------|------------------------|
| **Kafka** | Horror = flatness. Describe transformation like a tax form. |
| **Austen** | Irony lives in understatement. The humor is never announced. |
| **Doyle** | Holmes is always formal. His Hinglish = composed professional. |
| **Tagore** | Lyricism must survive. Sentences breathe. No flippancy. |
| **Stoker** | Dread builds through cold description, not exclamation. |
| **Dostoevsky** | Inner monologue spirals. Characters contradict themselves. |
| **Tolstoy** | Gravitas maintained. No casualness in serious moments. |
| **Chekhov** | Silence between lines matters. Understate everything. |
