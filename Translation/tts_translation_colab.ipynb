## TTS Punctuation Reference:
| Mark | Effect in TTS |
|------|---------------|
| `?` | Rising intonation (questions) |
| `???` | Strong doubt/disbelief |
| `...` | Pause, hesitation |

| `-` | Sudden interruption |

| CAPS | Word emphasis |

## Step 4: Define Translation Functions

````python
import ollama
import re
from IPython.display import display, Markdown, clear_output

def clean_translation(text):
    """Clean up translation artifacts."""
    # Remove thinking markers
    text = re.sub(r'<think>.*?</think>', '', text, flags=re.DOTALL)
    # Remove markdown code blocks
    text = re.sub(r'```\w*\n?', '', text)
    # Remove translation markers
    text = re.sub(r'(Translation:|Hindi Translation:|Here\'s the translation:)', '', text, flags=re.IGNORECASE)
    # Clean excessive whitespace
    lines = [line.strip() for line in text.split('\n')]
    text = '\n\n'.join(line for line in lines if line)
    return text.strip()

def translate_text(english_text, tier="ADVANCED", model=MODEL_NAME, temperature=0.4, stream=True):
    """
    Translate English text to Hindi with TTS-optimized punctuation.
    
    Args:
        english_text: The English text to translate
        tier: Translation tier (BASIC, INTERMEDIATE, ADVANCED)
        model: Ollama model to use
        temperature: Creativity level (0.0-1.0)
        stream: Whether to stream the output
    
    Returns:
        Translated Hindi text with TTS punctuation
    """
    prompts = TRANSLATION_PROMPTS[tier]
    user_prompt = prompts["user"].format(text=english_text)
    
    print(f"🤖 Translating with {model} (Tier: {tier})...")
    print(f"📝 Input: {len(english_text)} characters")
    print("─" * 50)
    
    if stream:
        # Streaming mode
        full_response = ""
        stream_response = ollama.chat(
            model=model,
            messages=[
                {"role": "system", "content": prompts["system"]},
                {"role": "user", "content": user_prompt}
            ],
            options={
                "temperature": temperature,
                "top_p": 0.75,
                "num_ctx": 8192
            },
            stream=True
        )
        
        for chunk in stream_response:
            if 'message' in chunk and 'content' in chunk['message']:
                token = chunk['message']['content']
                full_response += token
                print(token, end='', flush=True)
        
        print("\n" + "─" * 50)
        translated = clean_translation(full_response)
    else:
        # Non-streaming mode
        response = ollama.chat(
            model=model,
            messages=[
                {"role": "system", "content": prompts["system"]},
                {"role": "user", "content": user_prompt}
            ],
            options={
                "temperature": temperature,
                "top_p": 0.75,
                "num_ctx": 8192
            }
        )
        translated = clean_translation(response['message']['content'])
    
    print(f"\n✅ Translation complete!")
    print(f"📊 Output: {len(translated)} characters")
    print(f"📈 Ratio: {len(translated)/len(english_text):.2f}x")
    
    return translated

print("✅ Translation functions loaded!")
````
